# Model Prediction vs Actual Plots in Matplotlib

## Diagnostics for Regression
- Scatter of Actual ($y$) vs Predicted ($\hat{y}$) along with perfect prediction reference line ($y = \hat{y}$).
- Computing and displaying metrics directly on the plot ($R^2$, RMSE, MAE).

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
y_actual = np.random.uniform(50, 450, 200)
# Predicted with some noise and slight underprediction at high values
y_pred = y_actual * 0.95 + np.random.normal(0, 25, 200)

rmse = np.sqrt(np.mean((y_actual - y_pred)**2))
r2 = 1 - (np.sum((y_actual - y_pred)**2) / np.sum((y_actual - np.mean(y_actual))**2))

fig, ax = plt.subplots(figsize=(7, 7))

# Scatter with error magnitude encoded by color
errors = np.abs(y_actual - y_pred)
scatter = ax.scatter(y_actual, y_pred, c=errors, cmap='viridis', alpha=0.7, edgecolors='none', s=45)
cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label('Absolute Error |y - ŷ|')

# Perfect prediction identity line
min_val = min(y_actual.min(), y_pred.min()) - 10
max_val = max(y_actual.max(), y_pred.max()) + 10
ax.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2.5, label='Perfect Fit (y = ŷ)')

ax.set_xlim(min_val, max_val)
ax.set_ylim(min_val, max_val)
ax.set_xlabel('Actual Values (y)', fontweight='bold')
ax.set_ylabel('Predicted Values (ŷ)', fontweight='bold')
ax.set_title('Predicted vs Actual House Prices ($k)', fontsize=13, fontweight='bold')
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3)

# Display metrics box
stats_text = f"$R^2$: {r2:.3f}\nRMSE: {rmse:.2f}"
ax.text(0.7, 0.15, stats_text, transform=ax.transAxes,
        fontsize=11, verticalalignment='top',
        bbox=dict(boxstyle='round,pad=0.5', facecolor='white', alpha=0.8, edgecolor='gray'))

plt.tight_layout()
plt.show()
```
