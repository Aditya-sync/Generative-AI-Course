# Multiple Axes / Twin Axes in Matplotlib

## Use Cases
- Plotting two metrics with completely different units and scales on the same X-axis (e.g., Training Loss and Learning Rate, or Accuracy and Epoch Time).
- Functions: `ax.twinx()` (shares X, independent Y) and `ax.twiny()` (shares Y, independent X).

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. ax.twinx() for Loss vs Learning Rate
epochs = np.arange(1, 31)
loss = 2.5 * np.exp(-epochs / 8) + 0.1 * np.random.randn(30) * 0.2
lr = 1e-3 * (0.95 ** epochs)

fig, ax1 = plt.subplots(figsize=(9, 5))

color = 'tab:red'
ax1.set_xlabel('Epoch', fontweight='bold')
ax1.set_ylabel('Cross-Entropy Loss', color=color, fontweight='bold')
line1 = ax1.plot(epochs, loss, color=color, marker='o', lw=2, label='Training Loss')
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(True, alpha=0.3)

# Instantiate twin axes sharing X
ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Learning Rate (Cosine Schedule)', color=color, fontweight='bold')
line2 = ax2.plot(epochs, lr, color=color, linestyle='--', lw=2, label='Learning Rate')
ax2.tick_params(axis='y', labelcolor=color)

# Combine legends from both axes
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper center')

plt.title('Training Loss & Learning Rate Schedule on Twin Axes', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 2. Inset Axes (Zoom-in on critical region)
from mpl_toolkits.axes_grid1.inset_locator import inset_axes, mark_inset

x = np.linspace(0, 10, 1000)
y = np.sin(x) * np.exp(-x / 3) + np.sin(5 * x) * 0.1

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, y, color='navy')
ax.set_title('Main Signal with High-Frequency Zoom Inset', fontweight='bold')

# Create inset axes [x0, y0, width, height] in normalized coordinates
axins = inset_axes(ax, width="35%", height="35%", loc='upper right')
axins.plot(x, y, color='crimson')
axins.set_xlim(2.5, 4.0)
axins.set_ylim(-0.5, 0.5)
axins.grid(True, alpha=0.3)
axins.set_title('Zoomed (2.5 - 4.0s)', fontsize=9)

mark_inset(ax, axins, loc1=2, loc2=4, fc="none", ec="0.5")
plt.show()
```
