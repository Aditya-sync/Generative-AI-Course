# Colormaps in Matplotlib

## Colormap Categories
- **Sequential**: Uniform progression of lightness (e.g. `viridis`, `plasma`, `inferno`, `Blues`). Ideal for magnitudes.
- **Diverging**: Two contrasting hues meeting at a neutral midpoint (e.g. `coolwarm`, `seismic`, `bwr`). Ideal for residuals, correlations, zero-centered weights.
- **Qualitative**: Distinct non-ordered colors (e.g. `tab10`, `Set2`). Ideal for discrete classes.

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Visualizing Different Types of Colormaps
cmaps = {
    'Perceptually Uniform': ['viridis', 'plasma', 'inferno', 'cividis'],
    'Sequential': ['Blues', 'Greens', 'YlOrRd'],
    'Diverging': ['coolwarm', 'RdBu_r', 'seismic'],
    'Qualitative': ['tab10', 'Set2', 'Pastel1']
}

nrows = sum(len(v) for v in cmaps.values())
fig, axes = plt.subplots(nrows=nrows, figsize=(8, 8))
fig.subplots_adjust(top=0.95, bottom=0.01, left=0.2, right=0.99)

gradient = np.linspace(0, 1, 256).reshape(1, -1)
idx = 0
for cat, names in cmaps.items():
    for name in names:
        axes[idx].imshow(gradient, aspect='auto', cmap=plt.get_cmap(name))
        axes[idx].set_title(f"{cat}: '{name}'", loc='left', fontsize=10)
        axes[idx].set_axis_off()
        idx += 1

plt.tight_layout()
plt.show()
```

```python
# 2. Creating Custom Colormaps using LinearSegmentedColormap
from matplotlib.colors import LinearSegmentedColormap

colors = ['#1a2a6c', '#b21f1f', '#fdbb2d']  # Blue to Red to Gold
custom_cmap = LinearSegmentedColormap.from_list('aurora', colors, N=256)

data = np.random.randn(20, 20)
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(data, cmap=custom_cmap)
plt.colorbar(im, ax=ax)
ax.set_title('Custom LinearSegmentedColormap', fontweight='bold')
plt.show()
```
