# ROC & Precision-Recall Curves in Matplotlib

## When to use which:
- **ROC Curve (TPR vs FPR)**: Balanced datasets; compares models against random baseline (diagonal $y=x$).
- **PR Curve (Precision vs Recall)**: Severe class imbalance (fraud detection, rare disease diagnostics).

```python
import matplotlib.pyplot as plt
import numpy as np

# Simulated ROC curve data
fpr = np.linspace(0, 1, 100)
tpr_xgboost = 1 - (1 - fpr)**3.5
tpr_rf = 1 - (1 - fpr)**2.5
tpr_logistic = 1 - (1 - fpr)**1.8

# Simulated PR curve data
recall = np.linspace(0, 1, 100)
prec_xgboost = 0.95 - 0.45 * (recall**2)
prec_rf = 0.90 - 0.55 * (recall**2)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

# 1. ROC Curve
ax1.plot(fpr, tpr_xgboost, color='#1f77b4', lw=2.5, label='XGBoost (AUC = 0.94)')
ax1.plot(fpr, tpr_rf, color='#2ca02c', lw=2, label='Random Forest (AUC = 0.88)')
ax1.plot(fpr, tpr_logistic, color='#ff7f0e', lw=2, label='Logistic Reg (AUC = 0.81)')
ax1.plot([0, 1], [0, 1], color='gray', lw=1.5, linestyle='--', label='Random Chance (AUC = 0.50)')

ax1.set_xlim([0.0, 1.0])
ax1.set_ylim([0.0, 1.05])
ax1.set_xlabel('False Positive Rate (1 - Specificity)', fontweight='bold')
ax1.set_ylabel('True Positive Rate (Sensitivity)', fontweight='bold')
ax1.set_title('Receiver Operating Characteristic (ROC)', fontweight='bold')
ax1.legend(loc='lower right')
ax1.grid(True, alpha=0.3)

# 2. Precision-Recall Curve
ax2.plot(recall, prec_xgboost, color='#1f77b4', lw=2.5, label='XGBoost (PR-AUC = 0.82)')
ax2.plot(recall, prec_rf, color='#2ca02c', lw=2, label='Random Forest (PR-AUC = 0.74)')
ax2.axhline(0.1, color='gray', linestyle='--', label='Baseline Pos Rate (10%)')

ax2.set_xlim([0.0, 1.0])
ax2.set_ylim([0.0, 1.05])
ax2.set_xlabel('Recall (Sensitivity)', fontweight='bold')
ax2.set_ylabel('Precision (Positive Predictive Value)', fontweight='bold')
ax2.set_title('Precision-Recall Curve (Imbalanced Scenario)', fontweight='bold')
ax2.legend(loc='lower left')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
