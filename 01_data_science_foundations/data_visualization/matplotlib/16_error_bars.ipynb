# Error Bars in Matplotlib

## Overview
Error bars quantify uncertainty, confidence intervals, or standard deviations in measurements and ML benchmark scores.

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Continuous Error Bars (errorbar & fill_between)
x = np.linspace(0, 10, 30)
y = np.sin(x)
yerr = 0.15 + 0.1 * np.random.rand(len(x))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.errorbar(x, y, yerr=yerr, fmt='-o', color='#1f77b4',
             ecolor='gray', elinewidth=1.5, capsize=3, capthick=1.5, label='Signal ± Error')
ax1.set_title('Point-wise Error Bars (capsize & ecolor)', fontweight='bold')
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Amplitude')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Shaded Confidence Interval (Common in RL & Deep Learning training curves)
x_dense = np.linspace(0, 100, 200)
mean_reward = 100 * (1 - np.exp(-x_dense / 25))
std_reward = 8 * np.sin(x_dense / 10) + 12

ax2.plot(x_dense, mean_reward, color='#2ca02c', label='Mean Episode Reward')
ax2.fill_between(x_dense, mean_reward - std_reward, mean_reward + std_reward,
                 color='#2ca02c', alpha=0.2, label='±1 Std Dev')
ax2.set_title('Confidence Band via fill_between (RL Style)', fontweight='bold')
ax2.set_xlabel('Episodes')
ax2.set_ylabel('Reward')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# 2. Bar Chart with Asymmetric Error Bars
models = ['BERT', 'RoBERTa', 'DeBERTa', 'GPT-2', 'LLaMA']
f1_scores = [88.4, 90.1, 91.8, 85.3, 93.4]
# Asymmetric error: [lower_errors, upper_errors]
lower_err = [1.2, 0.8, 0.9, 2.1, 0.5]
upper_err = [0.8, 1.1, 0.7, 1.5, 0.4]
asym_err = [lower_err, upper_err]

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(models, f1_scores, yerr=asym_err, capsize=5, color='#3498db', edgecolor='navy', alpha=0.85)
ax.set_ylim(80, 100)
ax.set_ylabel('F1 Score (%)')
ax.set_title('Cross-Validation F1 Scores with Asymmetric Bounds', fontweight='bold')
ax.grid(axis='y', alpha=0.3)

for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval - 3, f'{yval:.1f}%', ha='center', va='bottom', color='white', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Edge case: handling NaNs or infs in error bars
x_err = [1, 2, 3]
y_err = [10, 20, 30]
err_with_nan = [1.0, np.nan, 2.0]

fig, ax = plt.subplots()
ax.errorbar(x_err, y_err, yerr=err_with_nan, fmt='o')
ax.set_title('Handling NaN error values (omits error bar for that point)')
plt.show()
```
