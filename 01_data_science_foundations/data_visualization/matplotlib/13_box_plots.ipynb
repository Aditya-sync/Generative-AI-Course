# Box Plots

## Definition
**Box plots** (box-and-whisker plots) display the **5-number summary** of a distribution:
minimum, Q1, median (Q2), Q3, and maximum — plus outliers.

### Anatomy of a Box Plot
```
        ┌───────┐
  ──────┤  IQR  ├──────  whisker extends to 1.5×IQR
        └───────┘
  min   Q1  Q2  Q3  max    (+ outlier points beyond whiskers)
```

### Key Parameters of `ax.boxplot()`
| Parameter | Description |
|-----------|-------------|
| `notch` | Draw notched box (confidence interval around median) |
| `vert` | Vertical (True) or horizontal (False) |
| `patch_artist` | Fill boxes with color |
| `widths` | Box width |
| `showfliers` | Show/hide outlier points |
| `flierprops` | Customize outlier markers |
| `medianprops` | Customize median line |
| `boxprops` | Customize box border |

````python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)

data_groups = {
    'Group A': rng.normal(100, 15, 200),
    'Group B': rng.normal(110, 25, 200),
    'Group C': rng.normal(95, 10, 200),
    'Group D': rng.normal(120, 30, 200)
}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Basic boxplot
bp = axes[0].boxplot(list(data_groups.values()),
                     labels=list(data_groups.keys()),
                     patch_artist=True, notch=False)

colors = ['steelblue', 'tomato', 'seagreen', 'darkorange']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

axes[0].set_title('Box Plot — Group Comparison', fontsize=13)
axes[0].set_ylabel('Value')
axes[0].grid(True, axis='y', alpha=0.3)
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)

# Horizontal boxplot
bp2 = axes[1].boxplot(list(data_groups.values()),
                      labels=list(data_groups.keys()),
                      patch_artist=True, vert=False, notch=True)
for patch, color in zip(bp2['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

axes[1].set_title('Horizontal Notched Box Plot', fontsize=13)
axes[1].set_xlabel('Value')
axes[1].grid(True, axis='x', alpha=0.3)
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
````

````python
# Styled boxplot with custom properties
fig, ax = plt.subplots(figsize=(10, 6))

data_all = [rng.normal(mu, sigma, 150)
            for mu, sigma in [(100,15),(110,25),(95,10),(120,30),(105,20)]]
labels = ['Model A', 'Model B', 'Model C', 'Model D', 'Model E']

bp = ax.boxplot(data_all, labels=labels, patch_artist=True,
    medianprops=dict(color='white', linewidth=2.5),
    flierprops=dict(marker='o', markerfacecolor='gray', markersize=5, alpha=0.5),
    whiskerprops=dict(linewidth=1.5, linestyle='--'),
    capprops=dict(linewidth=2))

palette = ['#2196F3', '#F44336', '#4CAF50', '#FF9800', '#9C27B0']
for patch, color in zip(bp['boxes'], palette):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)

ax.set_title('Model Score Distribution Comparison', fontsize=14)
ax.set_ylabel('Score')
ax.grid(True, axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
````

````python
# Box plot with overlaid data points (jittered)
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
n = 80
data_two = [rng.normal(0, 1, n), rng.normal(1, 1.5, n)]

# Standard box plot
bp = axes[0].boxplot(data_two, patch_artist=True, labels=['A', 'B'])
for patch, color in zip(bp['boxes'], ['steelblue', 'tomato']):
    patch.set_facecolor(color); patch.set_alpha(0.5)
axes[0].set_title('Standard Box Plot')

# Box plot + jittered points
for i, (d, color) in enumerate(zip(data_two, ['steelblue', 'tomato']), 1):
    jitter = rng.uniform(-0.15, 0.15, len(d))
    axes[1].scatter(np.full(len(d), i) + jitter, d, alpha=0.5,
                    s=15, color=color, zorder=3)

bp2 = axes[1].boxplot(data_two, patch_artist=True, labels=['A', 'B'],
                       showfliers=False)
for patch, color in zip(bp2['boxes'], ['steelblue', 'tomato']):
    patch.set_facecolor(color); patch.set_alpha(0.2)

axes[1].set_title('Box Plot + Jittered Points')

for ax in axes:
    ax.grid(True, axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
````
