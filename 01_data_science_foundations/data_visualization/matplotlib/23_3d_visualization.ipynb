# 3D Visualization with Matplotlib

## Overview
Using `projection='3d'` via `mpl_toolkits.mplot3d` for:
- 3D Scatter plots (clusters, latent embeddings)
- 3D Surface plots (`plot_surface`)
- 3D Wireframes (`plot_wireframe`)

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# 1. 3D Scatter Plot for Multi-Class Clusters (PCA/t-SNE latent space)
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

np.random.seed(42)
for cluster_id, color, label in [(0, 'blue', 'Class 0'), (1, 'green', 'Class 1'), (2, 'red', 'Class 2')]:
    center = np.random.randn(3) * 3
    pts = center + np.random.randn(100, 3) * 0.8
    ax.scatter(pts[:, 0], pts[:, 1], pts[:, 2], c=color, label=label, alpha=0.6, s=30, edgecolors='w')

ax.set_xlabel('Principal Component 1', fontweight='bold')
ax.set_ylabel('Principal Component 2', fontweight='bold')
ax.set_zlabel('Principal Component 3', fontweight='bold')
ax.set_title('3D Latent Embedding Scatter Plot', fontsize=14, fontweight='bold')
ax.legend()
ax.view_init(elev=25, azim=45)
plt.show()
```

```python
# 2. 3D Surface Plot of Optimization Loss Function
fig = plt.figure(figsize=(11, 7))
ax = fig.add_subplot(111, projection='3d')

X = np.linspace(-4, 4, 100)
Y = np.linspace(-4, 4, 100)
X, Y = np.meshgrid(X, Y)
# Ackley-like surface
Z = -20 * np.exp(-0.2 * np.sqrt(0.5 * (X**2 + Y**2))) - np.exp(0.5 * (np.cos(2*np.pi*X) + np.cos(2*np.pi*Y))) + np.e + 20

surf = ax.plot_surface(X, Y, Z, cmap='plasma', edgecolor='none', alpha=0.9, antialiased=True)
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, label='Loss Magnitude')

ax.set_xlabel('w1')
ax.set_ylabel('w2')
ax.set_zlabel('Loss(w1, w2)')
ax.set_title('3D Non-Convex Loss Optimization Landscape', fontweight='bold')
ax.view_init(elev=40, azim=120)
plt.show()
```
