# Distribution Visualization in Matplotlib for AI/ML

## Why it matters in ML
- Checking normality / skewness before feeding features to linear models.
- Identifying multimodality (indicating distinct subgroups or latent clusters).
- Comparing training vs test / production distribution shifts (covariate shift).

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
# Simulating features: Normal, Right-Skewed (Log-normal), Bimodal
normal_feat = np.random.normal(loc=50, scale=10, size=1500)
skewed_feat = np.random.exponential(scale=15, size=1500)
bimodal_feat = np.concatenate([np.random.normal(20, 5, 800), np.random.normal(60, 8, 700)])

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Subplot 1: Normal with KDE overlay simulation
ax = axes[0]
counts, bins, _ = ax.hist(normal_feat, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.7)
bin_centers = 0.5 * (bins[1:] + bins[:-1])
pdf = (1 / (10 * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((bin_centers - 50) / 10)**2)
ax.plot(bin_centers, pdf, 'r-', lw=2.5, label='Theoretical PDF')
ax.axvline(np.mean(normal_feat), color='darkred', linestyle='--', label=f'Mean: {np.mean(normal_feat):.1f}')
ax.set_title('Gaussian Distributed Feature', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

# Subplot 2: Skewed feature with Mean vs Median
ax = axes[1]
ax.hist(skewed_feat, bins=40, color='coral', edgecolor='black', alpha=0.7)
ax.axvline(np.mean(skewed_feat), color='blue', linestyle='--', lw=2, label=f'Mean: {np.mean(skewed_feat):.1f}')
ax.axvline(np.median(skewed_feat), color='green', linestyle='-', lw=2, label=f'Median: {np.median(skewed_feat):.1f}')
ax.set_title('Right-Skewed Feature (Mean > Median)', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

# Subplot 3: Covariate Shift (Train vs Test)
ax = axes[2]
train_dist = np.random.normal(30, 8, 1000)
test_dist = np.random.normal(38, 9, 1000)
ax.hist(train_dist, bins=30, alpha=0.6, label='Train Set', color='teal', density=True)
ax.hist(test_dist, bins=30, alpha=0.6, label='Production Set (Shifted)', color='purple', density=True)
ax.set_title('Dataset Covariate Shift Detection', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
