# Residual Plots in Matplotlib

## Core Diagnostic in Regression
- **Homoscedasticity vs Heteroscedasticity**: Residuals should show constant variance and zero mean across all predicted values.
- **Q-Q Plot simulation**: Checking normality of residuals.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
y_pred = np.linspace(10, 100, 300)
# Homoscedastic residuals vs Heteroscedastic residuals
res_homo = np.random.normal(0, 4, 300)
res_hetero = np.random.normal(0, scale=0.08 * y_pred, size=300)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Homoscedastic (Healthy model)
ax1.scatter(y_pred, res_homo, color='royalblue', alpha=0.6, edgecolors='none')
ax1.axhline(0, color='red', linestyle='--', lw=2)
ax1.axhline(np.std(res_homo) * 2, color='gray', linestyle=':')
ax1.axhline(-np.std(res_homo) * 2, color='gray', linestyle=':')
ax1.set_title('Well-Behaved Residuals (Homoscedastic)', fontweight='bold')
ax1.set_xlabel('Predicted Values')
ax1.set_ylabel('Residual (y - ŷ)')
ax1.grid(True, alpha=0.3)

# 2. Heteroscedastic (Funnel shape indicates error grows with prediction)
ax2.scatter(y_pred, res_hetero, color='darkorange', alpha=0.6, edgecolors='none')
ax2.axhline(0, color='red', linestyle='--', lw=2)
ax2.set_title('Funnel Pattern: Heteroscedasticity Warning', fontweight='bold')
ax2.set_xlabel('Predicted Values')
ax2.set_ylabel('Residual (y - ŷ)')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
