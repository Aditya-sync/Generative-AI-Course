# Annotations & Text

## Definition
Annotations and text allow you to **highlight specific data points**, explain trends, or
label important regions directly on the plot.

### Key Functions
| Function | Description |
|----------|-------------|
| `ax.text(x, y, str)` | Place text at data coordinates |
| `ax.annotate(str, xy, xytext)` | Arrow annotation pointing to data |
| `ax.axvline(x)` | Vertical reference line |
| `ax.axhline(y)` | Horizontal reference line |
| `ax.axvspan(x1, x2)` | Vertical shaded region |
| `ax.axhspan(y1, y2)` | Horizontal shaded region |
| `fig.text(x, y, str)` | Figure-level text (relative coords) |

### Annotation Arrow Styles
| arrowstyle | Appearance |
|------------|-----------|
| `'->'` | Simple arrow |
| `'-|>'` | Arrow with bar |
| `'fancy'` | Fancy arrow |
| `'wedge'` | Wedge shape |

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(11, 6))
ax.plot(x, y, 'steelblue', linewidth=2)

# Simple text
ax.text(0.1, 0.9, 'f(x) = sin(x)', fontsize=13,
        transform=ax.transAxes, color='steelblue',
        fontweight='bold')

# ax.annotate — arrow pointing to max
max_x = np.pi / 2
max_y = 1.0
ax.annotate('Maximum\n(π/2, 1)',
    xy=(max_x, max_y),              # point to annotate
    xytext=(max_x + 1, max_y - 0.3), # text position
    fontsize=11, color='darkred',
    arrowprops=dict(arrowstyle='->', color='darkred', lw=2),
    bbox=dict(boxstyle='round,pad=0.4', facecolor='lightyellow', edgecolor='darkred'))

# Arrow to minimum
ax.annotate('Minimum\n(3π/2, -1)',
    xy=(3*np.pi/2, -1),
    xytext=(3*np.pi/2 - 1, -0.6),
    fontsize=11, color='darkgreen',
    arrowprops=dict(arrowstyle='->', color='darkgreen', lw=2))

ax.set_title('Text & Annotations', fontsize=14)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x'); ax.set_ylabel('sin(x)')
plt.tight_layout()
plt.show()
```

```python
# Reference lines and spans
fig, ax = plt.subplots(figsize=(10, 5))
data = np.random.normal(100, 15, 300)
ax.plot(data[:100], 'o', color='steelblue', alpha=0.6, markersize=4)

# Horizontal reference lines
ax.axhline(y=100, color='black', linewidth=1.5, linestyle='--', label='Mean=100')
ax.axhline(y=130, color='tomato', linewidth=1.5, linestyle=':', label='Threshold=130')

# Vertical reference line
ax.axvline(x=50, color='seagreen', linewidth=2, linestyle='-', label='x=50')

# Shaded regions
ax.axhspan(85, 115, alpha=0.15, color='seagreen', label='±1σ band')
ax.axvspan(60, 80, alpha=0.15, color='orange', label='Period of interest')

ax.legend(fontsize=10, loc='upper right')
ax.set_title('Reference Lines & Spans')
plt.tight_layout()
plt.show()
```

```python
# Text in data vs axes coordinates
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, np.sin(x), 'steelblue', linewidth=2)

# Data coordinates (transform=None or default)
ax.text(np.pi, 0.5, 'Data coords (π, 0.5)', fontsize=10,
        ha='center', va='bottom',
        bbox=dict(boxstyle='round', facecolor='lightyellow'))

# Axes coordinates (0-1 relative to axes, transform=ax.transAxes)
ax.text(0.02, 0.02, 'Axes coords (0.02, 0.02)', fontsize=9,
        transform=ax.transAxes, color='darkred', style='italic')

# Figure coordinates
fig.text(0.5, 0.01, 'Figure coords (0.5, 0.01)', ha='center',
         fontsize=9, color='gray')

ax.set_title('Text Coordinate Systems')
plt.tight_layout()
plt.show()
```

```python
# Annotating multiple points (ML use case: labeling outliers)
rng = np.random.default_rng(42)
xs = rng.normal(0, 1, 100)
ys = rng.normal(0, 1, 100)

fig, ax = plt.subplots(figsize=(8, 7))
ax.scatter(xs, ys, alpha=0.5, color='steelblue', s=30)

# Find and annotate outliers (|x| > 2 or |y| > 2)
for i, (xi, yi) in enumerate(zip(xs, ys)):
    if abs(xi) > 2 or abs(yi) > 2:
        ax.annotate(f'#{i}', xy=(xi, yi), xytext=(xi+0.1, yi+0.1),
                    fontsize=8, color='darkred',
                    arrowprops=dict(arrowstyle='->', color='darkred', lw=0.8))

ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.set_title('Annotating Outliers')
ax.set_xlabel('Feature 1'); ax.set_ylabel('Feature 2')
plt.tight_layout()
plt.show()
```
