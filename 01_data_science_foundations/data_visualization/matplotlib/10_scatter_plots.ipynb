# Scatter Plots

## Definition
**Scatter plots** display individual data points as dots to reveal the **relationship between two numeric variables**.
They are the most important plot for ML feature analysis.

### When to Use
- Feature vs Target analysis
- Detecting correlations
- Visualizing clusters
- Outlier detection
- Model predictions vs actual values

### Key Parameters of `ax.scatter()`
| Parameter | Description |
|-----------|-------------|
| `x, y` | Data coordinates |
| `c` | Color (single color or array for colormap) |
| `s` | Marker size (can be array) |
| `alpha` | Transparency |
| `cmap` | Colormap (when `c` is numeric array) |
| `marker` | Marker style |
| `edgecolors` | Edge color |
| `linewidths` | Edge linewidth |
| `vmin`, `vmax` | Colormap range |

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)

# Basic scatter
n = 200
x = rng.normal(0, 1, n)
y = 0.7 * x + rng.normal(0, 0.5, n)

fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(x, y, color='steelblue', alpha=0.6, s=40, edgecolors='none')
ax.set_title('Basic Scatter Plot', fontsize=14)
ax.set_xlabel('Feature X'); ax.set_ylabel('Feature Y')
ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
# Colored by class label (classification visualization)
n = 150
X0 = rng.normal([-2, -2], 0.8, (n//3, 2))
X1 = rng.normal([0, 2], 0.8, (n//3, 2))
X2 = rng.normal([2, -1], 0.8, (n//3, 2))

X = np.vstack([X0, X1, X2])
y = np.array([0]*(n//3) + [1]*(n//3) + [2]*(n//3))

colors = ['steelblue', 'tomato', 'seagreen']
labels = ['Class 0', 'Class 1', 'Class 2']

fig, ax = plt.subplots(figsize=(8, 7))
for cls, (color, label) in enumerate(zip(colors, labels)):
    mask = y == cls
    ax.scatter(X[mask, 0], X[mask, 1], c=color, label=label,
               s=60, alpha=0.8, edgecolors='white', linewidths=0.5)

ax.set_title('Scatter by Class Label', fontsize=14)
ax.set_xlabel('Feature 1'); ax.set_ylabel('Feature 2')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Bubble chart — size encodes a third variable
n = 50
x_b = rng.uniform(0, 10, n)
y_b = rng.uniform(0, 10, n)
sizes = rng.uniform(50, 500, n)   # encode revenue
colors_b = rng.uniform(0, 1, n)  # encode growth rate

fig, ax = plt.subplots(figsize=(9, 7))
sc = ax.scatter(x_b, y_b, s=sizes, c=colors_b, cmap='RdYlGn',
                alpha=0.7, edgecolors='gray', linewidths=0.5)

cbar = plt.colorbar(sc, ax=ax)
cbar.set_label('Growth Rate', fontsize=11)
ax.set_title('Bubble Chart (size=revenue, color=growth rate)', fontsize=13)
ax.set_xlabel('Market Share'); ax.set_ylabel('Satisfaction')
plt.tight_layout()
plt.show()
```

```python
# Scatter with regression line overlay
from numpy.polynomial.polynomial import polyfit

x_r = rng.uniform(0, 10, 100)
y_r = 2.5 * x_r + rng.normal(0, 3, 100)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x_r, y_r, alpha=0.5, color='steelblue', s=40, label='Data')

# Regression line
coeffs = np.polyfit(x_r, y_r, 1)
x_line = np.linspace(0, 10, 100)
y_line = np.polyval(coeffs, x_line)
ax.plot(x_line, y_line, 'r-', linewidth=2.5,
        label=f'y = {coeffs[0]:.2f}x + {coeffs[1]:.2f}')

ax.set_title('Scatter + Regression Line', fontsize=14)
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# EXCEPTION: mismatched x and y lengths
try:
    fig, ax = plt.subplots()
    ax.scatter([1, 2, 3], [4, 5])  # lengths differ!
    plt.show()
except ValueError as e:
    print('ValueError:', e)
    plt.close()
```
