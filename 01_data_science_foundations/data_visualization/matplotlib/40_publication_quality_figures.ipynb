# Publication-Quality Figures (NeurIPS / ICML / Nature Style)

## Guidelines
- Column-width matching (single column: ~3.5 inches, double column: ~7 inches).
- Minimalist ink-to-data ratio (Tufte style: remove top/right spines, subtle grids).
- Legible typography matching IEEE / Springer font sizes (8–10 pt tick/legend, 10–12 pt labels).

```python
import matplotlib.pyplot as plt
import numpy as np

# Typical single-column conference width: 3.5 inches
fig, ax = plt.subplots(figsize=(4.5, 3.2), dpi=200)

np.random.seed(42)
steps = np.arange(1, 101)
base_loss = 1.8 / np.sqrt(steps) + 0.1
proposed_loss = 1.8 / (steps**0.65) + 0.05

ax.plot(steps, base_loss, label='Baseline (Transformer)', color='#7f8c8d', lw=1.5, linestyle='--')
ax.plot(steps, proposed_loss, label='Ours (Flash-Sparse)', color='#2980b9', lw=2.0)

ax.set_xlabel('Iteration Steps ($\\times 10^3$)', fontsize=9, fontweight='bold')
ax.set_ylabel('Validation Perplexity', fontsize=9, fontweight='bold')
ax.set_title('Language Model Convergence (ICLR / NeurIPS Format)', fontsize=10, fontweight='bold')

ax.legend(frameon=True, facecolor='white', edgecolor='none', fontsize=8, loc='upper right')
ax.grid(True, linestyle=':', alpha=0.5, color='gray')

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='both', which='major', labelsize=8)

plt.tight_layout()
plt.savefig('/tmp/publication_quality_demo.pdf', bbox_inches='tight')
plt.show()
```
