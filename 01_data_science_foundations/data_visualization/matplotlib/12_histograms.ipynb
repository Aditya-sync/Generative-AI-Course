# Histograms

## Definition
**Histograms** show the **frequency distribution** of a single numeric variable by dividing data
into bins and counting how many values fall in each bin.

### Key Parameters of `ax.hist()`
| Parameter | Description | Default |
|-----------|-------------|---------|
| `bins` | Number or list of bin edges | 'auto' |
| `density` | Normalize to probability density | False |
| `cumulative` | Cumulative distribution | False |
| `color` | Bar color | cycle |
| `edgecolor` | Bar border color | None |
| `alpha` | Transparency | 1.0 |
| `histtype` | 'bar', 'step', 'stepfilled' | 'bar' |
| `orientation` | 'vertical' or 'horizontal' | 'vertical' |

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)
data_normal = rng.normal(0, 1, 2000)
data_skewed = rng.exponential(scale=2, size=2000)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Basic histogram
n, bins, patches = axes[0].hist(data_normal, bins=40, color='steelblue',
                                 edgecolor='white', linewidth=0.5, alpha=0.85)
axes[0].set_title('Normal Distribution', fontsize=13)
axes[0].set_xlabel('Value'); axes[0].set_ylabel('Frequency')
axes[0].grid(True, alpha=0.3)

# Skewed with density normalization
axes[1].hist(data_skewed, bins=50, density=True, color='tomato',
             edgecolor='white', linewidth=0.5, alpha=0.85)
axes[1].set_title('Exponential Distribution (density=True)', fontsize=13)
axes[1].set_xlabel('Value'); axes[1].set_ylabel('Probability Density')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Overlapping histograms — compare distributions
group_a = rng.normal(5, 1.5, 1000)
group_b = rng.normal(7, 2.0, 1000)

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(group_a, bins=40, alpha=0.7, color='steelblue', edgecolor='white',
        density=True, label='Group A (μ=5, σ=1.5)')
ax.hist(group_b, bins=40, alpha=0.7, color='tomato', edgecolor='white',
        density=True, label='Group B (μ=7, σ=2.0)')

ax.set_title('Overlapping Histograms — Distribution Comparison', fontsize=13)
ax.set_xlabel('Value'); ax.set_ylabel('Density')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
# Histogram styles: bar, step, stepfilled
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
data = rng.normal(0, 1, 1000)

for ax, histtype, color in zip(axes,
    ['bar', 'step', 'stepfilled'],
    ['steelblue', 'tomato', 'seagreen']):
    ax.hist(data, bins=30, histtype=histtype, color=color, alpha=0.7)
    ax.set_title(f"histtype='{histtype}'")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Cumulative histogram
fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(data_normal, bins=50, cumulative=True, density=True,
        color='steelblue', histtype='step', linewidth=2, label='CDF')
ax.axhline(0.5, color='tomato', linestyle='--', label='50th percentile')
ax.axhline(0.95, color='seagreen', linestyle='--', label='95th percentile')

ax.set_title('Cumulative Distribution (CDF)', fontsize=13)
ax.set_xlabel('Value'); ax.set_ylabel('Cumulative Probability')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 2D histogram (hexbin)
x2d = rng.normal(0, 1, 5000)
y2d = rng.normal(0, 1, 5000) + 0.5 * x2d

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# hist2d
h, xedges, yedges, im = axes[0].hist2d(x2d, y2d, bins=40, cmap='Blues')
plt.colorbar(im, ax=axes[0])
axes[0].set_title('2D Histogram (hist2d)')

# hexbin
hb = axes[1].hexbin(x2d, y2d, gridsize=30, cmap='YlOrRd')
plt.colorbar(hb, ax=axes[1])
axes[1].set_title('Hexbin Plot')

plt.tight_layout()
plt.show()
```
