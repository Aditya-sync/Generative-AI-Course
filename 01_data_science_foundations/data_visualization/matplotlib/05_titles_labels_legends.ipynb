# Titles, Labels & Legends

## Definition
Titles, labels, and legends are the **text annotations** that make a plot interpretable.
Without them, even a beautiful plot is meaningless.

### Key Functions
| Function | Description |
|----------|-------------|
| `ax.set_title(str)` | Plot title |
| `fig.suptitle(str)` | Figure-level super title |
| `ax.set_xlabel(str)` | X-axis label |
| `ax.set_ylabel(str)` | Y-axis label |
| `ax.legend()` | Auto-detect labels and show legend |
| `ax.legend(handles, labels)` | Manual legend |
| `plt.colorbar()` | Colorbar for heatmaps |

### Legend Locations
| `loc` value | Position |
|-------------|---------|
| `'best'` | Auto pick best (default) |
| `'upper left/right'` | Corner positions |
| `'lower left/right'` | Corner positions |
| `'center'` | Center |
| `'outside right'` | `bbox_to_anchor=(1,1)` |

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 200)

fig, ax = plt.subplots(figsize=(10, 6))

# Plot multiple lines with labels
ax.plot(x, np.sin(x),     color='steelblue', linewidth=2, label='sin(x)')
ax.plot(x, np.cos(x),     color='tomato',    linewidth=2, label='cos(x)', linestyle='--')
ax.plot(x, np.sin(2*x)/2, color='seagreen',  linewidth=2, label='sin(2x)/2', linestyle='-.')

# Titles and labels
ax.set_title('Trigonometric Functions', fontsize=18, fontweight='bold', pad=15)
ax.set_xlabel('x (radians)', fontsize=13, labelpad=10)
ax.set_ylabel('y', fontsize=13, labelpad=10)

# Legend
ax.legend(fontsize=12, loc='upper right', framealpha=0.9,
          edgecolor='gray', fancybox=True)

ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Figure-level suptitle and tight_layout
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(x, np.sin(x), 'steelblue')
ax1.set_title('Subplot 1')
ax1.set_xlabel('x'); ax1.set_ylabel('sin(x)')

ax2.plot(x, np.cos(x), 'tomato')
ax2.set_title('Subplot 2')
ax2.set_xlabel('x'); ax2.set_ylabel('cos(x)')

# Figure-level title sits above all subplots
fig.suptitle('Figure Super Title', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Legend customization
fig, ax = plt.subplots(figsize=(9, 6))

lines = []
for i, (fn, name, color) in enumerate(
        [(np.sin, 'sin', 'steelblue'), (np.cos, 'cos', 'tomato'),
         (lambda x: x/np.pi - 1, 'linear', 'seagreen')]):
    line, = ax.plot(x, fn(x), color=color, linewidth=2, label=name)
    lines.append(line)

# Legend with custom properties
legend = ax.legend(
    handles=lines,
    labels=['sin(x)', 'cos(x)', 'Linear'],
    fontsize=12,
    title='Functions',
    title_fontsize=13,
    loc='upper right',
    framealpha=0.85,
    edgecolor='gray',
    shadow=True,
    ncol=1,             # number of legend columns
    handlelength=2.5,   # length of legend lines
    labelspacing=0.5,   # vertical spacing
)
ax.set_title('Custom Legend', fontsize=14)
plt.show()
```

```python
# Legend outside the plot
fig, ax = plt.subplots(figsize=(9, 5))
for i, (color, label) in enumerate(
    [('steelblue','Series A'),('tomato','Series B'),
     ('seagreen','Series C'),('darkorange','Series D')]):
    ax.plot(x, np.sin(x + i*0.5), color=color, label=label, linewidth=2)

# Place legend outside to the right
ax.legend(loc='upper left', bbox_to_anchor=(1, 1), fontsize=11)
ax.set_title('Legend Outside Plot')
plt.tight_layout()
plt.show()
```

```python
# Multi-line title, styled labels
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, np.sin(x), color='royalblue', linewidth=2)

# Multi-line title with LaTeX math
ax.set_title('$f(x) = \\sin(x)$\nRange: $[-1, 1]$',
             fontsize=14, fontweight='bold', color='darkblue')

# LaTeX in axis labels
ax.set_xlabel('$x \\in [0, 2\\pi]$', fontsize=12)
ax.set_ylabel('$\\sin(x)$', fontsize=12)

ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
