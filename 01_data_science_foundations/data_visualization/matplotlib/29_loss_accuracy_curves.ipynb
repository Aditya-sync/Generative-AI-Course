# Loss & Accuracy Curves in Matplotlib

## Deep Learning Training Diagnostics
- Detecting early stopping point
- Identifying overfitting when validation loss begins to diverge while train loss continues downward.

```python
import matplotlib.pyplot as plt
import numpy as np

epochs = np.arange(1, 41)
# Simulated DL metrics
train_loss = 2.0 * np.exp(-epochs / 8) + 0.05
val_loss = 2.0 * np.exp(-epochs / 8) + 0.05 + 0.0008 * (epochs - 18)**2 * (epochs > 18)

train_acc = 1 - 0.7 * np.exp(-epochs / 6)
val_acc = 1 - 0.7 * np.exp(-epochs / 6) - 0.002 * (epochs - 20) * (epochs > 20)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Loss Curve
ax1.plot(epochs, train_loss, 'b-', lw=2, label='Train Loss')
ax1.plot(epochs, val_loss, 'r--', lw=2, label='Val Loss')
ax1.axvline(18, color='gray', linestyle=':', lw=2, label='Optimal Early Stop (Epoch 18)')
ax1.set_title('Cross-Entropy Loss vs Epochs', fontweight='bold')
ax1.set_xlabel('Epochs')
ax1.set_ylabel('Loss')
ax1.legend()
ax1.grid(True, alpha=0.3)

# 2. Accuracy Curve
ax2.plot(epochs, train_acc * 100, 'b-', lw=2, label='Train Accuracy')
ax2.plot(epochs, val_acc * 100, 'g--', lw=2, label='Val Accuracy')
ax2.axvline(18, color='gray', linestyle=':', lw=2, label='Optimal Early Stop')
ax2.set_title('Classification Accuracy (%) vs Epochs', fontweight='bold')
ax2.set_xlabel('Epochs')
ax2.set_ylabel('Accuracy (%)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
