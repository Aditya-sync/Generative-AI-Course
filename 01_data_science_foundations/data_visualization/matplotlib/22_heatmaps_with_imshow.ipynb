# Heatmaps with imshow() in Matplotlib

## Core Applications
- Feature correlation matrices
- Confusion matrices
- Attention weight maps in Transformer models

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Feature Correlation Heatmap with Value Annotations
np.random.seed(42)
features = ['Age', 'Income', 'Credit_Score', 'Debt', 'Num_Cards', 'Default']
corr_matrix = np.random.uniform(-0.8, 0.8, size=(6, 6))
corr_matrix = (corr_matrix + corr_matrix.T) / 2
np.fill_diagonal(corr_matrix, 1.0)

fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)

ax.set_xticks(np.arange(len(features)))
ax.set_yticks(np.arange(len(features)))
ax.set_xticklabels(features, rotation=45, ha='right', fontweight='bold')
ax.set_yticklabels(features, fontweight='bold')

# Loop over data dimensions and create text annotations
for i in range(len(features)):
    for j in range(len(features)):
        text_color = 'white' if abs(corr_matrix[i, j]) > 0.5 else 'black'
        ax.text(j, i, f'{corr_matrix[i, j]:.2f}', ha='center', va='center',
                color=text_color, fontsize=10, fontweight='bold')

fig.colorbar(im, ax=ax, shrink=0.8, label='Pearson Correlation')
ax.set_title('Feature Correlation Matrix Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 2. Transformer Self-Attention Heatmap Simulation
tokens = ['[CLS]', 'The', 'model', 'learned', 'attention', 'weights', '[SEP]']
T = len(tokens)
np.random.seed(0)
# Softmax simulated attention
logits = np.random.randn(T, T) * 1.5
attn = np.exp(logits) / np.exp(logits).sum(axis=-1, keepdims=True)

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(attn, cmap='Blues')

ax.set_xticks(range(T))
ax.set_yticks(range(T))
ax.set_xticklabels(tokens, rotation=45, ha='right')
ax.set_yticklabels(tokens)

fig.colorbar(im, ax=ax, label='Attention Weight')
ax.set_title('Transformer Multi-Head Self-Attention Map', fontweight='bold')
plt.tight_layout()
plt.show()
```
