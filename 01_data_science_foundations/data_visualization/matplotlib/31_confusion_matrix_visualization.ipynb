# Confusion Matrix Visualization in Matplotlib

## Key Components
- Raw Counts vs Normalized Proportions
- Annotations with contrast coloring for high/low cell values.

```python
import matplotlib.pyplot as plt
import numpy as np

# 4-class Confusion Matrix
classes = ['Cat', 'Dog', 'Bird', 'Fish']
cm = np.array([
    [145,  12,   3,   0],
    [ 10, 160,   5,   1],
    [  4,   8, 128,  10],
    [  0,   2,   7, 151]
])

# Normalized CM
cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

# Raw count heatmap
im1 = ax1.imshow(cm, cmap='Blues')
ax1.set_xticks(range(len(classes)))
ax1.set_yticks(range(len(classes)))
ax1.set_xticklabels(classes, fontweight='bold')
ax1.set_yticklabels(classes, fontweight='bold')
ax1.set_xlabel('Predicted Label', fontweight='bold')
ax1.set_ylabel('True Label', fontweight='bold')
ax1.set_title('Confusion Matrix (Raw Counts)', fontweight='bold')

for i in range(len(classes)):
    for j in range(len(classes)):
        color = "white" if cm[i, j] > cm.max() / 2 else "black"
        ax1.text(j, i, str(cm[i, j]), ha="center", va="center", color=color, fontweight='bold')
fig.colorbar(im1, ax=ax1, shrink=0.8)

# Normalized heatmap
im2 = ax2.imshow(cm_norm, cmap='Greens', vmin=0, vmax=1)
ax2.set_xticks(range(len(classes)))
ax2.set_yticks(range(len(classes)))
ax2.set_xticklabels(classes, fontweight='bold')
ax2.set_yticklabels(classes, fontweight='bold')
ax2.set_xlabel('Predicted Label', fontweight='bold')
ax2.set_ylabel('True Label', fontweight='bold')
ax2.set_title('Confusion Matrix (Normalized Recall)', fontweight='bold')

for i in range(len(classes)):
    for j in range(len(classes)):
        color = "white" if cm_norm[i, j] > 0.5 else "black"
        ax2.text(j, i, f'{cm_norm[i, j]:.2f}', ha="center", va="center", color=color, fontweight='bold')
fig.colorbar(im2, ax=ax2, shrink=0.8)

plt.tight_layout()
plt.show()
```
