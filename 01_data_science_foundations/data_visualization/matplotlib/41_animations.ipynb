# Animations in Matplotlib

## Core Tool
- `matplotlib.animation.FuncAnimation`: updates artist properties per frame.
- Crucial for visualizing Gradient Descent convergence, live sensor feeds, or cellular automata.

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

# Simulating 1D Gradient Descent optimization of f(x) = x^2 - 4x + 6
f = lambda x: (x - 2)**2 + 2
df = lambda x: 2 * (x - 2)

x_trajectory = [5.0]
lr = 0.25
for _ in range(25):
    x_next = x_trajectory[-1] - lr * df(x_trajectory[-1])
    x_trajectory.append(x_next)

fig, ax = plt.subplots(figsize=(8, 5))
x_curve = np.linspace(-1, 6, 200)
ax.plot(x_curve, f(x_curve), 'b-', lw=2, label=r'$f(x) = (x-2)^2 + 2$')
point, = ax.plot([], [], 'ro', markersize=10, label='Optimizer State')
line_path, = ax.plot([], [], 'r--', alpha=0.5)

ax.set_xlim(-1, 6)
ax.set_ylim(0, 15)
ax.set_xlabel('Parameter x')
ax.set_ylabel('Objective Value')
ax.set_title('Gradient Descent Step-by-Step Trajectory', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

def init():
    point.set_data([], [])
    line_path.set_data([], [])
    return point, line_path

def animate(i):
    curr_x = x_trajectory[:i+1]
    curr_y = [f(x) for x in curr_x]
    point.set_data([curr_x[-1]], [curr_y[-1]])
    line_path.set_data(curr_x, curr_y)
    return point, line_path

anim = animation.FuncAnimation(fig, animate, init_func=init, frames=len(x_trajectory), interval=150, blit=True)
print("FuncAnimation object created:", type(anim).__name__)
plt.close(fig)  # Prevent static display in headless runner
```
