# Object-Oriented Plotting in Matplotlib

## Deep Dive into OO Paradigm
- Managing `Figure`, `Axes`, `Axis`, and `Artist` instances explicitly.
- Avoiding global state issues in production pipelines and multi-threaded environments.

```python
import matplotlib.pyplot as plt
import numpy as np

# Object-Oriented pipeline function
def plot_model_diagnostics(x_vals, y_true, y_pred, feature_name='Input Feature'):
    """Modular, thread-safe function returning Figure and Axes references."""
    fig, axes = plt.subplots(1, 2, figsize=(13, 5), dpi=100)
    ax_fit, ax_res = axes
    
    # Left: Fits
    ax_fit.scatter(x_vals, y_true, color='#2c3e50', alpha=0.5, label='Actual Ground Truth')
    sort_idx = np.argsort(x_vals)
    ax_fit.plot(x_vals[sort_idx], y_pred[sort_idx], color='#e74c3c', lw=2.5, label='Model Prediction')
    ax_fit.set_xlabel(feature_name, fontsize=11, fontweight='bold')
    ax_fit.set_ylabel('Target Value', fontsize=11, fontweight='bold')
    ax_fit.set_title('Regression Fit Analysis', fontsize=12, fontweight='bold')
    ax_fit.legend()
    ax_fit.grid(True, alpha=0.3)
    
    # Right: Residuals
    residuals = y_true - y_pred
    ax_res.scatter(y_pred, residuals, color='#8e44ad', alpha=0.6)
    ax_res.axhline(0, color='black', linestyle='--', lw=1.5)
    ax_res.set_xlabel('Predicted Value', fontsize=11, fontweight='bold')
    ax_res.set_ylabel('Residual Error', fontsize=11, fontweight='bold')
    ax_res.set_title('Residual Distribution', fontsize=12, fontweight='bold')
    ax_res.grid(True, alpha=0.3)
    
    # Despine clean appearance
    for ax in axes:
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        
    return fig, axes

# Example usage
np.random.seed(42)
x = np.random.uniform(0, 10, 150)
y = 3.5 * x + np.sin(x) * 4 + np.random.normal(0, 2, 150)
y_hat = 3.4 * x + 1.5

fig, axes = plot_model_diagnostics(x, y, y_hat)
plt.tight_layout()
plt.show()
```
