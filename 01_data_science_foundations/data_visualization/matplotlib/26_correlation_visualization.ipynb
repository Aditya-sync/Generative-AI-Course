# Correlation Visualization in Matplotlib

## Methods
- Clustered / Ordered Correlation Matrix
- Lower-triangle masked heatmap to eliminate redundancy.

```python
import matplotlib.pyplot as plt
import numpy as np

# Synthetic correlation matrix
np.random.seed(101)
n_features = 8
labels = [f'Feat_{i}' for i in range(1, n_features + 1)]
A = np.random.randn(n_features, n_features)
corr = np.corrcoef(A)

# Lower triangle mask
mask = np.triu(np.ones_like(corr, dtype=bool))
corr_masked = np.ma.masked_where(mask, corr)

fig, ax = plt.subplots(figsize=(8, 7))
cmap = plt.cm.coolwarm
cmap.set_bad('white')

im = ax.imshow(corr_masked, cmap=cmap, vmin=-1, vmax=1)

ax.set_xticks(range(n_features))
ax.set_yticks(range(n_features))
ax.set_xticklabels(labels, rotation=45, ha='right')
ax.set_yticklabels(labels)

for i in range(n_features):
    for j in range(n_features):
        if i > j:
            val = corr[i, j]
            text_color = "white" if abs(val) > 0.6 else "black"
            ax.text(j, i, f'{val:.2f}', ha='center', va='center', color=text_color, fontweight='bold', fontsize=9)

cbar = fig.colorbar(im, ax=ax, shrink=0.75)
cbar.set_label('Pearson Correlation Coefficient')
ax.set_title('Lower-Triangle Masked Correlation Heatmap', fontweight='bold')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
