# Creating Figures and Subplots

## Definition
Matplotlib provides multiple ways to create figures with one or more subplot areas.

### Methods
| Method | Description |
|--------|-------------|
| `plt.subplots(nrows, ncols)` | Regular grid of subplots |
| `plt.subplot(nrows, ncols, index)` | Add subplot to existing figure |
| `fig.add_subplot(nrows, ncols, index)` | OO version of above |
| `plt.subplot2grid(shape, loc)` | Irregular grid with spans |
| `GridSpec` | Full control over grid layout |
| `fig.add_axes([l,b,w,h])` | Arbitrary position axes |
| `plt.subplots_adjust(...)` | Manual spacing control |

### GridSpec (most powerful)
```python
import matplotlib.gridspec as gridspec
gs = gridspec.GridSpec(2, 3)
ax1 = fig.add_subplot(gs[0, :])    # spans all columns in row 0
ax2 = fig.add_subplot(gs[1, 0])    # row 1, col 0
ax3 = fig.add_subplot(gs[1, 1:])   # row 1, cols 1-2
```

````python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np

x = np.linspace(0, 2*np.pi, 100)

# Basic: 1 row, 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(x, np.sin(x), color='steelblue', linewidth=2)
ax1.set_title('Subplot 1: sin(x)')
ax2.plot(x, np.cos(x), color='tomato', linewidth=2)
ax2.set_title('Subplot 2: cos(x)')
fig.suptitle('1×2 Grid', fontsize=14)
plt.tight_layout()
plt.show()
````

````python
# 2D grid of subplots
fig, axes = plt.subplots(2, 3, figsize=(14, 8), sharex=True, sharey=False)

colors = ['steelblue', 'tomato', 'seagreen', 'darkorange', 'purple', 'brown']
fns = [np.sin, np.cos, np.tan, np.sinh, np.cosh, lambda x: x**2]
names = ['sin', 'cos', 'tan', 'sinh', 'cosh', 'x²']

for ax, fn, name, color in zip(axes.flat, fns, names, colors):
    y = fn(x)
    ax.plot(x, y.clip(-10, 10), color=color, linewidth=1.5)
    ax.set_title(name, fontsize=12)
    ax.grid(True, alpha=0.3)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

fig.suptitle('2×3 Subplot Grid', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# GridSpec — irregular layout
fig = plt.figure(figsize=(12, 8))
gs = gridspec.GridSpec(2, 3, figure=fig, hspace=0.4, wspace=0.3)

# Wide top plot spanning all columns
ax_top = fig.add_subplot(gs[0, :])
ax_top.plot(x, np.sin(x), color='steelblue', linewidth=2)
ax_top.set_title('Top Wide Plot (spans 3 columns)')

# Bottom-left plot spanning 2 columns
ax_bl = fig.add_subplot(gs[1, :2])
ax_bl.plot(x, np.cos(x), color='tomato', linewidth=2)
ax_bl.set_title('Bottom-Left (2 columns)')

# Bottom-right single column
ax_br = fig.add_subplot(gs[1, 2])
ax_br.hist(np.random.randn(1000), bins=30, color='seagreen')
ax_br.set_title('Bottom-Right')

fig.suptitle('GridSpec — Irregular Layout', fontsize=15, fontweight='bold')
plt.show()
````

````python
# subplot2grid
fig = plt.figure(figsize=(12, 6))

ax1 = plt.subplot2grid((2, 4), (0, 0), colspan=2)  # row0, col0-1
ax2 = plt.subplot2grid((2, 4), (0, 2), colspan=2)  # row0, col2-3
ax3 = plt.subplot2grid((2, 4), (1, 0), colspan=4)  # row1, all cols

ax1.plot(x, np.sin(x), 'steelblue'); ax1.set_title('sin(x)')
ax2.scatter(np.random.randn(50), np.random.randn(50), c='tomato', alpha=0.6)
ax2.set_title('Scatter')
ax3.plot(x, np.sin(x) * np.cos(x), 'seagreen', linewidth=2)
ax3.set_title('sin(x)·cos(x) — Full Width')

plt.tight_layout()
plt.show()
````

````python
# Inset axes (axes within axes)
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x, np.sin(x), color='steelblue', linewidth=2)
ax.set_title('Main Plot with Inset')
ax.set_xlabel('x'); ax.set_ylabel('sin(x)')

# Add inset axes (position: [left, bottom, width, height] relative to parent ax)
axins = ax.inset_axes([0.6, 0.6, 0.35, 0.35])
axins.plot(x[:20], np.sin(x[:20]), color='tomato', linewidth=1.5)
axins.set_title('Zoomed', fontsize=9)
axins.set_xlim(0, 0.5); axins.set_ylim(0, 0.5)

plt.show()
````

````python
# Adjusting spacing between subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for ax, i in zip(axes.flat, range(4)):
    ax.plot(x, np.sin(x + i * np.pi/4))
    ax.set_title(f'Plot {i+1}')

# Manual spacing
fig.subplots_adjust(left=0.1, right=0.95, top=0.9, bottom=0.1,
                    hspace=0.4, wspace=0.3)
fig.suptitle('Manually adjusted spacing', fontsize=13)
plt.show()
````
