# Interactive / Event-Based Plots in Matplotlib

## Event Handling & Widgets
- Matplotlib `mpl_connect` event callbacks (`button_press_event`, `motion_notify_event`, `key_press_event`).
- Interactive GUI widgets via `matplotlib.widgets.Slider` and `matplotlib.widgets.Button`.

```python
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider, Button
import numpy as np

# Slider-controlled Interactive Sine Wave
fig, ax = plt.subplots(figsize=(9, 6))
plt.subplots_adjust(bottom=0.25)

t = np.linspace(0.0, 1.0, 1000)
init_freq = 3.0
init_amp = 1.0
s = init_amp * np.sin(2 * np.pi * init_freq * t)
l, = ax.plot(t, s, lw=2, color='royalblue')
ax.set_title('Interactive Waveform Controller (Sliders)', fontweight='bold')
ax.grid(True, alpha=0.3)

# Add slider axes
ax_freq = plt.axes([0.25, 0.1, 0.65, 0.03])
ax_amp = plt.axes([0.25, 0.05, 0.65, 0.03])

slider_freq = Slider(ax_freq, 'Freq (Hz)', 0.1, 30.0, valinit=init_freq)
slider_amp = Slider(ax_amp, 'Amplitude', 0.1, 5.0, valinit=init_amp)

def update(val):
    amp = slider_amp.val
    freq = slider_freq.val
    l.set_ydata(amp * np.sin(2 * np.pi * freq * t))
    ax.set_ylim(-amp * 1.1, amp * 1.1)
    fig.canvas.draw_idle()

slider_freq.on_changed(update)
slider_amp.on_changed(update)

print("Interactive widgets registered successfully.")
plt.close(fig)
```

```python
# Matplotlib Event Handling Callback Demonstration
fig, ax = plt.subplots(figsize=(6, 4))
ax.set_title('Click event handler demo')

def on_click(event):
    if event.inaxes:
        print(f'Button clicked at coordinates: X={event.xdata:.2f}, Y={event.ydata:.2f}')

cid = fig.canvas.mpl_connect('button_press_event', on_click)
print('Connected event handler ID:', cid)
plt.close(fig)
```
