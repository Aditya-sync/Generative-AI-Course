# Learning Curves in Matplotlib

## Purpose
- Diagnose **High Bias (Underfitting)** vs **High Variance (Overfitting)** as a function of training dataset size.

```python
import matplotlib.pyplot as plt
import numpy as np

train_sizes = np.linspace(50, 1000, 10, dtype=int)

# Simulated scores with standard deviations
train_mean = 0.99 - 0.08 * np.exp(-train_sizes / 300)
train_std = 0.015 * np.ones_like(train_mean)

val_mean = 0.65 + 0.25 * (1 - np.exp(-train_sizes / 350))
val_std = 0.035 * np.exp(-train_sizes / 500) + 0.01

fig, ax = plt.subplots(figsize=(9, 5))

# Training score curve
ax.plot(train_sizes, train_mean, 'o-', color='#e74c3c', lw=2, label='Training Score')
ax.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.15, color='#e74c3c')

# Cross-validation score curve
ax.plot(train_sizes, val_mean, 'o-', color='#27ae60', lw=2, label='Cross-Validation Score')
ax.fill_between(train_sizes, val_mean - val_std, val_mean + val_std, alpha=0.15, color='#27ae60')

ax.set_xlabel('Training Set Size (Samples)', fontweight='bold')
ax.set_ylabel('Model Accuracy', fontweight='bold')
ax.set_ylim(0.5, 1.02)
ax.set_title('Model Learning Curve Analysis (Bias vs Variance)', fontsize=14, fontweight='bold')
ax.legend(loc='lower right')
ax.grid(True, alpha=0.3)

# Annotation for variance gap
mid_idx = 7
ax.annotate('Generalization Gap',
            xy=(train_sizes[mid_idx], (train_mean[mid_idx] + val_mean[mid_idx]) / 2),
            xytext=(train_sizes[mid_idx] - 200, 0.75),
            arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5),
            fontweight='bold')

plt.tight_layout()
plt.show()
```
