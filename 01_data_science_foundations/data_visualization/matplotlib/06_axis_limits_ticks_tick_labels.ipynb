# Axis Limits, Ticks & Tick Labels

## Definition
Controlling the **axis range** and **tick marks** is essential for highlighting relevant data
and creating interpretable plots.

### Key Functions
| Function | Description |
|----------|-------------|
| `ax.set_xlim(lo, hi)` | Set x-axis range |
| `ax.set_ylim(lo, hi)` | Set y-axis range |
| `ax.set_xticks([...])` | Set tick positions |
| `ax.set_xticklabels([...])` | Set tick labels |
| `ax.xaxis.set_major_locator(...)` | Set tick spacing rule |
| `ax.xaxis.set_major_formatter(...)` | Set tick label format |
| `ax.tick_params(...)` | Tick size, color, direction |
| `plt.xticks(rotation=45)` | Rotate x tick labels |

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np

x = np.linspace(0, 4*np.pi, 300)

# Setting axis limits
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].plot(x, np.sin(x), 'steelblue')
axes[0].set_title('Auto limits')

axes[1].plot(x, np.sin(x), 'steelblue')
axes[1].set_xlim(0, 2*np.pi)
axes[1].set_ylim(-1.5, 1.5)
axes[1].set_title('set_xlim / set_ylim')

axes[2].plot(x, np.sin(x), 'steelblue')
axes[2].axis([0, np.pi, -1, 1])  # [xmin, xmax, ymin, ymax]
axes[2].set_title('axis([xmin, xmax, ymin, ymax])')

plt.tight_layout()
plt.show()
```

```python
# Custom tick positions and labels
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(x, np.sin(x), 'steelblue', linewidth=2)

# Set custom x ticks at pi multiples
pi_ticks = [0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi, 5*np.pi/2, 3*np.pi, 4*np.pi]
pi_labels = ['0', 'π/2', 'π', '3π/2', '2π', '5π/2', '3π', '4π']

ax.set_xticks(pi_ticks)
ax.set_xticklabels(pi_labels, fontsize=11)
ax.set_title('Custom Pi Tick Labels')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Tick locators and formatters
fig, axes = plt.subplots(2, 2, figsize=(13, 8))
y = np.sin(x)

# MultipleLocator — ticks every N units
ax = axes[0, 0]
ax.plot(x, y, 'steelblue')
ax.xaxis.set_major_locator(ticker.MultipleLocator(np.pi))
ax.xaxis.set_minor_locator(ticker.MultipleLocator(np.pi/4))
ax.set_title('MultipleLocator(π) major, π/4 minor')
ax.grid(True, which='both', alpha=0.3)

# MaxNLocator — at most N ticks
ax = axes[0, 1]
ax.plot(x, y, 'tomato')
ax.xaxis.set_major_locator(ticker.MaxNLocator(5))
ax.set_title('MaxNLocator(5)')

# FormatStrFormatter — format tick labels
ax = axes[1, 0]
vals = np.linspace(0, 10000, 200)
ax.plot(vals, vals**2)
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.0f'))
ax.xaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{x:,.0f}'))
ax.set_title('FormatStrFormatter & FuncFormatter')

# LogLocator — for log scale
ax = axes[1, 1]
y_log = np.exp(np.linspace(0, 10, 100))
ax.semilogy(y_log, color='seagreen')
ax.set_title('Log scale (semilogy)')
ax.grid(True, which='both', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# tick_params — customize tick appearance
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, np.sin(x), 'steelblue', linewidth=2)

ax.tick_params(axis='x', which='major',
               length=8, width=2, color='darkred',
               labelcolor='darkred', labelsize=12, rotation=30)
ax.tick_params(axis='y', which='major',
               length=6, width=1.5, color='darkblue',
               labelcolor='darkblue', labelsize=11)

ax.set_title('Custom tick_params')
plt.tight_layout()
plt.show()
```

```python
# Date ticks — very common in ML/time-series
import matplotlib.dates as mdates
from datetime import datetime, timedelta

dates = [datetime(2023, 1, 1) + timedelta(days=i) for i in range(90)]
values = np.cumsum(np.random.randn(90))

fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(dates, values, color='steelblue', linewidth=1.5)

ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax.xaxis.set_minor_locator(mdates.WeekdayLocator(byweekday=0))

plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right')
ax.set_title('Date Ticks with Formatters')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
