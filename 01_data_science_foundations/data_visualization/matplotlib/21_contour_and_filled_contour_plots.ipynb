# Contour & Filled Contour Plots

## Functions
- `ax.contour()`: Draws isolines / level curves.
- `ax.contourf()`: Fills regions between isolines.
- Essential in ML for visualizing loss surfaces, optimization landscapes, and 2D decision surfaces.

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Loss Surface Contour Plot (Rosenbrock banana function)
def rosenbrock(x, y, a=1, b=100):
    return (a - x)**2 + b * (y - x**2)**2

x = np.linspace(-2, 2, 250)
y = np.linspace(-1, 3, 250)
X, Y = np.meshgrid(x, y)
Z = rosenbrock(X, Y)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Contour lines with labels
levels = np.logspace(0, 3.5, 20)
CS = ax1.contour(X, Y, Z, levels=levels, cmap='jet')
ax1.clabel(CS, inline=True, fontsize=8, fmt='%.1f')
ax1.plot(1, 1, 'r*', markersize=14, label='Global Min (1,1)')
ax1.set_title('Contour Lines with clabel()', fontweight='bold')
ax1.set_xlabel('Parameter w1')
ax1.set_ylabel('Parameter w2')
ax1.legend()

# Filled contour (contourf)
CF = ax2.contourf(X, Y, np.log1p(Z), levels=30, cmap='viridis')
fig.colorbar(CF, ax=ax2, label='log(1 + Loss)')
ax2.plot(1, 1, 'r*', markersize=14, label='Global Min')
ax2.set_title('Filled Contour (Loss Landscape)', fontweight='bold')
ax2.set_xlabel('Parameter w1')
ax2.set_ylabel('Parameter w2')
ax2.legend()

plt.tight_layout()
plt.show()
```
