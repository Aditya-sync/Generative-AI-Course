# Saving Figures in Matplotlib

## Formats & Options
- **Raster Formats**: PNG, JPEG, WebP (`dpi=300`, lossless/lossy compression).
- **Vector Formats**: SVG, PDF, EPS (resolution-independent; ideal for academic papers and posters).
- Essential arguments: `bbox_inches='tight'`, `transparent=True/False`, `pad_inches=0.1`.

```python
import matplotlib.pyplot as plt
import numpy as np
import os

fig, ax = plt.subplots(figsize=(6, 4), dpi=100)
x = np.linspace(-3, 3, 100)
ax.plot(x, 1 / (1 + np.exp(-x)), color='#16a085', lw=2.5, label='Sigmoid $\sigma(x)$')
ax.set_title('Sigmoid Activation Function', fontweight='bold')
ax.set_xlabel('Logits (z)')
ax.set_ylabel('Probability')
ax.grid(True, alpha=0.3)
ax.legend()

# 1. Saving to High-Resolution PNG (300 DPI)
png_path = '/tmp/sigmoid_plot_300dpi.png'
fig.savefig(png_path, dpi=300, bbox_inches='tight')

# 2. Saving to Vector SVG
svg_path = '/tmp/sigmoid_plot.svg'
fig.savefig(svg_path, format='svg', bbox_inches='tight')

# 3. Saving to PDF for Academic Papers
pdf_path = '/tmp/sigmoid_plot.pdf'
fig.savefig(pdf_path, format='pdf', bbox_inches='tight')

print(f"PNG size: {os.path.getsize(png_path)} bytes")
print(f"SVG size: {os.path.getsize(svg_path)} bytes")
print(f"PDF size: {os.path.getsize(pdf_path)} bytes")
plt.close(fig)
```
