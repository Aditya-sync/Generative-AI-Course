# Subplots & Complex Layouts in Matplotlib

## Layout Tools
1. `plt.subplots(nrows, ncols)`: Simple uniform grid.
2. `plt.subplot_mosaic()`: Semantic ASCII-like grid (introduced in Matplotlib 3.3).
3. `matplotlib.gridspec.GridSpec`: Fine-grained span and aspect ratio control.

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np

# 1. subplot_mosaic: cleanest way to design modern dashboard layouts
mosaic = [
    ['main', 'main', 'side_top'],
    ['main', 'main', 'side_bot'],
    ['bottom', 'bottom', 'bottom']
]

fig, ax_dict = plt.subplot_mosaic(mosaic, figsize=(12, 8))

# Main plot
x = np.linspace(-3, 3, 200)
ax_dict['main'].plot(x, np.sinc(x), color='crimson', lw=2)
ax_dict['main'].set_title('Primary Analysis (Main)', fontweight='bold')
ax_dict['main'].grid(True, alpha=0.3)

# Side top
ax_dict['side_top'].hist(np.random.randn(500), bins=20, color='teal', edgecolor='black')
ax_dict['side_top'].set_title('Distribution A', fontsize=10)

# Side bottom
ax_dict['side_bot'].scatter(np.random.rand(50), np.random.rand(50), color='purple', alpha=0.7)
ax_dict['side_bot'].set_title('Scatter Correlation', fontsize=10)

# Bottom full width
ax_dict['bottom'].plot(np.cumsum(np.random.randn(200)), color='darkorange')
ax_dict['bottom'].set_title('Residual Time Series (Full Width)', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# 2. GridSpec with unequal height/width ratios (Joint Plot layout)
fig = plt.figure(figsize=(9, 9))
gs = gridspec.GridSpec(4, 4, hspace=0.1, wspace=0.1)

ax_main = fig.add_subplot(gs[1:4, 0:3])
ax_xhist = fig.add_subplot(gs[0, 0:3], sharex=ax_main)
ax_yhist = fig.add_subplot(gs[1:4, 3], sharey=ax_main)

# Data
np.random.seed(42)
x_data = np.random.randn(1000)
y_data = 2 * x_data + np.random.randn(1000) * 1.5

ax_main.scatter(x_data, y_data, alpha=0.4, color='#34495e', edgecolors='none', s=20)
ax_main.set_xlabel('Feature X', fontsize=11)
ax_main.set_ylabel('Target Y', fontsize=11)

ax_xhist.hist(x_data, bins=35, color='#3498db', edgecolor='white')
ax_xhist.axis('off')

ax_yhist.hist(y_data, bins=35, orientation='horizontal', color='#2ecc71', edgecolor='white')
ax_yhist.axis('off')

fig.suptitle('Custom Joint Marginal Distribution via GridSpec', fontsize=14, fontweight='bold', y=0.94)
plt.show()
```
