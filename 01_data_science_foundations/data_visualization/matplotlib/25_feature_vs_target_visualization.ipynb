# Feature vs Target Visualization in Matplotlib

## Core Use Cases
- Continuous Feature vs Continuous Target (Regression trends, heteroscedasticity).
- Continuous Feature vs Categorical Target (Classification discriminative power).

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
N = 300
feature_continuous = np.linspace(10, 100, N)
# Non-linear relationship with heteroscedastic noise
noise = np.random.normal(0, scale=feature_continuous * 0.15)
target_continuous = 2.5 * feature_continuous + 0.05 * (feature_continuous ** 1.8) + noise

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Regression relationship
ax1.scatter(feature_continuous, target_continuous, alpha=0.6, color='steelblue', edgecolors='none', label='Observations')
# Trend line (polynomial fit)
z = np.polyfit(feature_continuous, target_continuous, 2)
p = np.poly1d(z)
ax1.plot(feature_continuous, p(feature_continuous), "r--", lw=2.5, label='Fitted Trend (Degree 2)')
ax1.set_xlabel('Engine Power (kW)')
ax1.set_ylabel('Fuel Consumption (L/100km)')
ax1.set_title('Feature vs Continuous Target (Regression)', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Classification relationship (Feature split by binary class)
class_0 = np.random.normal(loc=45, scale=12, size=200)
class_1 = np.random.normal(loc=65, scale=10, size=200)

bp = ax2.boxplot([class_0, class_1], labels=['Class 0 (Non-Churn)', 'Class 1 (Churn)'],
                 patch_artist=True, widths=0.5)
colors = ['#5dade2', '#ec7063']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
for median in bp['medians']:
    median.set(color='black', linewidth=2)

ax2.set_ylabel('Customer Engagement Score')
ax2.set_title('Feature Distribution by Class Target', fontweight='bold')
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
