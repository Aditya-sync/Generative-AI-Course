# Figure and Axes Architecture

## Definition
Matplotlib is built around two core objects:
- **Figure** (`fig`): The entire canvas / window. Can contain one or more Axes.
- **Axes** (`ax`): The actual plot area where data is drawn. A Figure can have multiple Axes.

### Hierarchy
```
Figure
  └── Axes (one or more)
        ├── XAxis
        │     ├── Ticks
        │     └── Label
        ├── YAxis
        │     ├── Ticks
        │     └── Label
        ├── Title
        ├── Legend
        └── Artists (lines, patches, text, images)
```

### Two Interfaces
| Interface | Style | Best for |
|-----------|-------|---------|
| **pyplot** (`plt.*`) | MATLAB-style, implicit | Quick exploration |
| **Object-Oriented** (`fig, ax`) | Explicit | Complex, production plots |

> ✅ Always prefer the **OO interface** (`fig, ax = plt.subplots()`) for anything beyond a quick sketch.

````python
import matplotlib.pyplot as plt
import numpy as np

# The anatomy of a matplotlib Figure
fig = plt.figure(figsize=(10, 6))

# Add Axes manually (left, bottom, width, height as fractions of Figure)
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])  # [left, bottom, width, height]

x = np.linspace(0, 2 * np.pi, 100)
ax.plot(x, np.sin(x), label='sin(x)', color='steelblue', linewidth=2)
ax.plot(x, np.cos(x), label='cos(x)', color='tomato', linewidth=2, linestyle='--')

ax.set_title('Figure & Axes Architecture Demo', fontsize=16, fontweight='bold')
ax.set_xlabel('x (radians)')
ax.set_ylabel('y')
ax.legend()
ax.grid(True, alpha=0.3)

plt.savefig('/tmp/fig_axes_demo.png', dpi=100, bbox_inches='tight')
plt.show()
print('Figure type:', type(fig))
print('Axes type:', type(ax))
````

````python
# Object-oriented interface — preferred approach
fig, ax = plt.subplots(figsize=(8, 5))

x = np.linspace(0, 4, 100)
ax.plot(x, x**2, color='darkorange', label='x²')
ax.fill_between(x, x**2, alpha=0.2, color='darkorange')

ax.set_title('Object-Oriented Interface', fontsize=14)
ax.set_xlabel('x')
ax.set_ylabel('x²')
ax.legend()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
print('Number of axes in figure:', len(fig.axes))
````

````python
# Accessing Figure and Axes properties
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [4, 5, 6])

print('Figure size (inches):', fig.get_size_inches())
print('Figure DPI:', fig.get_dpi())
print('Axes xlim:', ax.get_xlim())
print('Axes ylim:', ax.get_ylim())
print('Axes title:', ax.get_title())
print('XAxis label:', ax.get_xlabel())
print('All children of ax:', [type(c).__name__ for c in ax.get_children()[:5]])
plt.close()
````

````python
# Multiple axes in one figure
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

x = np.linspace(0, 2*np.pi, 100)
plots = [
    (np.sin(x),  'sin(x)',  'steelblue'),
    (np.cos(x),  'cos(x)',  'tomato'),
    (np.tan(x),  'tan(x)',  'green'),
    (x**2,       'x²',      'purple'),
    (np.exp(x/3),'e^(x/3)', 'orange'),
    (np.log(x+1),'log(x+1)','brown'),
]

for ax, (y, title, color) in zip(axes.flat, plots):
    ax.plot(x, y, color=color, linewidth=1.5)
    ax.set_title(title)
    ax.grid(True, alpha=0.3)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

fig.suptitle('Multiple Axes Demo', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

````python
# EXCEPTION: Plotting after plt.close()
fig, ax = plt.subplots()
ax.plot([1, 2, 3])
plt.close(fig)

try:
    ax.plot([4, 5, 6])   # can still call on closed fig
    print('Plotting on closed figure works (Axes object still valid)')
except Exception as e:
    print('Error:', e)

# EXCEPTION: Accessing non-existent attribute
try:
    ax.nonexistent_method()
except AttributeError as e:
    print('AttributeError:', e)
````
