# Figure Size, DPI & Resolution

## Definition
Understanding **figure size**, **DPI**, and **resolution** is essential for producing crisp output
whether for Jupyter notebooks, screen displays, or print publications.

### Key Concepts
| Property | Description | Default |
|----------|-------------|---------|
| `figsize` | Width × Height in **inches** | (6.4, 4.8) |
| `dpi` | Dots per inch (pixels = inches × dpi) | 100 |
| Pixel size | `figsize[0]*dpi × figsize[1]*dpi` | 640×480 |

### Recommended DPI by Use Case
| Use Case | DPI | Notes |
|----------|-----|-------|
| Jupyter/Screen | 100–150 | Default is fine |
| Web (PNG) | 150–200 | Crisp on HiDPI |
| Print / PDF | 300 | Standard for publications |
| Poster | 600 | Very high resolution |

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)

# Default figure size
fig1, ax1 = plt.subplots()   # default figsize=(6.4, 4.8), dpi=100
ax1.plot(x, np.sin(x), color='steelblue')
ax1.set_title(f'Default size: {fig1.get_size_inches()} inches @ {fig1.dpi} dpi')
ax1.set_title(f'Pixel dims: {int(fig1.get_figwidth()*fig1.dpi)} x {int(fig1.get_figheight()*fig1.dpi)}')
plt.show()
```

```python
# Custom figure sizes
sizes = [(4, 3), (8, 3), (6, 6), (12, 4)]
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

for ax, size in zip(axes.flat, sizes):
    f_inner, a_inner = plt.subplots(figsize=size)
    a_inner.plot(x, np.sin(x), 'steelblue')
    a_inner.set_title(f'figsize={size}')
    plt.close(f_inner)
    # Show size in parent
    ax.text(0.5, 0.5, f'figsize={size}\n{size[0]*100}x{size[1]*100} px @ 100dpi',
            ha='center', va='center', fontsize=12, transform=ax.transAxes,
            bbox=dict(boxstyle='round', facecolor='lightblue'))
    ax.set_axis_off()

fig.suptitle('Figure Size Options', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# DPI comparison
fig_lo, ax = plt.subplots(figsize=(4, 3), dpi=72)
ax.plot(x, np.sin(x), 'steelblue', linewidth=2)
ax.set_title(f'72 DPI — {int(4*72)}×{int(3*72)} px')
plt.savefig('/tmp/low_dpi.png', dpi=72)
plt.show()

fig_hi, ax = plt.subplots(figsize=(4, 3), dpi=300)
ax.plot(x, np.sin(x), 'steelblue', linewidth=2)
ax.set_title(f'300 DPI — {int(4*300)}×{int(3*300)} px')
plt.savefig('/tmp/high_dpi.png', dpi=300)
plt.show()
print('Both saved. High DPI file will be much larger.')
```

```python
# rc_context for temporary size changes
with plt.rc_context({'figure.figsize': (10, 3), 'figure.dpi': 120}):
    fig, ax = plt.subplots()
    ax.plot(x, np.sin(x), color='tomato', linewidth=2)
    ax.set_title(f'Inside rc_context: {fig.get_size_inches()} @ {fig.dpi} dpi')
    plt.show()

# Outside rc_context — default restored
fig2, ax2 = plt.subplots()
ax2.set_title(f'Outside rc_context: {fig2.get_size_inches()} @ {fig2.dpi} dpi')
plt.show()
plt.close('all')
```

```python
# Changing size after creation
fig, ax = plt.subplots()
ax.plot(x, np.sin(x))
print('Before resize:', fig.get_size_inches())

fig.set_size_inches(10, 3)
print('After resize:', fig.get_size_inches())
ax.set_title('Resized after creation (10x3 inches)')
plt.show()
```
