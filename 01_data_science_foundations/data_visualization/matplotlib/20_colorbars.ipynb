# Colorbars in Matplotlib

## Key Concepts
- Binding colorbars to `ScalarMappable` objects (`imshow`, `scatter`, `contourf`).
- Positioning, orientation, shrink, aspect, tick formatting, and extension arrows (`extend='both'`).

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import FormatStrFormatter

# 1. Customizing Colorbars
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Vertical colorbar with extend & label
im1 = ax1.imshow(Z, cmap='viridis', extent=[-3, 3, -3, 3], vmin=-0.8, vmax=0.8)
cbar1 = fig.colorbar(im1, ax=ax1, orientation='vertical', extend='both', shrink=0.85)
cbar1.set_label('Energy Potential (eV)', fontsize=11, fontweight='bold')
ax1.set_title('Vertical Colorbar with Bounds Clamping', fontweight='bold')

# Horizontal colorbar with custom tick formatting
im2 = ax2.imshow(Z, cmap='magma', extent=[-3, 3, -3, 3])
cbar2 = fig.colorbar(im2, ax=ax2, orientation='horizontal', pad=0.15, aspect=30)
cbar2.set_label('Activation Magnitude', fontsize=11)
cbar2.ax.xaxis.set_major_formatter(FormatStrFormatter('%.2f'))
ax2.set_title('Horizontal Colorbar with Precision Formatting', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# 2. Shared Colorbar across Multiple Subplots
fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)
vmin, vmax = -2, 2

for i, ax in enumerate(axes):
    data = np.random.randn(20, 20) + (i - 1)
    im = ax.imshow(data, cmap='coolwarm', vmin=vmin, vmax=vmax)
    ax.set_title(f'Condition {i+1}')

# Single unified colorbar for all 3 subplots
fig.colorbar(im, ax=axes.ravel().tolist(), orientation='horizontal', fraction=0.07, pad=0.2, label='Normalized Response')
plt.show()
```
