# Line Plots

## Definition
**Line plots** connect data points with lines and are the most common plot type.
They are ideal for showing **trends over time** or **continuous functions**.

### When to Use
- Time series data
- Training/validation loss curves
- Mathematical functions
- Trend analysis

### Key Parameters of `ax.plot()`
| Parameter | Description |
|-----------|-------------|
| `color` | Line color |
| `linewidth` / `lw` | Line thickness |
| `linestyle` / `ls` | Line style |
| `marker` | Marker style at each data point |
| `markersize` / `ms` | Marker size |
| `alpha` | Transparency |
| `label` | Legend label |
| `zorder` | Drawing order (higher = on top) |

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 4*np.pi, 300)

# Basic line plot
fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(x, np.sin(x),   color='steelblue', linewidth=2, label='sin(x)')
ax.plot(x, np.cos(x),   color='tomato',    linewidth=2, linestyle='--', label='cos(x)')
ax.plot(x, np.sin(2*x), color='seagreen',  linewidth=1.5, linestyle=':', label='sin(2x)')

ax.set_title('Basic Line Plot', fontsize=14)
ax.set_xlabel('x (radians)'); ax.set_ylabel('y')
ax.legend(fontsize=12)
ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
# Line plot with markers (step-by-step data)
n = 20
x_pts = np.linspace(0, 2*np.pi, n)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# Only line
axes[0].plot(x_pts, np.sin(x_pts), color='steelblue', linewidth=2)
axes[0].set_title('Line only')

# Line + markers
axes[1].plot(x_pts, np.sin(x_pts), color='steelblue', linewidth=2,
             marker='o', markersize=8, markerfacecolor='white',
             markeredgecolor='steelblue', markeredgewidth=2)
axes[1].set_title('Line + markers')

# Only markers (scatter-like)
axes[2].plot(x_pts, np.sin(x_pts), color='tomato', linewidth=0,
             marker='D', markersize=10)
axes[2].set_title('Markers only (no line)')

for ax in axes:
    ax.grid(True, alpha=0.3)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
# Multiple series — training/validation curves (ML use case)
epochs = np.arange(1, 51)
rng = np.random.default_rng(42)

train_loss = np.exp(-epochs/15) + rng.normal(0, 0.02, 50)
val_loss   = np.exp(-epochs/20) + 0.05 + rng.normal(0, 0.03, 50)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(epochs, train_loss, color='steelblue', linewidth=2, label='Train Loss')
ax.plot(epochs, val_loss,   color='tomato',    linewidth=2, linestyle='--', label='Val Loss')

# Shade the gap (overfitting region)
ax.fill_between(epochs, train_loss, val_loss,
                where=(val_loss > train_loss),
                alpha=0.2, color='tomato', label='Overfitting gap')

ax.axvline(x=30, color='gray', linestyle=':', linewidth=1.5)
ax.text(31, 0.3, 'Early stopping?', fontsize=10, color='gray')

ax.set_title('Training vs Validation Loss Curve', fontsize=14)
ax.set_xlabel('Epoch'); ax.set_ylabel('Loss')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Step plot (for categorical/discrete data)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

x_step = np.arange(10)
y_step = np.random.randint(1, 10, 10)

axes[0].step(x_step, y_step, color='steelblue', linewidth=2, where='post')
axes[0].set_title("step() plot (where='post')")
axes[0].grid(True, alpha=0.3)

# Fill area under line
axes[1].plot(x_step, y_step, 'o-', color='tomato', linewidth=2)
axes[1].fill_between(x_step, y_step, alpha=0.3, color='tomato')
axes[1].fill_between(x_step, y_step, 5, where=(y_step > 5),
                     alpha=0.4, color='seagreen', label='Above 5')
axes[1].axhline(5, color='gray', linestyle='--')
axes[1].legend()
axes[1].set_title('fill_between()')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
