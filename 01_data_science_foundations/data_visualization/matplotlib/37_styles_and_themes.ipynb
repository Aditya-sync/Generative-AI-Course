# Styles & Themes in Matplotlib

## Built-in Matplotlib Styles
- Exploring `plt.style.available`
- Comparing `seaborn-v0_8-whitegrid`, `ggplot`, `dark_background`, and `bmh`.

```python
import matplotlib.pyplot as plt
import numpy as np

styles = ['seaborn-v0_8-whitegrid', 'ggplot', 'dark_background', 'bmh']
x = np.linspace(0, 2 * np.pi, 100)

fig = plt.figure(figsize=(14, 8))

for idx, style in enumerate(styles, 1):
    with plt.style.context(style):
        ax = fig.add_subplot(2, 2, idx)
        ax.plot(x, np.sin(x), label='sin(x)', lw=2)
        ax.plot(x, np.cos(x), label='cos(x)', lw=2)
        ax.set_title(f'Style: "{style}"', fontsize=12)
        ax.legend()

plt.tight_layout()
plt.show()
```

```python
# List all available styles
print("Available styles count:", len(plt.style.available))
print("Sample styles:", plt.style.available[:8])
```
