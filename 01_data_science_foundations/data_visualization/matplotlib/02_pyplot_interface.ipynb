# plt Pyplot Interface

## Definition
`matplotlib.pyplot` (imported as `plt`) is the **state-based interface** to Matplotlib.
It maintains an implicit current Figure and current Axes, making quick plots simple.

### pyplot vs Object-Oriented
```python
# pyplot (implicit, quick)
plt.plot([1, 2, 3])
plt.title('Quick plot')
plt.show()

# OO (explicit, recommended for serious work)
fig, ax = plt.subplots()
ax.plot([1, 2, 3])
ax.set_title('OO plot')
plt.show()
```

### Most Used pyplot Functions
| Function | Description |
|----------|-------------|
| `plt.figure()` | Create a new Figure |
| `plt.subplots()` | Create Figure + Axes |
| `plt.plot()` | Line/scatter plot |
| `plt.show()` | Display the figure |
| `plt.close()` | Close figure (free memory) |
| `plt.clf()` | Clear current figure |
| `plt.cla()` | Clear current axes |
| `plt.savefig()` | Save to file |
| `plt.gcf()` | Get current figure |
| `plt.gca()` | Get current axes |

````python
import matplotlib.pyplot as plt
import numpy as np

# Basic pyplot workflow
x = np.linspace(0, 2 * np.pi, 200)

plt.figure(figsize=(10, 5))
plt.plot(x, np.sin(x), label='sin(x)', color='steelblue', linewidth=2)
plt.plot(x, np.sin(2*x), label='sin(2x)', color='tomato', linewidth=2, linestyle='--')
plt.title('pyplot Quick Plot', fontsize=14)
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
# plt.subplots() — most common entry point
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

x = np.linspace(0, 10, 100)
ax1.plot(x, np.sin(x), color='steelblue')
ax1.set_title('Sine Wave')
ax1.set_xlabel('x')

ax2.plot(x, np.exp(-x/5) * np.cos(x), color='tomato')
ax2.set_title('Damped Oscillation')
ax2.set_xlabel('x')

fig.suptitle('Two Subplots with plt.subplots()', fontsize=13)
plt.tight_layout()
plt.show()
````

````python
# plt.gcf() and plt.gca() — get current figure/axes
plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3, 4], [1, 4, 9, 16], 'o-', color='purple')

# Access the automatically created figure and axes
current_fig = plt.gcf()
current_ax  = plt.gca()

print('Current figure:', type(current_fig).__name__, current_fig.number)
print('Current axes:', type(current_ax).__name__)
print('Current xlim:', current_ax.get_xlim())

plt.title('Using plt.gcf() and plt.gca()')
plt.show()
````

````python
# plt.clf(), plt.cla(), plt.close() — clearing and closing
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [1, 4, 9])
ax.set_title('Original Plot')
print('Figure number before close:', fig.number)

# Clear axes only (keeps figure)
plt.cla()
ax.set_title('After cla() — Axes cleared')
plt.plot([1, 2, 3], [1, 2, 3])  # replot
plt.show()

# Close figure (frees memory)
plt.close('all')  # close all open figures
print('All figures closed')
````

````python
# Multiple figures with plt.figure(num=...)
fig1 = plt.figure(1, figsize=(5, 3))
plt.plot([1, 2, 3], [1, 4, 9], color='blue')
plt.title('Figure 1')

fig2 = plt.figure(2, figsize=(5, 3))
plt.plot([1, 2, 3], [3, 2, 1], color='red')
plt.title('Figure 2')

# Switch back to figure 1
plt.figure(1)
plt.plot([1, 2, 3], [2, 4, 6], 'g--')   # add to Figure 1

plt.figure(1); plt.tight_layout(); plt.show()
plt.figure(2); plt.tight_layout(); plt.show()
plt.close('all')
````

````python
# EXCEPTION: Forgetting plt.show() in scripts
# In Jupyter: output renders inline — plt.show() is optional
# In scripts: must call plt.show() or plt.savefig() or nothing displays

# EXCEPTION: Modifying wrong axes after subplots
fig, (ax1, ax2) = plt.subplots(1, 2)
ax1.plot([1, 2, 3], color='blue')

# Common mistake: plt.title() sets title on CURRENT axes (may not be ax1)
plt.title('This goes to the CURRENT active axes, not necessarily ax1!')
ax_current = plt.gca()
print('Current axes is ax2?', ax_current is ax2)  # True after second subplot created
plt.close('all')
````
