# Hyperparameter & Experiment Visualization

## Common Visuals
- Validation Score vs Hyperparameter curve with cross-validation confidence bounds.
- 2D Hyperparameter Grid Search Heatmap (e.g. Learning Rate vs Batch Size).

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. 2D Hyperparameter Grid Heatmap (C vs Gamma in SVM, or LR vs Depth in GBDT)
learning_rates = [0.001, 0.01, 0.05, 0.1, 0.2]
max_depths = [3, 4, 5, 6, 8, 10]

# Simulated F1 scores matrix
np.random.seed(42)
scores = np.array([
    [0.72, 0.78, 0.81, 0.83, 0.80],
    [0.75, 0.82, 0.86, 0.87, 0.83],
    [0.76, 0.85, 0.91, 0.89, 0.84],
    [0.77, 0.87, 0.93, 0.90, 0.82],
    [0.75, 0.86, 0.90, 0.88, 0.79],
    [0.72, 0.83, 0.88, 0.85, 0.75]
])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

im = ax1.imshow(scores, cmap='YlGnBu')
ax1.set_xticks(range(len(learning_rates)))
ax1.set_yticks(range(len(max_depths)))
ax1.set_xticklabels(learning_rates)
ax1.set_yticklabels(max_depths)
ax1.set_xlabel('Learning Rate', fontweight='bold')
ax1.set_ylabel('Max Depth', fontweight='bold')
ax1.set_title('Grid Search CV: F1-Score Landscape', fontweight='bold')

for i in range(len(max_depths)):
    for j in range(len(learning_rates)):
        val = scores[i, j]
        color = "white" if val > 0.85 else "black"
        ax1.text(j, i, f'{val:.2f}', ha='center', va='center', color=color, fontweight='bold')
fig.colorbar(im, ax=ax1, shrink=0.8, label='Validation F1 Score')

# Highlight best parameter combination
best_i, best_j = np.unravel_index(np.argmax(scores), scores.shape)
ax1.scatter(best_j, best_i, s=200, facecolors='none', edgecolors='red', lw=3, label='Best Model')
ax1.legend(loc='upper left')

# 2. 1D Parameter Sensitivity Curve with Std Bounds
param_range = np.logspace(-4, 0, 10)
train_scores = 0.98 - 0.2 * np.exp(-param_range * 10)
test_scores = 0.70 + 0.22 * (param_range**0.3) / (1 + param_range**0.3) - 0.1 * (param_range > 0.1)

ax2.semilogx(param_range, train_scores, 'o-', color='crimson', label='Training Score')
ax2.semilogx(param_range, test_scores, 's-', color='navy', label='Validation Score')
ax2.set_xlabel('Regularization Parameter C (Log Scale)', fontweight='bold')
ax2.set_ylabel('ROC-AUC Score', fontweight='bold')
ax2.set_title('Hyperparameter Sensitivity Curve (C)', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3, which='both')

plt.tight_layout()
plt.show()
```
