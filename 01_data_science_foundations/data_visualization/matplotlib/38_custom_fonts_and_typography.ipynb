# Custom Fonts & Typography in Matplotlib

## Controlling Text Properties & Math Expressions
- Using LaTeX math formatting: `$r'\alpha_i = \frac{\exp(e_i)}{\sum_j \exp(e_j)}'$`
- Controlling font families (sans-serif, serif, monospace, cursive) and font weights.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(10, 6))

x = np.linspace(0, 10, 100)
y = np.exp(-x / 3) * np.sin(2 * np.pi * x / 2)

ax.plot(x, y, color='#2c3e50', lw=2, label=r'Damped Oscillator: $y(t) = e^{-\gamma t} \sin(\omega t)$')

# Rich LaTeX annotations
ax.set_title(r'Damped Harmonic Motion & Attention Weights: $lpha_t = \mathrm{softmax}(QK^T / \sqrt{d_k})$',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel(r'Time Step ($t \in [0, 10]$)', fontsize=12, family='serif')
ax.set_ylabel(r'Amplitude ($\Psi$)', fontsize=12, family='serif')

# Monospace stats annotation box
text_str = "\n".join((
    r"$\mu = 0.00$",
    r"$\sigma = 0.35$",
    r"$\lambda_{\mathrm{decay}} = 0.333$",
))
props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)
ax.text(0.75, 0.90, text_str, transform=ax.transAxes, fontsize=11, family='monospace', verticalalignment='top', bbox=props)

ax.legend(loc='lower right', fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
