# Customizing rcParams in Matplotlib

## Control Global Defaults Dynamically
- Figure size, DPI, font sizes, grid styling, colors, and line widths across entire notebooks.

```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np

# 1. Inspecting current rcParams
print("Default DPI:", mpl.rcParams['figure.dpi'])
print("Default figure size:", mpl.rcParams['figure.figsize'])
print("Default font family:", mpl.rcParams['font.sans-serif'][:3])

# 2. Applying contextual rcParams using plt.rc_context
custom_settings = {
    'figure.figsize': (10, 5),
    'figure.dpi': 120,
    'axes.titlesize': 14,
    'axes.labelsize': 12,
    'lines.linewidth': 2.5,
    'lines.markersize': 8,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'grid.alpha': 0.35,
    'grid.linestyle': '--'
}

x = np.linspace(0, 10, 50)
y = np.cos(x)

with plt.rc_context(custom_settings):
    fig, ax = plt.subplots()
    ax.plot(x, y, marker='o', label='Cos(x) with Contextual rcParams')
    ax.set_title('Custom Look via rc_context')
    ax.set_xlabel('Timestamp')
    ax.set_ylabel('Amplitude')
    ax.grid(True)
    ax.legend()
    plt.show()

# After context ends, original defaults are safely restored
print("Restored figure size:", mpl.rcParams['figure.figsize'])
```
