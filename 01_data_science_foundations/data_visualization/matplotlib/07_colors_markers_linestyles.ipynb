# Colors, Markers, Linestyles & Line Properties

## Definition
Controlling the visual appearance of plot elements is key for clarity and professional aesthetics.

### Colors
| Specification | Example |
|--------------|---------|
| Named CSS color | `'steelblue'`, `'tomato'`, `'seagreen'` |
| Hex code | `'#2196F3'` |
| RGB tuple | `(0.5, 0.1, 0.8)` |
| RGBA tuple | `(0.5, 0.1, 0.8, 0.5)` |
| Shorthand | `'r'`, `'g'`, `'b'`, `'k'`, `'w'` |
| Grayscale | `'0.7'` (0=black, 1=white) |

### Linestyles
| Style | Code |
|-------|------|
| Solid | `'-'` |
| Dashed | `'--'` |
| Dotted | `':'` |
| Dash-dot | `'-.'` |
| None | `''` or `'None'` |

### Markers
`'o'` circle, `'s'` square, `'^'` triangle, `'D'` diamond,
`'+'` plus, `'x'` cross, `'*'` star, `'.'` point, `'|'` vline

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 50)

# Color options
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

color_examples = [
    ('steelblue', 'Named CSS'),
    ('#E91E63', 'Hex code'),
    ((0.2, 0.7, 0.2), 'RGB tuple'),
    ((1.0, 0.5, 0.0, 0.7), 'RGBA (alpha=0.7)'),
    ('r', 'Shorthand r'),
    ('0.6', 'Grayscale 0.6'),
]

for i, (color, label) in enumerate(color_examples):
    axes[0].plot(x, np.sin(x) + i*2.5, color=color, linewidth=2, label=label)

axes[0].set_title('Color Options')
axes[0].legend(fontsize=9, loc='upper left')
axes[0].set_yticks([])

# Linestyle options
linestyles = ['-', '--', ':', '-.', (0, (5, 2, 1, 2)), (0, (3, 1, 1, 1, 1, 1))]
labels = ['solid -', 'dashed --', 'dotted :', 'dashdot -.', 'custom 1', 'custom 2']
for i, (ls, label) in enumerate(zip(linestyles, labels)):
    axes[1].plot(x, np.sin(x) + i*2.5, linestyle=ls, color='steelblue',
                 linewidth=2, label=label)
axes[1].set_title('Linestyle Options')
axes[1].legend(fontsize=9, loc='upper left')
axes[1].set_yticks([])

plt.tight_layout()
plt.show()
```

```python
# Marker styles
markers = ['o', 's', '^', 'D', '+', 'x', '*', '.', 'v', 'p', 'h', '8']
labels  = ['circle', 'square', 'triangle_up', 'diamond', 'plus', 'cross',
           'star', 'point', 'triangle_down', 'pentagon', 'hexagon', 'octagon']

x_pts = np.linspace(0, 1, 10)
fig, ax = plt.subplots(figsize=(13, 7))

colors = plt.cm.tab20(np.linspace(0, 1, len(markers)))
for i, (marker, label, color) in enumerate(zip(markers, labels, colors)):
    ax.plot(x_pts, [i]*10, marker=marker, markersize=10, color=color,
            linestyle='-', linewidth=1, label=label, markerfacecolor=color,
            markeredgecolor='black', markeredgewidth=0.5)

ax.set_title('Marker Styles', fontsize=14)
ax.set_yticks(range(len(markers)))
ax.set_yticklabels(labels)
ax.set_xlim(-0.1, 1.1)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Line properties: linewidth, alpha, markerfacecolor, markeredgecolor
fig, ax = plt.subplots(figsize=(10, 6))
x_sparse = np.linspace(0, 2*np.pi, 15)

ax.plot(x_sparse, np.sin(x_sparse),
        color='steelblue', linewidth=3,
        marker='o', markersize=10,
        markerfacecolor='white', markeredgecolor='steelblue', markeredgewidth=2,
        label='Hollow markers', alpha=0.9)

ax.plot(x_sparse, np.cos(x_sparse),
        color='tomato', linewidth=2, linestyle='--',
        marker='s', markersize=9,
        markerfacecolor='tomato', markeredgecolor='darkred',
        label='Filled markers', alpha=0.8)

ax.set_title('Line Properties Demo')
ax.legend(fontsize=12)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Format string shorthand: 'color+marker+linestyle'
fig, ax = plt.subplots(figsize=(9, 5))
x_s = np.linspace(0, 2*np.pi, 20)

# format: [color][marker][linestyle]
ax.plot(x_s, np.sin(x_s),    'bo-',  label='blue, circle, solid')
ax.plot(x_s, np.cos(x_s),    'r^--', label='red, triangle, dashed')
ax.plot(x_s, np.sin(2*x_s),  'gs:',  label='green, square, dotted')
ax.plot(x_s, -np.cos(x_s),   'kD-.', label='black, diamond, dash-dot')

ax.set_title('Format String Shorthand: [color][marker][linestyle]')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# EXCEPTION: Invalid color specification
try:
    fig, ax = plt.subplots()
    ax.plot([1,2,3], color='invalidcolor')
    plt.show()
except ValueError as e:
    print('ValueError:', e)
    plt.close()

# EXCEPTION: Invalid marker
try:
    fig, ax = plt.subplots()
    ax.plot([1,2,3], marker='Z')
    plt.show()
except Exception as e:
    print(type(e).__name__+':', e)
    plt.close()
```
