# Violin Plots

## Definition
**Violin plots** combine a **box plot** with a **kernel density estimate (KDE)** to show the
full distribution shape. Wider sections have more data points.

### Violin vs Box Plot
| Feature | Box Plot | Violin Plot |
|---------|---------|-------------|
| Shows median, IQR | ✅ | ✅ (inner box) |
| Shows distribution shape | ❌ | ✅ |
| Shows multimodality | ❌ | ✅ |
| Compact | ✅ | ❌ |

### Key Parameters of `ax.violinplot()`
| Parameter | Description |
|-----------|-------------|
| `showmeans` | Show mean marker |
| `showmedians` | Show median line |
| `showextrema` | Show min/max whiskers |
| `bw_method` | KDE bandwidth |
| `vert` | Vertical (True) or horizontal |
| `widths` | Width of each violin |

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)

# Create bimodal and unimodal distributions
data = [
    np.concatenate([rng.normal(-2, 0.5, 200), rng.normal(2, 0.5, 200)]),  # bimodal
    rng.normal(0, 1.5, 400),     # unimodal wide
    rng.exponential(1, 400),     # right-skewed
    rng.uniform(-3, 3, 400),     # uniform
]
labels = ['Bimodal', 'Wide Normal', 'Skewed', 'Uniform']

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Violin plot
vp = axes[0].violinplot(data, positions=range(1, 5),
                         showmeans=True, showmedians=True, showextrema=True)

# Color each violin
colors = ['steelblue', 'tomato', 'seagreen', 'darkorange']
for body, color in zip(vp['bodies'], colors):
    body.set_facecolor(color)
    body.set_alpha(0.7)

for part in ['cbars', 'cmins', 'cmaxes', 'cmedians', 'cmeans']:
    if part in vp:
        vp[part].set_color('black')
        vp[part].set_linewidth(1.5)

axes[0].set_xticks(range(1, 5))
axes[0].set_xticklabels(labels, fontsize=11)
axes[0].set_title('Violin Plot — Distribution Shapes', fontsize=13)
axes[0].set_ylabel('Value')
axes[0].grid(True, axis='y', alpha=0.3)

# Box vs violin comparison
axes[1].violinplot(data, positions=[1,2,3,4], showmedians=True)
axes[1].boxplot(data, positions=[1,2,3,4], widths=0.1,
                medianprops=dict(color='red', linewidth=2))
axes[1].set_xticks(range(1, 5))
axes[1].set_xticklabels(labels, fontsize=10)
axes[1].set_title('Violin + Box Plot Overlaid', fontsize=13)
axes[1].grid(True, axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Half violin plot — comparing two groups
fig, ax = plt.subplots(figsize=(9, 6))

n = 300
group_a = rng.normal(5, 1.5, n)
group_b = rng.normal(6.5, 2.0, n)
positions = [1, 2, 3, 4]

data_a = [rng.normal(5+i*0.3, 1.5, n) for i in range(4)]
data_b = [rng.normal(6.5+i*0.2, 2.0, n) for i in range(4)]

vp_a = ax.violinplot(data_a, positions=positions, showmedians=True, widths=0.4)
vp_b = ax.violinplot(data_b, positions=[p+0.5 for p in positions], showmedians=True, widths=0.4)

for body in vp_a['bodies']:
    body.set_facecolor('steelblue'); body.set_alpha(0.7)
for body in vp_b['bodies']:
    body.set_facecolor('tomato'); body.set_alpha(0.7)

ax.set_xticks([p+0.25 for p in positions])
ax.set_xticklabels(['Q1', 'Q2', 'Q3', 'Q4'])

# Manual legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor='steelblue', alpha=0.7, label='Group A'),
                   Patch(facecolor='tomato', alpha=0.7, label='Group B')]
ax.legend(handles=legend_elements, fontsize=11)
ax.set_title('Violin Plot — A/B Comparison per Quarter', fontsize=13)
ax.grid(True, axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Bandwidth effect on violin shape
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
data_bw = np.concatenate([rng.normal(-2, 0.5, 100), rng.normal(2, 0.5, 100)])

for ax, bw, title in zip(axes,
    [0.05, 0.5, 1.5],
    ['Bandwidth=0.05 (overfit)', 'Bandwidth=0.5 (good)', 'Bandwidth=1.5 (oversmooth)']):
    vp = ax.violinplot([data_bw], bw_method=bw, showmedians=True)
    vp['bodies'][0].set_facecolor('steelblue')
    vp['bodies'][0].set_alpha(0.7)
    ax.set_title(title, fontsize=10)
    ax.set_xticks([])
    ax.grid(True, axis='y', alpha=0.3)

plt.suptitle('Effect of KDE Bandwidth on Violin Shape', fontsize=13)
plt.tight_layout()
plt.show()
```
