# Decision Boundary Visualization in Matplotlib

## Approach
1. Create a dense 2D coordinate grid using `np.meshgrid`.
2. Generate model predictions over all grid points.
3. Draw boundaries using `contourf` (for probability confidence) and `contour` (for threshold boundary).

```python
import matplotlib.pyplot as plt
import numpy as np

# Synthetic 2D data: 2 non-linearly separable classes
np.random.seed(42)
N = 150
X0 = np.random.randn(N, 2) * 0.8 + np.array([-1, -1])
X1 = np.random.randn(N, 2) * 0.8 + np.array([1, 1])
X = np.vstack([X0, X1])
y = np.array([0]*N + [1]*N)

# Simulated logistic model decision function: z = w1*x1 + w2*x2 + non-linear term
def predict_proba(grid_x, grid_y):
    z = 1.2 * grid_x + 1.4 * grid_y - 0.4 * (grid_x**2) + 0.3 * (grid_y**2)
    return 1 / (1 + np.exp(-z))

# 1. Coordinate grid
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))
probs = predict_proba(xx, yy)

# 2. Plotting
fig, ax = plt.subplots(figsize=(8, 6))

# Filled probability contour
contour = ax.contourf(xx, yy, probs, levels=25, cmap='RdBu_r', alpha=0.7)
cbar = fig.colorbar(contour, ax=ax)
cbar.set_label('P(Class = 1)', fontweight='bold')

# Decision threshold boundary line (P = 0.5)
ax.contour(xx, yy, probs, levels=[0.5], colors='black', linewidths=2.5, linestyles='--')

# Data scatter points
ax.scatter(X0[:, 0], X0[:, 1], c='blue', edgecolors='k', s=50, label='Class 0')
ax.scatter(X1[:, 0], X1[:, 1], c='red', edgecolors='k', s=50, label='Class 1')

ax.set_title('Classifier Non-Linear Decision Boundary (P = 0.5 Threshold)', fontsize=13, fontweight='bold')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.legend(loc='lower right')
ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```
