# Count Plots in Seaborn (countplot)

## Frequency of Categorical Variables
- Dedicated function for counting occurrences (like a histogram for discrete bins).
- Supports horizontal counts and hue breakdowns.

```python
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset('titanic')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Vertical count plot
sns.countplot(data=titanic, x='class', hue='survived', palette='Set1', ax=ax1)
ax1.set_title('Passenger Class Counts by Survival', fontweight='bold')
ax1.set_ylabel('Number of Passengers')

# Horizontal count plot
sns.countplot(data=titanic, y='embark_town', hue='class', palette='Set2', ax=ax2)
ax2.set_title('Embarkation Town by Ticket Class', fontweight='bold')
ax2.set_xlabel('Passenger Count')

plt.tight_layout()
plt.show()
```
