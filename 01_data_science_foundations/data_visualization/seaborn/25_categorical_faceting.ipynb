# Categorical Faceting in Seaborn (catplot)

## Figure-Level Categorical Dashboard
- `sns.catplot()` can generate any categorical plot (`box`, `violin`, `bar`, `count`, `strip`, `swarm`) across multiple rows and columns.

```python
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset('titanic')

# Faceted Catplot: Survival rates across Class, Sex, and Embarkation Town
g = sns.catplot(
    data=titanic,
    x='class', y='survived',
    hue='sex', col='embark_town',
    kind='bar', palette='Set1',
    height=4, aspect=0.9
)

g.set_axis_labels('Passenger Class', 'Survival Probability')
g.fig.suptitle('Multi-Faceted Survival Rate Analysis (catplot)', y=1.04, fontweight='bold', fontsize=14)
plt.show()
```
