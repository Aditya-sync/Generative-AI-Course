# Multi-Group Distribution Comparison

## Methods Compared
- Boxplot vs Violinplot vs KDE overlay vs Faceted Histograms.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# 1. Overlay KDE
sns.kdeplot(data=tips, x='total_bill', hue='day', common_norm=False, fill=True, alpha=0.2, ax=axes[0])
axes[0].set_title('KDE Multi-Group Overlay', fontweight='bold')

# 2. Box Plot
sns.boxplot(data=tips, x='day', y='total_bill', palette='Set3', ax=axes[1])
axes[1].set_title('Box Plot Multi-Group Comparison', fontweight='bold')

# 3. Violin Plot with inner quartiles
sns.violinplot(data=tips, x='day', y='total_bill', inner='quartile', palette='Pastel1', ax=axes[2])
axes[2].set_title('Violin Plot Multi-Group Comparison', fontweight='bold')

plt.tight_layout()
plt.show()
```
