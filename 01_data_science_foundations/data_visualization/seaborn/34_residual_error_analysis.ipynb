# Residual & Error Analysis in Seaborn

## Regression Diagnostic Tools
- `sns.residplot()`: Automatically plots residuals of a linear regression.
- Residual distribution check using `sns.kdeplot()`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
x = np.random.uniform(10, 100, 200)
# True non-linear relationship
y = 3 * x + 0.05 * (x**2) + np.random.normal(0, 15, 200)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. residplot detecting non-linearity
sns.residplot(x=x, y=y, lowess=True, color='teal',
              scatter_kws={'alpha': 0.6}, line_kws={'color': 'red', 'lw': 2}, ax=ax1)
ax1.set_title('sns.residplot with LOWESS Trend Curve', fontweight='bold')
ax1.set_xlabel('Feature X')
ax1.set_ylabel('Residual')
ax1.grid(True, alpha=0.3)

# 2. Residual Distribution (Normality Check)
residuals = y - (np.poly1d(np.polyfit(x, y, 1)))(x)
sns.kdeplot(residuals, fill=True, color='purple', ax=ax2)
ax2.axvline(0, color='red', linestyle='--', lw=2)
ax2.set_title('Residual Density vs Zero Baseline', fontweight='bold')
ax2.set_xlabel('Residual Error')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
