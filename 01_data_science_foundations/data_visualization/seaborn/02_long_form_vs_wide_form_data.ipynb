# Long-Form vs Wide-Form Data in Seaborn

## Concepts
- **Long-Form (Tidy Data)**: Each variable is a column, each observation is a row. Ideal for semantic mapping (`hue`, `style`, `size`).
- **Wide-Form**: Data is spread across columns (e.g. time steps or feature values as columns).
- Transforming between formats using `pandas.melt()` and `pandas.pivot()`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 1. Wide-form data representation
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=50)
wide_df = pd.DataFrame({
    'Model_A_Loss': np.exp(-np.linspace(0, 3, 50)) + np.random.normal(0, 0.05, 50),
    'Model_B_Loss': np.exp(-np.linspace(0, 2.5, 50)) + np.random.normal(0, 0.05, 50),
    'Model_C_Loss': np.exp(-np.linspace(0, 4, 50)) + np.random.normal(0, 0.05, 50)
}, index=dates)

print("Wide-Form Shape:", wide_df.shape)
print(wide_df.head(2))

# Plotting wide-form directly
fig, ax = plt.subplots(figsize=(8, 4))
sns.lineplot(data=wide_df, ax=ax)
ax.set_title('Seaborn Plotting Wide-Form DataFrame Directly', fontweight='bold')
ax.set_ylabel('Loss Value')
plt.show()
```

```python
# 2. Converting to Long-Form (Tidy) using melt
long_df = wide_df.reset_index().melt(
    id_vars=['index'],
    value_vars=['Model_A_Loss', 'Model_B_Loss', 'Model_C_Loss'],
    var_name='Model',
    value_name='Loss'
).rename(columns={'index': 'Date'})

print("Long-Form Shape:", long_df.shape)
print(long_df.head(4))

# Plotting long-form with rich semantics
fig, ax = plt.subplots(figsize=(9, 4))
sns.lineplot(data=long_df, x='Date', y='Loss', hue='Model', style='Model', markers=True, dashes=False, ax=ax)
ax.set_title('Plotting Long-Form Data (Semantic Hue & Style)', fontweight='bold')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
