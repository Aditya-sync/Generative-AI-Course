# Violin Plots in Seaborn (violinplot)

## Hybrid Box Plot + KDE
- Reveals multi-modality hidden inside box plots.
- Asymmetric split violins for binary categories (`split=True`).

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Standard Violin with inner quartile markers
sns.violinplot(data=tips, x='day', y='total_bill', inner='quartile', palette='muted', ax=ax1)
ax1.set_title('Violin Plot with Quartile Dividers', fontweight='bold')

# 2. Split Violin Plot for Direct Binary Comparison
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', split=True,
               inner='stick', palette={'Male': '#3498db', 'Female': '#e74c3c'}, ax=ax2)
ax2.set_title('Split Violin Plot (Male vs Female)', fontweight='bold')

plt.tight_layout()
plt.show()
```
