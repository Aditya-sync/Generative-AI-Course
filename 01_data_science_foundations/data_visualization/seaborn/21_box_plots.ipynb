# Box Plots in Seaborn (boxplot)

## Five-Number Summary
- Min, Q1 (25th), Median (50th), Q3 (75th), Max ($1.5 \times \text{IQR}$).
- Outlier detection and notch confidence intervals.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Standard Box Plot with Hue
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker', palette='vlag', ax=ax1)
ax1.set_title('Box Plot Split by Smoker Status', fontweight='bold')

# Notched Box Plot (Confidence interval around median)
sns.boxplot(data=tips, x='day', y='total_bill', notch=True, palette='Pastel1', ax=ax2)
ax2.set_title('Notched Box Plot (Median Significance Test)', fontweight='bold')

plt.tight_layout()
plt.show()
```
