# Seaborn Architecture: Figure-Level vs Axes-Level Functions

## Core Distinction
Seaborn provides two tiers of plotting functions:
1. **Axes-level functions**: (e.g. `sns.histplot`, `sns.scatterplot`, `sns.boxplot`) draw onto a specific matplotlib `Axes` object via the `ax=` parameter.
2. **Figure-level functions**: (e.g. `sns.displot`, `sns.relplot`, `sns.catplot`) control the entire figure via a `FacetGrid` wrapper, enabling faceting across rows and columns.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load built-in sample dataset
tips = sns.load_dataset('tips')
print("Tips dataset head:")
print(tips.head(3))

# 1. Axes-level function: Placed inside custom subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', ax=ax1)
ax1.set_title('Axes-Level: sns.scatterplot on ax1', fontweight='bold')

sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2', ax=ax2)
ax2.set_title('Axes-Level: sns.boxplot on ax2', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# 2. Figure-level function: sns.relplot (creates its own FacetGrid)
g = sns.relplot(
    data=tips,
    x='total_bill', y='tip',
    hue='smoker', col='time', row='sex',
    kind='scatter', height=3.5, aspect=1.1
)
g.fig.suptitle('Figure-Level: sns.relplot (Row=Sex, Col=Time)', y=1.03, fontweight='bold')
plt.show()
print("Figure-level object type:", type(g).__name__)
```

```python
# Common Pitfall: Passing ax to a figure-level function
try:
    fig, ax = plt.subplots()
    sns.relplot(data=tips, x='total_bill', y='tip', ax=ax)
except TypeError as e:
    print('Caught TypeError (figure-level functions do not accept ax=):', e)
plt.close(fig)
```
