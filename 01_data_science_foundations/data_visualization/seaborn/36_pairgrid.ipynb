# PairGrid in Seaborn

## Generalized Subplot Grid for Variable Pairs
- Unlike `pairplot` which is a convenience function, `PairGrid` gives full programmatic control over:
  - `g.map_upper()`
  - `g.map_lower()`
  - `g.map_diag()`
  - `g.map_offdiag()`

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')

g = sns.PairGrid(iris, hue='species', palette='Dark2', diag_sharey=False)

# Lower triangle: scatter with alpha
g.map_lower(sns.scatterplot, s=35, alpha=0.7)

# Upper triangle: KDE density contours
g.map_upper(sns.kdeplot, alpha=0.5)

# Diagonal: KDE univariate plots
g.map_diag(sns.kdeplot, fill=True, common_norm=False, alpha=0.3)

g.add_legend(title='Iris Species')
g.fig.suptitle('PairGrid Custom Diagonal & Triangular Layout', y=1.02, fontweight='bold', fontsize=13)
plt.show()
```
