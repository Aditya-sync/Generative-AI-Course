# Pairwise Feature Analysis in Seaborn

## PairGrid Customization
- Diagonal: univariate histograms/KDE.
- Lower triangle: scatter / regression plots.
- Upper triangle: 2D KDE contour maps.

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()
cols = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm']

g = sns.PairGrid(penguins, vars=cols, hue='species', palette='Set1', diag_sharey=False)

# Map distinct plots to each matrix section
g.map_upper(sns.kdeplot, alpha=0.6)
g.map_lower(sns.scatterplot, s=30, alpha=0.7)
g.map_diag(sns.histplot, multiple='stack', element='step')

g.add_legend(title='Species')
g.fig.suptitle('Advanced PairGrid with Distinct Section Mappings', y=1.02, fontweight='bold')
plt.show()
```
