# Feature-Target Relationships in Seaborn

## Analysis
- Evaluating predictive power of individual features against target variables.
- Using regression lines, box splits, and probability distributions.

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Feature vs Continuous Target with Group Separation
sns.scatterplot(
    data=penguins,
    x='flipper_length_mm', y='body_mass_g',
    hue='species', style='sex',
    s=60, alpha=0.8, ax=ax1
)
sns.regplot(data=penguins, x='flipper_length_mm', y='body_mass_g',
            scatter=False, ax=ax1, color='gray')
ax1.set_title('Continuous Feature vs Continuous Target', fontweight='bold')

# 2. Continuous Feature vs Discrete Target (KDE Density split)
sns.kdeplot(
    data=penguins,
    x='bill_length_mm', hue='species',
    fill=True, common_norm=False, alpha=0.35, ax=ax2
)
ax2.set_title('Feature Separability across Classes', fontweight='bold')

plt.tight_layout()
plt.show()
```
