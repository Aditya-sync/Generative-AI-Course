# Outlier Visualization in Seaborn

## Identifying Anomalies
- Boxplots with fliers highlighted (`flierprops`).
- Jittered stripplots exposing leverage points.
- Mahalanobis / PCA distance scatter.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Inject outliers into dataset
np.random.seed(42)
clean = np.random.normal(50, 10, 300)
outliers = np.array([5, 9, 105, 120, 135])
vals = np.concatenate([clean, outliers])
group = ['Normal Data'] * 300 + ['Extreme Outlier'] * 5
df = pd.DataFrame({'Value': vals, 'Status': group})

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Boxplot with custom outlier marker properties
sns.boxplot(data=df, y='Value', ax=ax1, color='lightblue',
            flierprops={'marker': 'o', 'markerfacecolor': 'red', 'markersize': 8})
ax1.set_title('Boxplot Outlier Detection (1.5 x IQR)', fontweight='bold')

# 2. Stripplot + Violin
sns.violinplot(data=df, y='Value', color='wheat', inner=None, ax=ax2)
sns.stripplot(data=df, y='Value', hue='Status', palette={'Normal Data': 'blue', 'Extreme Outlier': 'red'},
              jitter=0.25, size=6, alpha=0.7, ax=ax2)
ax2.set_title('Violin + Stripplot Outlier Exposure', fontweight='bold')

plt.tight_layout()
plt.show()
```
