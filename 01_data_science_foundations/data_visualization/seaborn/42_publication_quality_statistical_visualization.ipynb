# Publication-Quality Statistical Visualization in Seaborn

## Academic Standards (Nature / Science / PNAS)
- Raincloud plot (Combining Half-Violin / KDE, Boxplot, and Raw Stripplot jitter).
- Crisp vector export and colorblind-safe palettes.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

penguins = sns.load_dataset('penguins').dropna()

# Nature / PNAS publication styling
sns.set_theme(style='ticks', font='DejaVu Sans')
fig, ax = plt.subplots(figsize=(7, 4.5), dpi=150)

# Half-violin / Box / Strip hybrid (Raincloud aesthetic)
sns.boxplot(
    data=penguins,
    x='species', y='flipper_length_mm',
    palette='colorblind',
    width=0.25,
    showcaps=True,
    boxprops={'zorder': 2, 'alpha': 0.8},
    showfliers=False,
    ax=ax
)

sns.stripplot(
    data=penguins,
    x='species', y='flipper_length_mm',
    palette='colorblind',
    size=4,
    jitter=0.15,
    alpha=0.6,
    zorder=1,
    ax=ax
)

# Statistical significance bracket annotation (Matplotlib)
x1, x2 = 0, 1
y_max = penguins['flipper_length_mm'].max() + 5
h = 3
ax.plot([x1, x1, x2, x2], [y_max, y_max + h, y_max + h, y_max], lw=1.2, c='black')
ax.text((x1 + x2) * 0.5, y_max + h + 1, '*** p < 0.001', ha='center', va='bottom', color='black', fontsize=9, fontweight='bold')

ax.set_title('Morphological Differentiation across Pygoscelis Species', fontsize=11, fontweight='bold', pad=15)
ax.set_xlabel('Species', fontsize=10, fontweight='bold')
ax.set_ylabel('Flipper Length (mm)', fontsize=10, fontweight='bold')
sns.despine(trim=True)

plt.tight_layout()
plt.savefig('/tmp/nature_style_seaborn.pdf', bbox_inches='tight')
plt.show()
```
