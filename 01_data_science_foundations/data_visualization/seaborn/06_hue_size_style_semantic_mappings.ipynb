# Hue, Size & Style Semantic Mappings

## Multi-Dimensional Encoding
Seaborn allows mapping up to 5 dimensions onto a single 2D plot:
- $X$ and $Y$ (Spatial positions)
- `hue` (Color hue)
- `size` (Marker size / stroke width)
- `style` (Marker shape / line dash pattern)

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

fig, ax = plt.subplots(figsize=(10, 6))

# Mapping 5 dimensions: bill_length (X), flipper_length (Y), species (hue), sex (style), body_mass (size)
sns.scatterplot(
    data=penguins,
    x='bill_length_mm',
    y='flipper_length_mm',
    hue='species',
    style='sex',
    size='body_mass_g',
    sizes=(30, 200),
    alpha=0.8,
    palette='Set1',
    ax=ax
)

ax.set_title('5D Data Encoding: X, Y, Hue, Style, and Size', fontsize=14, fontweight='bold')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
