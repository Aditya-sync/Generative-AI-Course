# Model Performance Visualization in Seaborn

## Benchmarking Across Models & Folds
- Cross-validation score boxplots across multiple architectures.
- Metric bar comparisons with error bounds.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Cross-validation benchmark simulation
np.random.seed(42)
models = ['LogisticReg', 'RandomForest', 'LightGBM', 'XGBoost', 'CatBoost']
records = []
for model in models:
    mean_score = {'LogisticReg': 0.78, 'RandomForest': 0.85, 'LightGBM': 0.89, 'XGBoost': 0.90, 'CatBoost': 0.91}[model]
    scores = np.random.normal(mean_score, 0.015, 10)  # 10 folds
    for s in scores:
        records.append({'Model': model, 'ROC_AUC': s})

cv_df = pd.DataFrame(records)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Boxplot CV comparison
sns.boxplot(data=cv_df, x='Model', y='ROC_AUC', palette='Blues_r', ax=ax1)
sns.stripplot(data=cv_df, x='Model', y='ROC_AUC', color='black', alpha=0.5, jitter=0.2, ax=ax1)
ax1.set_title('10-Fold CV Distribution Across Architectures', fontweight='bold')
ax1.grid(axis='y', alpha=0.3)

# 2. Point plot showing mean and confidence intervals
sns.pointplot(data=cv_df, x='Model', y='ROC_AUC', color='crimson', capsize=0.1, errorbar=('ci', 95), ax=ax2)
ax2.set_title('Mean ROC-AUC ± 95% Bootstrap CI', fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
