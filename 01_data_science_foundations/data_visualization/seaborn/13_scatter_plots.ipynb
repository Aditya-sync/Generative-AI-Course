# Scatter Plots in Seaborn (scatterplot & relplot)

## Capabilities
- Multi-dimensional semantic mappings: `hue`, `style`, `size`.
- Marker sizing range control (`sizes=(min, max)`).
- Continuous color normalization (`hue_norm`).

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

# 1. Categorical Hue + Style
sns.scatterplot(
    data=penguins,
    x='bill_length_mm', y='bill_depth_mm',
    hue='species', style='sex',
    s=70, alpha=0.8, palette='Set2',
    ax=ax1
)
ax1.set_title('Categorical Semantics (Species & Sex)', fontweight='bold')
ax1.grid(True, alpha=0.3)

# 2. Continuous Hue + Sizing
sns.scatterplot(
    data=penguins,
    x='flipper_length_mm', y='body_mass_g',
    hue='bill_length_mm', size='bill_length_mm',
    sizes=(20, 180), palette='viridis',
    ax=ax2
)
ax2.set_title('Continuous Numeric Encoding', fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
