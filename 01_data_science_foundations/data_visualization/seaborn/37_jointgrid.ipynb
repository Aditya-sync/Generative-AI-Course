# JointGrid in Seaborn

## Granular Control over Joint Plots
- Directly access `g.ax_joint`, `g.ax_marg_x`, and `g.ax_marg_y`.
- Plot completely custom artists on the joint and marginal axes.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

tips = sns.load_dataset('tips')

# Initialize JointGrid
g = sns.JointGrid(data=tips, x='total_bill', y='tip', height=6)

# Custom plot mappings
g.plot_joint(sns.scatterplot, s=50, alpha=0.6, color='#2c3e50')
g.plot_joint(sns.kdeplot, color='red', levels=5, alpha=0.6)

# Marginals
sns.histplot(tips['total_bill'], kde=True, ax=g.ax_marg_x, color='#3498db')
sns.histplot(y=tips['tip'], kde=True, ax=g.ax_marg_y, color='#2ecc71')

g.fig.suptitle('JointGrid: Custom Composite Marginals and Contours', y=1.02, fontweight='bold')
plt.show()
```
