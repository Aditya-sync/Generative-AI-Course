# Combining Seaborn with Matplotlib

## Synergistic Power
- Leverage Seaborn for high-level statistical visualization.
- Use Matplotlib for fine-grained annotations, custom spines, dual axes, and layout geometry.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

tips = sns.load_dataset('tips')

# Create matplotlib Figure and Axes explicitly
fig, ax = plt.subplots(figsize=(9, 5), dpi=100)

# 1. Use Seaborn for plotting statistical distribution
sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2', width=0.4,
            boxprops=dict(alpha=0.8), ax=ax)

# 2. Use Matplotlib for custom benchmark threshold and annotations
median_overall = tips['total_bill'].median()
ax.axhline(median_overall, color='crimson', linestyle='--', lw=2,
           label=f'Overall Median: ${median_overall:.2f}')

# Custom text annotation pointing to high outliers
max_tip_row = tips.loc[tips['total_bill'].idxmax()]
ax.annotate(
    f"Max Bill: ${max_tip_row['total_bill']:.2f}",
    xy=(3, max_tip_row['total_bill']),
    xytext=(2.2, max_tip_row['total_bill'] + 2),
    arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=7),
    fontweight='bold', fontsize=10
)

# Matplotlib formatting
ax.set_title('Hybrid Matplotlib + Seaborn Chart', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Day of Week', fontsize=11, fontweight='bold')
ax.set_ylabel('Total Bill ($)', fontsize=11, fontweight='bold')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.legend(loc='upper left', frameon=True)
ax.grid(axis='y', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
