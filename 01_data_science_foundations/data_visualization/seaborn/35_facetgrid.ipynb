# FacetGrid in Seaborn

## Multi-Plot Grid for Conditional Subsets
- Initialize grid with `sns.FacetGrid(df, row=..., col=..., hue=...)`.
- Map plotting functions (`g.map()` or `g.map_dataframe()`).
- Add legend and adjust titles (`g.set_titles()`, `g.add_legend()`).

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Initialize FacetGrid
g = sns.FacetGrid(tips, col='time', row='smoker', hue='sex', margin_titles=True, height=3.5, aspect=1.2)

# Map scatter plot onto each facet
g.map(sns.scatterplot, 'total_bill', 'tip', alpha=0.7)

# Customize titles and labels
g.set_axis_labels('Total Bill ($)', 'Tip ($)')
g.set_titles(col_template='Time: {col_name}', row_template='Smoker: {row_name}')
g.add_legend(title='Gender')

g.fig.subplots_adjust(top=0.9)
g.fig.suptitle('Faceted Grid of Tip vs Bill across 4 Subgroups', fontweight='bold', fontsize=14)
plt.show()
```
