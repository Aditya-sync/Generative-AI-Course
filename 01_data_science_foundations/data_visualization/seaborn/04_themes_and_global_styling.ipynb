# Themes & Global Styling in Seaborn

## Global Theme Controls
- `sns.set_theme()`: Comprehensive theme configuration.
- `sns.set_style()`: Visual style (`darkgrid`, `whitegrid`, `dark`, `white`, `ticks`).
- `sns.set_context()`: Scale formatting (`paper`, `notebook`, `talk`, `poster`).
- `sns.despine()`: Streamlined removal of top and right spines.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# 1. Comparing Styles
styles = ['darkgrid', 'whitegrid', 'white', 'ticks']
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, style in zip(axes.flat, styles):
    sns.set_style(style)
    sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', ax=ax)
    ax.set_title(f'Style: "{style}"', fontweight='bold')
    if style in ['white', 'ticks']:
        sns.despine(ax=ax)

plt.tight_layout()
plt.show()
```

```python
# 2. Controlling Context Scale
contexts = ['paper', 'notebook', 'talk', 'poster']
fig, axes = plt.subplots(1, 4, figsize=(18, 4))

for ax, ctx in zip(axes, contexts):
    with sns.plotting_context(ctx):
        sns.lineplot(x=[0, 1, 2], y=[0, 1, 4], ax=ax)
        ax.set_title(f'Context: {ctx}')

plt.tight_layout()
plt.show()

# Reset to standard defaults
sns.set_theme(style='whitegrid', context='notebook')
```
