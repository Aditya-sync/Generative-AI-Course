# Class Distribution Analysis (Class Imbalance)

## Diagnosing Imbalance
- Visualizing positive vs negative class imbalance (e.g. 98:2).
- Class proportion barplots with percentage annotations.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Simulating imbalanced fraud dataset
classes = ['Legitimate'] * 950 + ['Fraudulent'] * 50
df = pd.DataFrame({'Transaction_Type': classes})

fig, ax = plt.subplots(figsize=(7, 5))
palette = {'Legitimate': '#2ecc71', 'Fraudulent': '#e74c3c'}

sns.countplot(data=df, x='Transaction_Type', palette=palette, ax=ax)
ax.set_title('Class Imbalance Diagnostic (Fraud Detection)', fontsize=13, fontweight='bold')
ax.set_ylabel('Transaction Count')

total = len(df)
for p in ax.patches:
    height = p.get_height()
    pct = 100 * height / total
    ax.annotate(f'{height} ({pct:.1f}%)',
                (p.get_x() + p.get_width() / 2., height + 15),
                ha='center', va='bottom', fontsize=11, fontweight='bold')

ax.set_ylim(0, 1100)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
