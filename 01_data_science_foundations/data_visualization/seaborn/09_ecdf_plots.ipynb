# Empirical Cumulative Distribution Function (ECDF) Plots

## Why ECDF is Superior to Histograms/KDE
- No binning bias or bandwidth tuning artifacts.
- Every single data point is directly plotted.
- $y$-value directly represents $P(X \le x)$ (Percentile rank).

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Standard ECDF across species
sns.ecdfplot(data=penguins, x='flipper_length_mm', hue='species', ax=ax1)
ax1.set_title('ECDF of Flipper Length by Species', fontweight='bold')
ax1.axhline(0.5, color='gray', linestyle='--', label='Median line (50%)')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Complementary ECDF (Survival Function: P(X > x))
sns.ecdfplot(data=penguins, x='body_mass_g', complementary=True, hue='species', ax=ax2)
ax2.set_title('Complementary ECDF: P(Mass > x)', fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
