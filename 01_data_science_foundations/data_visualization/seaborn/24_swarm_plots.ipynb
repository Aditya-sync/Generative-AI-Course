# Swarm Plots in Seaborn (swarmplot)

## Beeswarm Visualization
- Avoids point overlapping while preserving exact values.
- Tuning point size and warnings on high-density data.

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

fig, ax = plt.subplots(figsize=(10, 5))

sns.swarmplot(data=penguins, x='species', y='body_mass_g', hue='sex',
              palette={'Male': '#2980b9', 'Female': '#8e44ad'}, size=5, dodge=True, ax=ax)

ax.set_title('Penguin Body Mass Beeswarm Distribution', fontsize=14, fontweight='bold')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
