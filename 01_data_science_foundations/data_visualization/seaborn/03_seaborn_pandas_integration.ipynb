# Seaborn + Pandas Integration

## Seamless DataFrame Workflows
- Seaborn automatically uses DataFrame column names for axes labels and legend titles.
- Native handling of Pandas categorical data types, datetimes, and Index structures.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Create rich pandas DataFrame
np.random.seed(42)
df = pd.DataFrame({
    'Department': pd.Categorical(['Engineering', 'Sales', 'Marketing', 'HR'] * 50,
                                 categories=['HR', 'Marketing', 'Sales', 'Engineering'], ordered=True),
    'Experience_Years': np.random.uniform(1, 15, 200),
    'Salary_k': np.random.normal(90, 25, 200),
    'Remote_Worker': np.random.choice([True, False], size=200, p=[0.6, 0.4])
})
# Inject correlation
df['Salary_k'] += df['Experience_Years'] * 4.5

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Automatic categorical ordering & labels
sns.boxplot(data=df, x='Department', y='Salary_k', hue='Remote_Worker', ax=ax1, palette='Blues')
ax1.set_title('Categorical Order Automatically Respected', fontweight='bold')

# 2. Aggregation & estimation on DataFrame columns
sns.barplot(data=df, x='Department', y='Experience_Years', estimator=np.median,
            errorbar=('ci', 95), capsize=0.1, ax=ax2, palette='mako')
ax2.set_title('Median Experience with 95% Bootstrap CI', fontweight='bold')

plt.tight_layout()
plt.show()
```
