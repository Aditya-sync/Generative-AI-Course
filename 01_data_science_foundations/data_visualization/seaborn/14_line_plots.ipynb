# Line Plots in Seaborn (lineplot)

## Aggregation & Confidence Bands
- Automatic aggregation of repeated measurements over time steps.
- Bootstrap confidence interval bands (`errorbar=('ci', 95)` or `errorbar='sd'`).

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Simulating multiple runs of ML training across random seeds
np.random.seed(42)
records = []
for run in range(8):
    for epoch in range(1, 26):
        loss_adam = 2.0 * np.exp(-epoch / 6) + np.random.normal(0, 0.08)
        loss_sgd = 2.5 * np.exp(-epoch / 12) + np.random.normal(0, 0.12)
        records.append({'Epoch': epoch, 'Loss': loss_adam, 'Optimizer': 'Adam', 'Seed': run})
        records.append({'Epoch': epoch, 'Loss': loss_sgd, 'Optimizer': 'SGD', 'Seed': run})

df_runs = pd.DataFrame(records)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. 95% Bootstrap Confidence Interval
sns.lineplot(data=df_runs, x='Epoch', y='Loss', hue='Optimizer', errorbar=('ci', 95), ax=ax1)
ax1.set_title('Mean Loss with 95% CI Bands (8 Seeds)', fontweight='bold')
ax1.grid(True, alpha=0.3)

# 2. Standard Deviation Errorbar + Markers
sns.lineplot(data=df_runs, x='Epoch', y='Loss', hue='Optimizer', style='Optimizer',
             markers=True, errorbar='sd', ax=ax2)
ax2.set_title('Mean Loss with ±1 Std Dev Bands', fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
