# Seaborn Objects (`seaborn.objects`) API

## Declarative Grammar of Graphics
- Introduced in Seaborn 0.12 (inspired by ggplot2).
- `so.Plot()` chain: Marks, Transforms, Facets, and Themes.

```python
import seaborn.objects as so
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

# Declarative Plot: Points with Linear Model Smoother per Species
p = (
    so.Plot(penguins, x='bill_length_mm', y='bill_depth_mm', color='species')
    .add(so.Dots(alpha=0.6, pointsize=5))
    .add(so.Line(linewidth=2), so.PolyFit(order=1))
    .label(
        x='Bill Length (mm)',
        y='Bill Depth (mm)',
        color='Penguin Species',
        title='Declarative seaborn.objects Grammar: PolyFit per Species'
    )
    .theme({'axes.grid': True, 'grid.alpha': 0.3})
)

# Render
fig = plt.figure(figsize=(8, 5))
p.on(fig).plot()
plt.show()
```
