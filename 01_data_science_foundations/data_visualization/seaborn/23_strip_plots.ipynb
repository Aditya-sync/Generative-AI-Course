# Strip Plots in Seaborn (stripplot)

## Overlaying Raw Points on Box/Violin Plots
- Combining box plots with strip plots for transparency (displaying both summary stats and underlying data density).

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, ax = plt.subplots(figsize=(9, 5))

# Underlay boxplot
sns.boxplot(data=tips, x='day', y='total_bill', color='lightgray', showfliers=False, ax=ax)

# Overlay stripplot
sns.stripplot(data=tips, x='day', y='total_bill', hue='time', jitter=0.2,
              palette='dark', alpha=0.7, size=6, ax=ax)

ax.set_title('Box Plot Underlay + Stripplot Overlay', fontsize=13, fontweight='bold')
ax.legend(title='Meal Time')
plt.tight_layout()
plt.show()
```
