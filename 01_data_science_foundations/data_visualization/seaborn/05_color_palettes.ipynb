# Color Palettes in Seaborn

## Palette Types
1. **Qualitative**: `deep`, `muted`, `bright`, `pastel`, `dark`, `colorblind` (Accessible for color vision deficiency).
2. **Sequential**: `rocket`, `mako`, `flare`, `crest`, `Blues`.
3. **Diverging**: `vlag`, `icefire`, `coolwarm`.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Built-in Palettes Demo
fig, axes = plt.subplots(6, 1, figsize=(9, 7))

palettes = ['colorblind', 'deep', 'rocket', 'mako', 'vlag', 'icefire']
for ax, p in zip(axes, palettes):
    sns.palplot(sns.color_palette(p, 10), ax=ax)
    ax.set_title(f"Palette: '{p}'", loc='left', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# 2. Applying Custom & Colorblind Palettes to Real Data
penguins = sns.load_dataset('penguins')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Colorblind-safe palette (crucial for publications)
sns.boxplot(data=penguins, x='species', y='body_mass_g', palette='colorblind', ax=ax1)
ax1.set_title('Colorblind-Safe Qualitative Palette', fontweight='bold')

# Continuous sequential palette for numeric encoding
sns.scatterplot(data=penguins, x='bill_length_mm', y='bill_depth_mm',
                hue='body_mass_g', palette='viridis', size='body_mass_g', ax=ax2)
ax2.set_title('Sequential Palette (Viridis) for Numeric Feature', fontweight='bold')

plt.tight_layout()
plt.show()
```
