# Joint Plots in Seaborn (jointplot)

## Joint & Marginal Distributions
- Combines bivariate relationship in center with univariate marginals on top and right.
- `kind`: `'scatter'`, `'kde'`, `'hist'`, `'hex'`, `'reg'`.

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

# 1. Joint plot with regression & KDE marginals
g1 = sns.jointplot(data=penguins, x='flipper_length_mm', y='body_mass_g', kind='reg',
                   color='#2b5c8f', height=5.5)
g1.fig.suptitle('Joint Regression + Marginal Densities', y=1.02, fontweight='bold')
plt.show()

# 2. Joint plot with Hexbin for density
g2 = sns.jointplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', kind='hex',
                   color='#27ae60', height=5.5)
g2.fig.suptitle('Joint Hexbin Representation', y=1.02, fontweight='bold')
plt.show()
```
