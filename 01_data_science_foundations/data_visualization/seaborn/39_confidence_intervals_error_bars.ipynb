# Confidence Intervals & Error Bars in Seaborn

## Seaborn 0.12+ `errorbar` API
- `errorbar=('ci', 95)`: Percentile bootstrap confidence interval.
- `errorbar='sd'`: Standard deviation.
- `errorbar='se'`: Standard error of the mean.
- `errorbar=('pi', 90)`: Prediction interval.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, axes = plt.subplots(1, 4, figsize=(18, 4.5))

options = [
    (('ci', 95), '95% Bootstrap CI', '#2980b9'),
    ('sd', 'Standard Deviation (sd)', '#27ae60'),
    ('se', 'Standard Error (se)', '#8e44ad'),
    (('pi', 80), '80% Prediction Interval', '#d35400')
]

for ax, (eb, title, color) in zip(axes, options):
    sns.barplot(data=tips, x='day', y='total_bill', errorbar=eb, color=color, capsize=0.15, ax=ax)
    ax.set_title(title, fontweight='bold', fontsize=11)
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
