# Rug Plots in Seaborn

## Fine-Grained Marginal Observation
- `sns.rugplot()`: Draws single ticks on axes for each individual data point.
- Excellent when paired with KDE or Scatter plots to expose individual sample locations and outliers.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

tips = sns.load_dataset('tips')

fig, ax = plt.subplots(figsize=(9, 5))

# Combine KDE + Rug plot
sns.kdeplot(data=tips, x='total_bill', fill=True, color='teal', alpha=0.3, ax=ax)
sns.rugplot(data=tips, x='total_bill', height=0.1, color='darkslategray', lw=1.2, ax=ax)

ax.set_title('Kernel Density Estimate with Marginal Rug Plot', fontsize=13, fontweight='bold')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
