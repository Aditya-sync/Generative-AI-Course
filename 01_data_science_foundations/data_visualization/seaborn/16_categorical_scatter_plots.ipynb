# Categorical Scatter Plots (Strip & Swarm)

## Comparison
- `sns.stripplot`: Jittered points across categories.
- `sns.swarmplot`: Non-overlapping points positioned to reflect distribution density.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Strip plot with jitter
sns.stripplot(data=tips, x='day', y='total_bill', hue='sex', jitter=0.25, dodge=True, alpha=0.7, ax=ax1)
ax1.set_title('Strip Plot (Jittered Points)', fontweight='bold')
ax1.grid(True, alpha=0.3)

# 2. Swarm plot (Beeswarm algorithm)
sns.swarmplot(data=tips, x='day', y='total_bill', hue='sex', dodge=True, size=5, ax=ax2)
ax2.set_title('Swarm Plot (Non-Overlapping Points)', fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
