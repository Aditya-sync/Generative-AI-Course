# Bivariate Distribution Visualization

## Exploring Pairwise Joint Distributions
- Hexbin plots for dense distributions.
- 2D Histograms & Contour maps via `sns.displot(kind='hist' / 'kde')`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Dense synthetic dataset (10,000 points)
np.random.seed(42)
mean = [0, 0]
cov = [[1, 0.7], [0.7, 1]]
x, y = np.random.multivariate_normal(mean, cov, 10000).T

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

# 1. Hexbin plot for massive datasets
hb = ax1.hexbin(x, y, gridsize=35, cmap='Blues', mincnt=1)
fig.colorbar(hb, ax=ax1, label='Point Density')
ax1.set_title('Bivariate Hexbin (10k Samples)', fontweight='bold')
ax1.set_xlabel('Feature 1')
ax1.set_ylabel('Feature 2')

# 2. 2D Seaborn histplot with colorbar
sns.histplot(x=x, y=y, bins=35, pthresh=0.02, cmap='viridis', cbar=True, ax=ax2)
ax2.set_title('Seaborn 2D Histogram Heatmap', fontweight='bold')
ax2.set_xlabel('Feature 1')
ax2.set_ylabel('Feature 2')

plt.tight_layout()
plt.show()
```
