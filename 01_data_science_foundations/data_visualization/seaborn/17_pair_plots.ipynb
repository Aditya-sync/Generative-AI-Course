# Pair Plots in Seaborn (pairplot)

## Instant Multidimensional EDA
- Generates an $N \times N$ grid of scatter plots and marginal distributions.
- Essential first inspection tool for classification datasets.

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

# Pairplot colored by target class
g = sns.pairplot(
    data=penguins,
    hue='species',
    diag_kind='kde',
    plot_kws={'alpha': 0.6, 's': 25},
    palette='Set1'
)
g.fig.suptitle('Pairwise Feature Matrix by Target Class', y=1.02, fontweight='bold', fontsize=14)
plt.show()
```
