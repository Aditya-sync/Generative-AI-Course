# Multivariate Analysis in Seaborn

## Visualizing 3+ Variables Simultaneously
- Faceting with `row` and `col`
- Interaction effects between features.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# relplot faceting by two conditioning variables (time and smoker)
g = sns.relplot(
    data=tips,
    x='total_bill', y='tip',
    hue='sex', size='size',
    col='time', row='smoker',
    sizes=(20, 150),
    palette='muted',
    height=3.5, aspect=1.1
)

g.fig.suptitle('Multivariate Interaction: Total Bill, Tip, Sex, Party Size, Time & Smoker',
               y=1.04, fontweight='bold', fontsize=13)
plt.show()
```
