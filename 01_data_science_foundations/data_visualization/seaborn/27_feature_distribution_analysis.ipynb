# Feature Distribution Analysis for ML

## What to Inspect
- Skewness and kurtosis
- Outlier tails
- Distribution transformations (e.g. log transform)

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Skewed feature simulation (e.g., Transaction Amount)
np.random.seed(42)
raw_amounts = np.random.exponential(scale=50, size=2000)
log_amounts = np.log1p(raw_amounts)

df = pd.DataFrame({'Raw_Amount': raw_amounts, 'Log_Amount': log_amounts})

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Raw Skewed Distribution
sns.histplot(data=df, x='Raw_Amount', kde=True, color='crimson', ax=ax1)
ax1.set_title(f'Raw Amount Distribution (Skew: {df["Raw_Amount"].skew():.2f})', fontweight='bold')

# Transformed Normal Distribution
sns.histplot(data=df, x='Log_Amount', kde=True, color='teal', ax=ax2)
ax2.set_title(f'Log1p Transformed Distribution (Skew: {df["Log_Amount"].skew():.2f})', fontweight='bold')

plt.tight_layout()
plt.show()
```
