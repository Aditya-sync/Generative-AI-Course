# Regression Plots in Seaborn (regplot & lmplot)

## Capabilities
- Linear, polynomial (`order=2`), logistic (`logistic=True`), and robust regression (`robust=True`).
- Faceting regressions across subsets via `sns.lmplot()`.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Linear Fit with 95% CI
sns.regplot(data=tips, x='total_bill', y='tip', ax=ax1,
            scatter_kws={'alpha': 0.6, 'color': '#2980b9'},
            line_kws={'color': '#c0392b', 'linewidth': 2})
ax1.set_title('Linear Regression with 95% CI Band', fontweight='bold')
ax1.grid(True, alpha=0.3)

# 2. Polynomial Fit (order=2)
sns.regplot(data=tips, x='total_bill', y='tip', order=2, ax=ax2,
            scatter_kws={'alpha': 0.6, 'color': '#27ae60'},
            line_kws={'color': '#8e44ad', 'linewidth': 2})
ax2.set_title('2nd Degree Polynomial Fit', fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# 3. Faceted Regression using lmplot
g = sns.lmplot(data=tips, x='total_bill', y='tip', hue='smoker', col='time', height=4, aspect=1.2)
g.fig.suptitle('Faceted Regression Fit (lmplot)', y=1.03, fontweight='bold')
plt.show()
```
