# Kernel Density Estimation (KDE) Plots

## Capabilities
- Smooth nonparametric estimation of probability density function (`sns.kdeplot`).
- Bandwidth smoothing control via `bw_adjust`.
- Bivariate 2D KDE contour and filled surface plots.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. 1D KDE with Bandwidth Adjustment
sns.kdeplot(data=tips, x='total_bill', hue='time', fill=True, common_norm=False, palette='crest', alpha=0.4, ax=ax1)
ax1.set_title('1D Density Estimate (common_norm=False)', fontweight='bold')

# 2. 2D Bivariate Density Contours
sns.kdeplot(data=tips, x='total_bill', y='tip', hue='sex', levels=8, fill=True, alpha=0.5, ax=ax2)
ax2.set_title('2D Bivariate KDE Contours by Sex', fontweight='bold')

plt.tight_layout()
plt.show()
```
