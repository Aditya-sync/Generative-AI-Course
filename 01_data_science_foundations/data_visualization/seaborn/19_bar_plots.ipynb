# Bar Plots in Seaborn (barplot)

## Statistical Aggregation
- Automatically computes summary statistic (mean by default) and confidence intervals.
- Custom estimators via `estimator=np.median` or `estimator=sum`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Mean with 95% Bootstrap CI
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', errorbar=('ci', 95),
            palette='Blues', capsize=0.1, ax=ax1)
ax1.set_title('Mean Total Bill ± 95% CI', fontweight='bold')

# 2. Median Estimator
sns.barplot(data=tips, x='day', y='tip', hue='time', estimator=np.median,
            errorbar='sd', palette='Pastel2', capsize=0.1, ax=ax2)
ax2.set_title('Median Tip ± Std Dev', fontweight='bold')

plt.tight_layout()
plt.show()
```
