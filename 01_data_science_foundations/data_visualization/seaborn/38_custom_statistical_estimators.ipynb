# Custom Statistical Estimators in Seaborn

## Beyond the Mean
- Using `estimator=np.median`, `trimmed_mean`, or custom robust summary metrics.
- Vectorized estimators accepting array inputs.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

tips = sns.load_dataset('tips')

# Define custom 10% trimmed mean estimator
def trimmed_mean(x):
    return stats.trim_mean(x, proportiontocut=0.1)

# Define Geometric Mean estimator
def geometric_mean(x):
    return stats.gmean(x[x > 0])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Standard Mean vs Trimmed Mean
sns.barplot(data=tips, x='day', y='total_bill', estimator=np.mean,
            errorbar=None, color='#3498db', alpha=0.6, label='Arithmetic Mean', ax=ax1)
sns.barplot(data=tips, x='day', y='total_bill', estimator=trimmed_mean,
            errorbar=None, color='#e74c3c', alpha=0.6, label='10% Trimmed Mean', ax=ax1)
ax1.set_title('Mean vs Robust Trimmed Mean', fontweight='bold')
ax1.legend()

# 2. Geometric Mean Estimator
sns.barplot(data=tips, x='day', y='tip', estimator=geometric_mean,
            errorbar=('ci', 90), palette='viridis', capsize=0.1, ax=ax2)
ax2.set_title('Geometric Mean Tip with 90% CI', fontweight='bold')

plt.tight_layout()
plt.show()
```
