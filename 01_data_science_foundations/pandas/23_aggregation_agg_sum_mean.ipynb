# Aggregation — agg(), sum(), mean(), etc.

## Definition
**Aggregation** reduces a DataFrame or group to summary statistics.
It's the backbone of EDA and feature engineering.

### Direct Aggregation on DataFrame
These apply to entire DataFrame or specific columns:
`df.sum()`, `df.mean()`, `df.median()`, `df.std()`, `df.var()`,
`df.min()`, `df.max()`, `df.count()`, `df.nunique()`, `df.first()`, `df.last()`

### `agg()` — Flexible Multi-Function Aggregation
```python
df.agg({'col1': 'sum', 'col2': ['mean', 'std']})
# or
df.groupby('key').agg(new_name=('col', 'func'))
```

### Named Aggregation (pandas >= 0.25)
```python
df.groupby('dept').agg(
    avg_salary = ('salary', 'mean'),
    max_score  = ('score', 'max')
)
```

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'dept':   ['Eng', 'Mkt', 'Eng', 'HR', 'Eng', 'Mkt', 'HR'],
    'salary': [90000, 65000, 75000, 85000, 95000, 80000, 70000],
    'score':  [88, 72, 91, 75, 84, 78, 69],
    'yoe':    [5, 3, 2, 10, 7, 6, 8]
})
print(df)
````

````python
# Direct aggregation — whole DataFrame
print('Column sums:')
print(df[['salary', 'score', 'yoe']].sum())

print('\nColumn means:')
print(df[['salary', 'score', 'yoe']].mean().round(2))
````

````python
# agg() with a single function
print('agg with single func:')
print(df[['salary', 'score']].agg('mean'))
````

````python
# agg() with multiple functions per column
print('agg with list of funcs:')
print(df[['salary', 'score']].agg(['mean', 'median', 'std', 'min', 'max']))
````

````python
# agg() with dict — different functions per column
print('agg with dict:')
print(df.agg({
    'salary': ['mean', 'std', 'max'],
    'score': ['mean', 'min'],
    'yoe': 'sum'
}))
````

````python
# Named aggregation after groupby
result = df.groupby('dept').agg(
    num_employees = ('salary', 'count'),
    avg_salary    = ('salary', 'mean'),
    median_score  = ('score', 'median'),
    total_yoe     = ('yoe', 'sum'),
    max_salary    = ('salary', 'max'),
    min_salary    = ('salary', 'min')
).round(0)
print('Named aggregation result:')
print(result)
````

````python
# Custom aggregation function
def salary_iqr(x):
    return x.quantile(0.75) - x.quantile(0.25)

result = df.groupby('dept')['salary'].agg(['mean', salary_iqr])
result.columns = ['mean_salary', 'salary_iqr']
print('Custom agg function:')
print(result)
````

````python
# Aggregate with transform (keeps original index)
df['dept_avg_salary'] = df.groupby('dept')['salary'].transform('mean')
df['salary_vs_dept_avg'] = df['salary'] - df['dept_avg_salary']
print('Salary vs dept average:')
print(df[['name' if 'name' in df.columns else 'dept', 'dept', 'salary', 'dept_avg_salary', 'salary_vs_dept_avg']])
````

````python
# describe() as aggregation
print('Detailed stats per dept:')
print(df.groupby('dept')['salary'].describe())
````
