# Datetime & Time-Series Operations

## Definition
Datetime handling is critical in ML — nearly every real dataset has timestamps.
Pandas provides the `.dt` accessor and `DatetimeIndex` for vectorized datetime operations.

### Key Concepts
| Concept | Type | Description |
|---------|------|-------------|
| `Timestamp` | Scalar | Single point in time |
| `DatetimeIndex` | Index | Array of timestamps |
| `Period` | Scalar | Span of time (year, month) |
| `Timedelta` | Duration | Difference between timestamps |

### `.dt` Accessor Properties
| Property | Returns |
|----------|---------|
| `.dt.year` | Year |
| `.dt.month` | Month (1-12) |
| `.dt.day` | Day (1-31) |
| `.dt.hour` | Hour |
| `.dt.dayofweek` | Mon=0, Sun=6 |
| `.dt.dayofyear` | Day of year (1-365) |
| `.dt.week` | Week number |
| `.dt.quarter` | Quarter (1-4) |
| `.dt.is_weekend` | Bool |

```python
import pandas as pd
import numpy as np

# Creating datetime Series
dates = pd.to_datetime(['2023-01-15', '2023-03-22', '2023-07-04', '2023-12-25'])
s = pd.Series(dates)
print('DatetimeSeries:')
print(s)
print('dtype:', s.dtype)
```

```python
# .dt accessor properties
df = pd.DataFrame({
    'event': ['Product Launch', 'Marketing Campaign', 'Sales Event', 'Year End Review'],
    'date': pd.to_datetime(['2023-01-15', '2023-03-22', '2023-07-04', '2023-12-25'])
})

df['year']      = df['date'].dt.year
df['month']     = df['date'].dt.month
df['day']       = df['date'].dt.day
df['dayofweek'] = df['date'].dt.dayofweek  # Monday=0, Sunday=6
df['day_name']  = df['date'].dt.day_name()
df['quarter']   = df['date'].dt.quarter
df['is_weekend']= df['date'].dt.dayofweek >= 5
print(df)
```

```python
# Date arithmetic — Timedelta
df['days_from_new_year'] = (df['date'] - pd.Timestamp('2023-01-01')).dt.days
df['next_week'] = df['date'] + pd.Timedelta(days=7)
print(df[['event', 'date', 'days_from_new_year', 'next_week']])
```

```python
# DatetimeIndex — time-series indexing
idx = pd.date_range('2023-01-01', periods=12, freq='ME')   # ME = Month End
ts = pd.Series(np.random.randn(12), index=idx, name='value')
print('Monthly time series:')
print(ts)
```

```python
# Resampling — change frequency
daily_idx = pd.date_range('2023-01-01', periods=90, freq='D')
daily = pd.Series(np.random.randn(90), index=daily_idx)

# Resample to weekly mean
weekly = daily.resample('W').mean()
print('Daily -> Weekly mean:')
print(weekly.head(5))

# Resample to monthly aggregates
monthly = daily.resample('ME').agg(['mean', 'std', 'min', 'max'])
print('Daily -> Monthly stats:')
print(monthly.head(4))
```

```python
# Date parsing formats
raw_dates = ['15/01/2023', '22-03-2023', 'July 4, 2023', '20231225']
parsed = pd.to_datetime(raw_dates, dayfirst=False, infer_datetime_format=True)
print('Parsed dates:', parsed)

# Specify format explicitly for speed and correctness
df_custom = pd.DataFrame({'raw': ['2023_01_15', '2023_07_04']})
df_custom['date'] = pd.to_datetime(df_custom['raw'], format='%Y_%m_%d')
print(df_custom)
```

```python
# Feature engineering from datetime (ML use case)
df = pd.DataFrame({
    'order_date': pd.date_range('2022-01-01', periods=100, freq='3D'),
    'amount': np.random.uniform(10, 500, 100)
})

# Extract time features for ML
df['year']       = df['order_date'].dt.year
df['month']      = df['order_date'].dt.month
df['day_of_week'] = df['order_date'].dt.dayofweek
df['day_of_year'] = df['order_date'].dt.dayofyear
df['is_weekend']  = (df['order_date'].dt.dayofweek >= 5).astype(int)
df['week_of_year'] = df['order_date'].dt.isocalendar().week.astype(int)
print('Time features:')
print(df[['order_date', 'month', 'day_of_week', 'is_weekend', 'week_of_year']].head(8))
```
