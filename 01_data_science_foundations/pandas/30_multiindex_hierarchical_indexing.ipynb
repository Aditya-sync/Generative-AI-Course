# MultiIndex / Hierarchical Indexing

## Definition
A **MultiIndex** (or hierarchical index) allows multiple levels of indexing on rows or columns.
This is useful for working with:
- Panel data (multiple entities over time)
- Grouped / nested data
- Results of groupby aggregations

### Creating MultiIndex
```python
pd.MultiIndex.from_tuples([('A', 1), ('A', 2), ('B', 1)])
pd.MultiIndex.from_product([['A', 'B'], [1, 2, 3]])
df.set_index(['col1', 'col2'])  # from DataFrame columns
```

### Accessing MultiIndex Data
```python
df.loc['A']               # outer level
df.loc[('A', 1)]          # specific row
df.loc[['A', 'B'], :]     # multiple outer levels
df.xs('label', level='level_name')  # cross-section
```

````python
import pandas as pd
import numpy as np

# Create from groupby result (most common source)
df = pd.DataFrame({
    'dept':   ['Eng', 'Eng', 'Mkt', 'Mkt', 'HR', 'HR'],
    'level':  ['Sr', 'Jr', 'Sr', 'Jr', 'Sr', 'Jr'],
    'salary': [90000, 70000, 85000, 65000, 80000, 60000],
    'score':  [88, 82, 78, 72, 85, 75]
})

# Set MultiIndex from two columns
df_mi = df.set_index(['dept', 'level'])
print('MultiIndex DataFrame:')
print(df_mi)
print('\nIndex:', df_mi.index)
````

````python
# Accessing MultiIndex data
print('Engineering dept (all levels):')
print(df_mi.loc['Eng'])

print('\nEng, Sr level (specific row):')
print(df_mi.loc[('Eng', 'Sr')])

print('\nMultiple depts:')
print(df_mi.loc[['Eng', 'HR']])
````

````python
# xs() — cross-section (select on inner level)
print('All Sr level employees:')
print(df_mi.xs('Sr', level='level'))
````

````python
# Creating MultiIndex explicitly
index = pd.MultiIndex.from_product(
    [['2022', '2023'], ['Q1', 'Q2', 'Q3', 'Q4']],
    names=['year', 'quarter']
)
data = pd.DataFrame({
    'revenue': np.random.randint(100, 500, len(index)),
    'cost': np.random.randint(50, 300, len(index))
}, index=index)
print('Multi-year quarterly data:')
print(data)
````

````python
# Aggregation with MultiIndex result
result = df.groupby(['dept', 'level']).agg({'salary': 'mean', 'score': 'mean'})
print('GroupBy MultiIndex result:')
print(result)
print()

# Unstacking inner level to columns
print('After unstack():')
print(result.unstack(level='level'))
````

````python
# reset_index — flatten MultiIndex
flat = result.reset_index()
print('After reset_index:')
print(flat)
````

````python
# MultiIndex on columns
cols = pd.MultiIndex.from_tuples([
    ('score', 'math'), ('score', 'science'),
    ('grade', 'math'), ('grade', 'science')
])
df_mc = pd.DataFrame(np.random.randint(60, 100, (3, 4)),
    columns=cols, index=['Alice', 'Bob', 'Carol'])
print('MultiIndex columns:')
print(df_mc)
print()
print('Access score columns:')
print(df_mc['score'])
````

````python
# EXCEPTION: Key error on MultiIndex
try:
    df_mi.loc[('Eng', 'Intern')]  # 'Intern' doesn't exist
except KeyError as e:
    print('KeyError:', e)

# swaplevel — change order of index levels
print('\nSwapped levels:')
print(df_mi.swaplevel())
````
