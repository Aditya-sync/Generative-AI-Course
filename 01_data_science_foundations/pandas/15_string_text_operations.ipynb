# String / Text Operations

## Definition
Pandas provides the `.str` accessor — a powerful vectorized string processing interface
that applies Python string methods to an entire Series without loops.

### The `.str` accessor
Access via: `df['text_col'].str.<method>()`

### Common .str Methods
| Method | Description |
|--------|-------------|
| `.str.lower()` / `.str.upper()` | Case conversion |
| `.str.strip()` / `.str.lstrip()` / `.str.rstrip()` | Remove whitespace |
| `.str.split(pat)` | Split string |
| `.str.contains(pat)` | Substring check (regex) |
| `.str.startswith()` / `.str.endswith()` | Prefix/suffix check |
| `.str.replace(old, new)` | Replace substring |
| `.str.extract(regex)` | Extract groups using regex |
| `.str.findall(regex)` | Find all occurrences |
| `.str.len()` | String length |
| `.str.count(pat)` | Count occurrences |
| `.str.get(i)` | Get element by position |
| `.str.join(sep)` | Join list elements |

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name':  ['  Alice Smith ', 'BOB JONES', 'carol white  ', 'Dave Brown'],
    'email': ['alice@gmail.com', 'bob@yahoo.com', 'carol@gmail.com', 'dave@outlook.com'],
    'phone': ['123-456-7890', '(234) 567-8901', '345.678.9012', '456-789-0123'],
    'bio':   ['ML engineer with 3yrs exp', 'Data Scientist, Python lover',
              'Backend dev. Java, Python', 'UX Designer with 5 years']
})
print(df)
```

```python
# Basic string operations
print('lower:', df['name'].str.lower().values)
print('upper:', df['name'].str.upper().values)
print('strip:', df['name'].str.strip().values)
print('len:',   df['name'].str.strip().str.len().values)
```

```python
# contains, startswith, endswith
print('Gmail users:', df[df['email'].str.contains('@gmail')])

print('\nName starts with A or B:')
print(df[df['name'].str.strip().str.lower().str[0].isin(['a', 'b'])])

# count occurrences
print('\nComma count in bio:', df['bio'].str.count(',').values)
```

```python
# split — returns list of strings
domains = df['email'].str.split('@').str[1]  # get part after @
print('Email domains:', domains.values)

# Split and expand into multiple columns
name_parts = df['name'].str.strip().str.split(' ', expand=True)
name_parts.columns = ['first', 'last']
print('\nSplit name:')
print(name_parts)
```

```python
# replace — regex supported
# Standardize phone numbers
phones_clean = (df['phone']
    .str.replace(r'[\(\)\.\s\-]', '', regex=True)  # remove non-digit chars
)
print('Cleaned phones:', phones_clean.values)
```

```python
# extract — regex groups
# Extract years of experience from bio
yoe = df['bio'].str.extract(r'(\d+)\s*yr')
print('Extracted years of experience:')
print(yoe)

# Extract multiple groups
email_parts = df['email'].str.extract(r'([^@]+)@([^.]+)\.(.+)')
email_parts.columns = ['user', 'provider', 'tld']
print('\nEmail parts:')
print(email_parts)
```

```python
# String padding, zero-fill
ids = pd.Series([1, 23, 456, 7890])
print('zfill (6 chars):', ids.astype(str).str.zfill(6).values)
print('rjust:', ids.astype(str).str.rjust(8, ' ').values)
```

```python
# EXCEPTION: str methods on NaN propagate NaN
s = pd.Series(['hello', None, 'world'])
print('str.upper with NaN:', s.str.upper().values)
print('str.len with NaN:', s.str.len().values)  # NaN preserved

# EXCEPTION: wrong regex
try:
    df['email'].str.contains('[invalid', regex=True)
except Exception as e:
    print('Error:', type(e).__name__, str(e)[:60])
```
