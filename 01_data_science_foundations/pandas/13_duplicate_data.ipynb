# Duplicate Data

## Definition
**Duplicate rows** are rows where all (or a subset of) column values are identical.
In ML, duplicates can:
- **Bias your model** — duplicate training samples have inflated influence
- **Inflate metrics** — if test set has duplicates from train set → data leakage
- **Waste computation** — processing the same data multiple times

### Key Functions
| Function | Description |
|----------|-------------|
| `df.duplicated()` | Boolean mask — True for duplicate rows |
| `df.drop_duplicates()` | Remove duplicate rows |
| `df.duplicated(subset=[...])` | Check duplicates on specific columns |
| `df['col'].duplicated()` | Duplicates in a single column |
| `df.value_counts()` | See frequency of each row |

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'id':   [1, 2, 3, 2, 4, 1, 5],
    'name': ['Alice', 'Bob', 'Carol', 'Bob', 'Dave', 'Alice', 'Eve'],
    'score': [85, 72, 91, 72, 68, 85, 88]
})
print(df)
```

```python
# Detecting duplicates
print('duplicated() mask:')
print(df.duplicated())  # True for all duplicates EXCEPT the first occurrence

print('\nTotal duplicates:', df.duplicated().sum())
print('\nDuplicate rows:')
print(df[df.duplicated(keep=False)])  # show ALL copies of duplicated rows
```

```python
# Different keep options
print('keep=first (default) — mark all but FIRST as duplicate:')
print(df.duplicated(keep='first').values)

print('keep=last — mark all but LAST as duplicate:')
print(df.duplicated(keep='last').values)

print('keep=False — mark ALL occurrences as duplicate:')
print(df.duplicated(keep=False).values)
```

```python
# drop_duplicates()
print('After drop_duplicates (keep first):')
print(df.drop_duplicates())

print('\nAfter drop_duplicates (keep last):')
print(df.drop_duplicates(keep='last'))
```

```python
# Check duplicates on subset of columns
# e.g., same id and name but different score
print('Duplicates on [id, name]:')
print(df.duplicated(subset=['id', 'name'], keep=False))

# Keep only unique (id, name) pairs
df_dedup = df.drop_duplicates(subset=['id', 'name'])
print('\nAfter dedup on [id, name]:')
print(df_dedup)
```

```python
# Real ML scenario: dedup before train/test split
data = pd.DataFrame({
    'feature': [1, 2, 3, 2, 4, 5, 3],
    'label':   [0, 1, 0, 1, 1, 0, 0]
})
print('Before dedup:', data.shape)
data_clean = data.drop_duplicates()
print('After dedup:', data_clean.shape)
```

```python
# Finding duplicate values in a column
s = pd.Series([1, 2, 2, 3, 3, 3, 4])
print('Duplicated values:', s[s.duplicated(keep=False)].unique())
print('Value counts:\n', s.value_counts())
```
