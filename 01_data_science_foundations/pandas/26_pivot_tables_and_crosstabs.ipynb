# Pivot Tables & Crosstabs

## Definition
**Pivot tables** summarize and reorganize data by grouping on one axis (rows) and another (columns),
similar to Excel pivot tables.

**Cross-tabulations (crosstabs)** compute frequency tables of two or more factors.

### Functions
| Function | Description |
|----------|-------------|
| `df.pivot_table()` | Full-featured pivot with aggregation |
| `df.pivot()` | Simple reshape (no aggregation) |
| `pd.crosstab()` | Frequency table (count by default) |

### pivot_table() Parameters
| Param | Description |
|-------|-------------|
| `values` | Column(s) to aggregate |
| `index` | Row grouping |
| `columns` | Column grouping |
| `aggfunc` | Aggregation function(s) |
| `fill_value` | Replace NaN in result |
| `margins` | Add row/col totals |

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'dept':    ['Eng', 'Mkt', 'Eng', 'HR', 'Eng', 'Mkt', 'HR', 'Mkt'],
    'level':   ['Sr', 'Jr', 'Jr', 'Sr', 'Sr', 'Sr', 'Jr', 'Jr'],
    'gender':  ['M', 'F', 'F', 'M', 'M', 'F', 'F', 'M'],
    'salary':  [90000, 65000, 75000, 85000, 95000, 80000, 70000, 72000],
    'score':   [88, 72, 91, 75, 84, 78, 69, 74],
    'name':    ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank', 'Grace', 'Henry']
})
print(df)
```

```python
# Basic pivot table: mean salary by dept x level
pt = df.pivot_table(values='salary', index='dept', columns='level', aggfunc='mean')
print('Mean salary by dept x level:')
print(pt)
```

```python
# Multiple aggregation functions
pt2 = df.pivot_table(values='salary', index='dept', columns='level',
                     aggfunc=['mean', 'count'], fill_value=0)
print('Multiple agg functions:')
print(pt2)
```

```python
# Margins — add row and column totals
pt3 = df.pivot_table(values='salary', index='dept', columns='level',
                     aggfunc='mean', margins=True, margins_name='Total',
                     fill_value=0)
print('With margins:')
print(pt3.round(0))
```

```python
# Multiple values
pt4 = df.pivot_table(values=['salary', 'score'], index='dept', columns='gender',
                     aggfunc='mean', fill_value=0)
print('Salary and score by dept x gender:')
print(pt4.round(1))
```

```python
# pd.crosstab() — frequency count table
ct = pd.crosstab(df['dept'], df['level'])
print('Crosstab (count):')
print(ct)
```

```python
# crosstab with normalize
print('Row proportions:')
print(pd.crosstab(df['dept'], df['gender'], normalize='index').round(3))

print('\nWith margins:')
print(pd.crosstab(df['dept'], df['level'], margins=True))
```

```python
# crosstab with values and aggfunc
print('Mean salary in crosstab:')
print(pd.crosstab(df['dept'], df['level'], values=df['salary'], aggfunc='mean').round(0))
```
