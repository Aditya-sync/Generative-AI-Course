# isna(), notna(), fillna(), dropna()

## Definition
These four functions form the **core toolkit** for handling missing data in pandas.

| Function | Description |
|----------|-------------|
| `isna()` / `isnull()` | Returns boolean DataFrame/Series (True = missing) |
| `notna()` / `notnull()` | Inverse of isna() |
| `fillna(value/method)` | Replace NaN with a value or strategy |
| `dropna(axis, how, thresh)` | Remove rows or columns with NaN |

### fillna() methods
| Method | Description |
|--------|-------------|
| `value=X` | Fill with constant |
| `method='ffill'` | Forward fill (propagate last valid) |
| `method='bfill'` | Backward fill (propagate next valid) |
| `df.mean()` | Fill with column means |
| `df.groupby('g').transform('mean')` | Group-aware mean fill |

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1.0, np.nan, 3.0, np.nan, 5.0],
    'B': [np.nan, 2.0, np.nan, 4.0, 5.0],
    'C': ['x', None, 'y', 'z', None]
})
print(df)
```

```python
# isna() / isnull() — identical
print('isna():\n', df.isna())
print('\nnull count:\n', df.isna().sum())
```

```python
# notna() — select valid rows
print('Rows where A is not NaN:')
print(df[df['A'].notna()])
```

```python
# fillna() — fill with scalar
df1 = df.copy()
df1['A'] = df1['A'].fillna(0)
df1['C'] = df1['C'].fillna('missing')
print('After scalar fillna:')
print(df1)
```

```python
# fillna() — fill with dict (different values per column)
df2 = df.copy()
df2 = df2.fillna({'A': df2['A'].mean(), 'B': df2['B'].median(), 'C': 'unknown'})
print('Dict fillna:')
print(df2)
```

```python
# fillna() with method (deprecated in new pandas, use ffill/bfill directly)
s = pd.Series([1.0, np.nan, np.nan, 4.0, np.nan, 6.0])
print('ffill:', s.ffill().values)
print('bfill:', s.bfill().values)
print('ffill limit=1:', s.ffill(limit=1).values)  # fill max 1 consecutive NaN
```

```python
# dropna() — detailed control
df3 = pd.DataFrame({
    'A': [1, np.nan, 3],
    'B': [np.nan, np.nan, 6],
    'C': [7, 8, 9]
})
print('Original:')
print(df3)
print('\ndropna() (any NaN in row):')
print(df3.dropna())
print('\ndropna(how=all):')
print(df3.dropna(how='all'))
print('\ndropna(subset=[A]):')
print(df3.dropna(subset=['A']))
print('\ndropna(axis=1, how=any) — drop columns:')
print(df3.dropna(axis=1, how='any'))
print('\ndropna(thresh=2) — keep rows with >=2 non-null:')
print(df3.dropna(thresh=2))
```

```python
# Group-aware imputation (ML best practice!)
df_group = pd.DataFrame({
    'dept': ['Eng', 'Eng', 'Mkt', 'Mkt', 'HR', 'HR'],
    'salary': [80000, np.nan, 70000, np.nan, 60000, np.nan]
})

# Fill salary NaN with department mean (better than global mean)
df_group['salary_filled'] = df_group.groupby('dept')['salary'].transform(
    lambda x: x.fillna(x.mean())
)
print(df_group)
```
