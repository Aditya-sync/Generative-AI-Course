# groupby()

## Definition
`groupby()` is one of the most powerful pandas operations. It follows the **split-apply-combine** pattern:
1. **Split** the data into groups based on some criteria
2. **Apply** a function to each group independently
3. **Combine** the results back into a single DataFrame

### Split → Apply → Combine
```
dept = [Eng, Mkt, Eng, HR, Mkt]
salary = [80k, 70k, 90k, 60k, 85k]
  ↓ groupby('dept')
Eng: [80k, 90k]  →  mean → 85k
Mkt: [70k, 85k]  →  mean → 77.5k
HR:  [60k]       →  mean → 60k
  ↓ combine
{Eng: 85k, Mkt: 77.5k, HR: 60k}
```

### Types of GroupBy Operations
| Type | Function | Description |
|------|----------|-------------|
| Aggregation | `agg()`, `sum()`, `mean()`, `count()` | Reduce group to scalar |
| Transformation | `transform()` | Return same-length result per group |
| Filtration | `filter()` | Keep/discard entire groups |
| Application | `apply()` | Custom function per group |
| Iteration | `for name, group in df.groupby(...)` | Iterate over groups |

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name':   ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank', 'Grace', 'Henry'],
    'dept':   ['Eng', 'Mkt', 'Eng', 'HR', 'Eng', 'Mkt', 'HR', 'Eng'],
    'level':  ['Sr', 'Jr', 'Jr', 'Sr', 'Sr', 'Sr', 'Jr', 'Jr'],
    'salary': [90000, 65000, 75000, 85000, 95000, 80000, 70000, 70000],
    'score':  [88, 72, 91, 75, 84, 78, 69, 82]
})
print(df)
````

````python
# Basic groupby + aggregation
print('Mean salary by dept:')
print(df.groupby('dept')['salary'].mean().sort_values(ascending=False))
````

````python
# Multiple aggregation functions
print('Salary stats by dept:')
print(df.groupby('dept')['salary'].agg(['mean', 'median', 'std', 'count']))
````

````python
# Group by multiple columns
print('Mean salary by dept AND level:')
print(df.groupby(['dept', 'level'])['salary'].mean())
````

````python
# Aggregate multiple columns simultaneously
print('Salary and score stats by dept:')
print(df.groupby('dept').agg(
    avg_salary=('salary', 'mean'),
    max_score=('score', 'max'),
    employee_count=('name', 'count')
).round(0))
````

````python
# Iterating over groups
print('Group sizes:')
for name, group in df.groupby('dept'):
    print(f'  {name}: {len(group)} employees, avg salary: {group["salary"].mean():.0f}')
````

````python
# Get specific group
eng_group = df.groupby('dept').get_group('Eng')
print('Engineering group:')
print(eng_group)
````

````python
# groupby with custom aggregation using agg() dict
custom = df.groupby('dept').agg(
    n_employees=('name', 'count'),
    avg_salary=('salary', lambda x: x.mean()),
    salary_spread=('salary', lambda x: x.max() - x.min()),
    top_score=('score', 'max')
)
print('Custom aggregation:')
print(custom)
````

````python
# EXCEPTION: groupby result index
gb_result = df.groupby('dept')['salary'].mean()
print('GroupBy result index:', gb_result.index.tolist())
# Reset index to make dept a column again
print('\nAfter reset_index:')
print(gb_result.reset_index())
````
