# Concatenation — concat()

## Definition
`pd.concat()` joins **multiple DataFrames** along an axis (rows or columns).
It is the simplest way to combine DataFrames when they have **similar structure**.

### concat() vs merge()
| | `concat()` | `merge()` |
|--|------------|-----------|
| Combines on | Shared index or column | Matching key values |
| Use for | Appending rows, side-by-side | SQL-style joins |
| Alignment | By index/position | By key column |

### Key Parameters
| Parameter | Default | Description |
|-----------|---------|-------------|
| `axis` | 0 | 0=row-wise, 1=column-wise |
| `join` | 'outer' | 'outer' or 'inner' |
| `ignore_index` | False | Reset index 0,1,2,... |
| `keys` | None | Create hierarchical index |
| `sort` | False | Sort columns/index |
| `verify_integrity` | False | Check for duplicate indices |

```python
import pandas as pd
import numpy as np

# Three DataFrames representing monthly data
jan = pd.DataFrame({'name': ['Alice', 'Bob'],   'sales': [100, 150], 'month': 'Jan'})
feb = pd.DataFrame({'name': ['Carol', 'Dave'],  'sales': [200, 130], 'month': 'Feb'})
mar = pd.DataFrame({'name': ['Eve', 'Frank'],   'sales': [175, 210], 'month': 'Mar'})

print('Jan:'); print(jan)
print('Feb:'); print(feb)
```

```python
# Row-wise concat (axis=0) — stacking vertically
all_months = pd.concat([jan, feb, mar], axis=0)
print('Row concat:')
print(all_months)
print('Note: duplicate index values (0, 1)!')
```

```python
# ignore_index=True — reset to 0,1,2,...
all_months = pd.concat([jan, feb, mar], axis=0, ignore_index=True)
print('With ignore_index:')
print(all_months)
```

```python
# keys — create hierarchical index for traceability
all_months_keyed = pd.concat([jan, feb, mar], keys=['Jan', 'Feb', 'Mar'])
print('With keys (MultiIndex):')
print(all_months_keyed)
print('\nAccess Jan data:')
print(all_months_keyed.loc['Jan'])
```

```python
# Column-wise concat (axis=1) — joining side by side
df_names = pd.DataFrame({'name': ['Alice', 'Bob', 'Carol']})
df_scores = pd.DataFrame({'math': [85, 72, 91], 'science': [88, 78, 94]})
df_grades = pd.DataFrame({'grade': ['B', 'C', 'A']})

combined = pd.concat([df_names, df_scores, df_grades], axis=1)
print('Column concat:')
print(combined)
```

```python
# join='inner' vs 'outer' (when columns differ)
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'B': [5, 6], 'C': [7, 8]})

outer = pd.concat([df1, df2], axis=0, join='outer', ignore_index=True)
inner = pd.concat([df1, df2], axis=0, join='inner', ignore_index=True)

print('Outer join (NaN for missing):')
print(outer)
print('Inner join (only common cols):')
print(inner)
```

```python
# Building a dataset incrementally with concat
results = []
for i in range(3):
    batch = pd.DataFrame({'batch': [i], 'value': [i*10], 'score': [i*0.1]})
    results.append(batch)

final = pd.concat(results, ignore_index=True)
print('Concatenated batches:')
print(final)
```

```python
# EXCEPTION: verify_integrity catches duplicate indices
df1 = pd.DataFrame({'A': [1, 2]}, index=[0, 1])
df2 = pd.DataFrame({'A': [3, 4]}, index=[0, 1])  # same indices!

try:
    pd.concat([df1, df2], verify_integrity=True)
except ValueError as e:
    print('ValueError:', str(e)[:80])

# Default: silently allows duplicates
result = pd.concat([df1, df2])
print('Without verify_integrity:', result.index.tolist())
```
