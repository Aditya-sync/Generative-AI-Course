# Replacing & Mapping Values

## Definition
**Replacing** substitutes specific values with others. **Mapping** transforms values using a dictionary or function.
Both are essential for feature engineering and data cleaning.

### Methods Overview
| Method | Best For |
|--------|----------|
| `df.replace(old, new)` | Substituting specific values |
| `df['col'].map(dict)` | Applying a lookup dict to a Series |
| `df['col'].map(func)` | Applying a Python function element-wise |
| `pd.cut()` | Bin continuous values into categories |
| `pd.qcut()` | Quantile-based binning |
| `np.where()` | Conditional replacement |
| `df.replace({'col': {old: new}})` | Column-specific replacement dict |

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'grade':    ['A', 'B', 'C', 'D', 'F', 'B', 'A', 'C'],
    'rating':   [5, 4, 3, 2, 1, 4, 5, 3],
    'gender':   ['M', 'F', 'M', 'male', 'female', 'F', 'Male', 'Female'],
    'score':    [95, 82, 71, 60, 45, 80, 92, 73],
    'outcome':  ['Yes', 'No', 'Yes', 'No', 'No', 'Yes', 'Yes', 'No']
})
print(df)
```

```python
# replace() — single value
df2 = df.copy()
df2['outcome'] = df2['outcome'].replace({'Yes': 1, 'No': 0})
print('After replace Yes/No:')
print(df2[['outcome']])
```

```python
# replace() — normalize messy gender data
gender_map = {'M': 'male', 'F': 'female', 'Male': 'male', 'male': 'male',
              'female': 'female', 'Female': 'female'}
df['gender'] = df['gender'].replace(gender_map)
print('Normalized gender:', df['gender'].unique())
```

```python
# replace with regex
s = pd.Series(['$100', '$200.50', '$3,000', 'N/A'])
s_clean = s.replace(r'[\$,]', '', regex=True).replace('N/A', np.nan)
print('After replace regex:', s_clean.values)
```

```python
# map() — Series to Series mapping
grade_to_gpa = {'A': 4.0, 'B': 3.0, 'C': 2.0, 'D': 1.0, 'F': 0.0}
df['gpa'] = df['grade'].map(grade_to_gpa)
print('GPA from grade:')
print(df[['grade', 'gpa']])
```

```python
# map() with a function
df['score_norm'] = df['score'].map(lambda x: (x - 50) / 50)
print('Normalized scores:', df['score_norm'].round(2).values)
```

```python
# pd.cut() — bin into equal-width intervals
df['score_bin'] = pd.cut(df['score'],
    bins=[0, 60, 70, 80, 90, 100],
    labels=['F', 'D', 'C', 'B', 'A'],
    right=True   # intervals are (left, right]
)
print('Score bins:')
print(df[['score', 'score_bin']])
```

```python
# pd.qcut() — quantile-based bins (equal frequency)
df['score_quartile'] = pd.qcut(df['score'], q=4,
    labels=['Q1', 'Q2', 'Q3', 'Q4'])
print('Score quartiles:')
print(df[['score', 'score_quartile']])
```

```python
# np.where for conditional replacement
df['pass_fail'] = np.where(df['score'] >= 60, 'Pass', 'Fail')
df['adjusted_score'] = np.where(df['score'] < 50, df['score'] * 1.1, df['score'])
print(df[['score', 'pass_fail', 'adjusted_score']])
```
