# Reshaping — pivot(), melt(), stack(), unstack()

## Definition
Reshaping changes the **structure** (wide ↔ long format) of a DataFrame without changing the data itself.

### Wide vs Long Format
```
WIDE (one column per variable):
   name    math  science  english
   Alice    85      92      78

LONG (one row per observation):
   name    subject    score
   Alice    math       85
   Alice    science    92
   Alice    english    78
```

### Functions
| Function | Direction | Description |
|----------|-----------|-------------|
| `pivot()` | Long → Wide | Reshape without aggregation |
| `melt()` | Wide → Long | Unpivot columns to rows |
| `stack()` | Wide → Long | Column level → row level |
| `unstack()` | Long → Wide | Row level → column level |

````python
import pandas as pd
import numpy as np

# Wide format
wide_df = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Carol'],
    'math':    [85, 72, 91],
    'science': [88, 78, 94],
    'english': [79, 85, 82]
})
print('Wide format:')
print(wide_df)
````

````python
# melt() — Wide to Long
long_df = wide_df.melt(
    id_vars='student',         # columns to keep as-is
    value_vars=['math', 'science', 'english'],  # columns to unpivot
    var_name='subject',        # name for the new variable column
    value_name='score'         # name for the new value column
)
print('Long format:')
print(long_df.sort_values('student'))
````

````python
# pivot() — Long to Wide (reverse of melt)
wide_again = long_df.pivot(
    index='student',    # become rows
    columns='subject',  # become columns
    values='score'      # fill values
)
wide_again.columns.name = None  # remove column index name
wide_again = wide_again.reset_index()
print('Back to wide:')
print(wide_again)
````

````python
# stack() — columns become row index level
df = pd.DataFrame({
    ('math', 'Q1'): [85, 72, 91],
    ('math', 'Q2'): [88, 75, 93],
    ('science', 'Q1'): [78, 80, 88],
    ('science', 'Q2'): [82, 77, 90]
}, index=['Alice', 'Bob', 'Carol'])
df.columns = pd.MultiIndex.from_tuples(df.columns)
print('Before stack:')
print(df)

stacked = df.stack(level=0)  # stack the subject level
print('\nAfter stack:')
print(stacked)
````

````python
# unstack() — row index level becomes columns
unstacked = stacked.unstack(level=0)  # unstack student level... no, unstack subject
print('After unstack:')
print(unstacked)
````

````python
# Real ML use: converting time-series to wide format for ML
ts_data = pd.DataFrame({
    'user_id': [1, 1, 1, 2, 2, 2, 3, 3, 3],
    'month':   ['Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar'],
    'purchases': [5, 3, 7, 2, 8, 4, 9, 1, 6]
})
print('Long format:')
print(ts_data)

# Pivot to wide — each month becomes a feature
wide_users = ts_data.pivot(index='user_id', columns='month', values='purchases')
wide_users.columns.name = None
print('\nWide format (for ML):')
print(wide_users)
````

````python
# melt with multiple id_vars
metadata_df = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Carol'],
    'q1_score': [85, 72, 91],
    'q2_score': [88, 75, 93],
    'q1_grade': ['B', 'C', 'A'],
    'q2_grade': ['B', 'C', 'A']
})
melted = metadata_df.melt(id_vars=['id', 'name'])
print('Multi id_vars melt:')
print(melted.head(8))
````
