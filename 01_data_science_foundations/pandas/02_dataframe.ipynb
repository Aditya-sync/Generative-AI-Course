# Pandas DataFrame

## Definition
A **DataFrame** is a 2-dimensional, size-mutable, labeled data structure with columns of potentially different data types.
Think of it as a table (like an Excel spreadsheet or SQL table) where:
- **Rows** are identified by an **index**
- **Columns** are identified by **column labels**
- Each **column** is a **Series** sharing the same index

```
       Name   Age   Score
  0   Alice    28    85.0
  1     Bob    34    72.0
  2   Carol    25    91.0
```

DataFrames are the primary data container in pandas — every ML dataset starts as a DataFrame.

````python
import pandas as pd
import numpy as np

# Create from dict of lists (most common)
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Carol', 'Dave', 'Eve'],
    'Age': [28, 34, 25, 31, 29],
    'Score': [85.0, 72.0, 91.0, 68.0, 88.0],
    'Passed': [True, True, True, False, True]
})
print(df)
````

````python
# DataFrame attributes
print('shape:', df.shape)        # (5, 4) = (rows, cols)
print('dtypes:\n', df.dtypes)
print('columns:', df.columns.tolist())
print('index:', df.index.tolist())
print('size:', df.size)
print('ndim:', df.ndim)
````

````python
# Create from list of dicts (each dict = one row)
records = [
    {'city': 'Mumbai', 'population': 20.7, 'country': 'India'},
    {'city': 'Delhi',  'population': 16.3, 'country': 'India'},
    {'city': 'NYC',    'population': 8.3,  'country': 'USA'},
]
cities_df = pd.DataFrame(records)
print(cities_df)
````

````python
# Create from NumPy array
arr = np.random.randint(0, 100, size=(4, 3))
df_np = pd.DataFrame(arr, columns=['X', 'Y', 'Z'], index=['r1', 'r2', 'r3', 'r4'])
print(df_np)
````

````python
# Column access (returns Series)
print(df['Name'])           # dict-like access
print(df.Age)               # attribute access (only works for simple column names)
print(type(df['Name']))     # <class 'pandas.core.series.Series'>
````

````python
# Multi-column selection (returns DataFrame)
subset = df[['Name', 'Score']]
print(subset)
print(type(subset))  # DataFrame
````

````python
# Row access
print('Row by label (index=1):')
print(df.loc[1])

print('\nRow by position (2nd row):')
print(df.iloc[1])
````

````python
# Adding new column
df['Grade'] = pd.cut(df['Score'], bins=[0, 60, 75, 90, 100],
                     labels=['F', 'C', 'B', 'A'])
print(df)
````

````python
# EXCEPTIONS
# TypeError: wrong type for column data
try:
    bad = pd.DataFrame({'a': [1, 2], 'b': [1, 2, 3]})  # mismatched lengths
except ValueError as e:
    print('ValueError:', e)

# KeyError: column doesn't exist
try:
    df['NonExistent']
except KeyError as e:
    print('KeyError:', e)
````
