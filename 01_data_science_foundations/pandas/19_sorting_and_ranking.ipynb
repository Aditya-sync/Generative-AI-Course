# Sorting & Ranking

## Definition
**Sorting** reorders rows/columns by value. **Ranking** assigns a rank (position) to each element.
Both are crucial for data analysis, feature engineering, and creating ranked metrics.

### Key Functions
| Function | Description |
|----------|-------------|
| `df.sort_values(by, ascending)` | Sort rows by one or more columns |
| `df.sort_index(axis, ascending)` | Sort by index |
| `df['col'].rank(method, ascending)` | Rank each element |
| `df.nlargest(n, col)` | Top N rows |
| `df.nsmallest(n, col)` | Bottom N rows |

### Rank Methods
| method | Description |
|--------|-------------|
| `average` | Tied values get average rank |
| `min` | Tied values get lowest rank |
| `max` | Tied values get highest rank |
| `first` | First occurrence gets lower rank |
| `dense` | Like min but ranks don't skip |

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name':   ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank'],
    'dept':   ['Eng', 'Mkt', 'Eng', 'HR', 'Eng', 'Mkt'],
    'salary': [80000, 75000, 95000, 65000, 80000, 90000],
    'yoe':    [5, 8, 3, 12, 5, 6]
})
print(df)
```

```python
# Sort by single column
print('Sort by salary (descending):')
print(df.sort_values('salary', ascending=False))
```

```python
# Sort by multiple columns
print('Sort by dept (asc), then salary (desc):')
print(df.sort_values(['dept', 'salary'], ascending=[True, False]))
```

```python
# Sort with NaN handling
df_na = df.copy()
df_na.loc[2, 'salary'] = np.nan
print('na_position=last (default):')
print(df_na.sort_values('salary', na_position='last'))
print('\nna_position=first:')
print(df_na.sort_values('salary', na_position='first'))
```

```python
# Ranking
df['salary_rank'] = df['salary'].rank(ascending=False, method='min')
print('Salary ranks:')
print(df[['name', 'salary', 'salary_rank']].sort_values('salary_rank'))
```

```python
# Rank with ties
tied = pd.Series([100, 90, 90, 80, 80, 70])
print('Tied values:', tied.values)
print('average rank:', tied.rank(method='average').values)
print('min rank:    ', tied.rank(method='min').values)
print('max rank:    ', tied.rank(method='max').values)
print('dense rank:  ', tied.rank(method='dense').values)
print('first rank:  ', tied.rank(method='first').values)
```

```python
# Percentile rank (pct=True)
df['salary_pct_rank'] = df['salary'].rank(pct=True)
print('Percentile ranks:')
print(df[['name', 'salary', 'salary_pct_rank']])
```

```python
# nlargest and nsmallest
print('Top 3 by salary:')
print(df.nlargest(3, 'salary')[['name', 'salary']])

print('\nBottom 2 by yoe:')
print(df.nsmallest(2, 'yoe')[['name', 'yoe']])
```

```python
# Group-wise ranking (e.g., rank within each dept)
df['rank_in_dept'] = df.groupby('dept')['salary'].rank(ascending=False, method='min')
print('Rank within department:')
print(df.sort_values(['dept', 'rank_in_dept']))
```
