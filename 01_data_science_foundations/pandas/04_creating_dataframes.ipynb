# Creating DataFrames

## Definition
There are many ways to construct a DataFrame in pandas. Knowing all of them is essential because
you'll encounter data in many different formats throughout an ML project.

### Creation Methods
| Method | When to Use |
|--------|-------------|
| From dict of lists | Most common — columns as keys |
| From list of dicts | When data comes as records |
| From NumPy array | When data is already numeric |
| From CSV/Excel/JSON | Loading from file (next notebook) |
| From scratch + append | Building incrementally |
| `pd.DataFrame.from_records()` | Named tuples, list of tuples |
| `pd.DataFrame.from_dict()` | Dict with `orient` options |

```python
import pandas as pd
import numpy as np

# 1. From dict of lists (most common)
df1 = pd.DataFrame({
    'product': ['Laptop', 'Phone', 'Tablet', 'Watch'],
    'price':   [999, 699, 499, 299],
    'units':   [50, 120, 80, 200],
    'in_stock': [True, True, False, True]
})
print('From dict:\n', df1)
```

```python
# 2. From list of dicts (each dict = one row)
records = [
    {'id': 1, 'name': 'Alice', 'score': 85},
    {'id': 2, 'name': 'Bob',   'score': 72},
    {'id': 3, 'name': 'Carol', 'score': 91},
]
df2 = pd.DataFrame(records)
print('From list of dicts:\n', df2)
```

```python
# 3. From NumPy array
rng = np.random.default_rng(42)
arr = rng.integers(0, 100, size=(5, 4))
df3 = pd.DataFrame(arr,
    columns=['feat_1', 'feat_2', 'feat_3', 'feat_4'],
    index=[f'sample_{i}' for i in range(5)])
print('From NumPy array:\n', df3)
```

```python
# 4. From list of tuples
tuples = [('Alice', 28, 'NY'), ('Bob', 34, 'CA'), ('Carol', 25, 'TX')]
df4 = pd.DataFrame(tuples, columns=['Name', 'Age', 'State'])
print('From tuples:\n', df4)
```

```python
# 5. pd.DataFrame.from_dict with orient options
# orient='index': keys are row labels
data = {
    'row0': {'a': 1, 'b': 2},
    'row1': {'a': 3, 'b': 4},
    'row2': {'a': 5, 'b': 6},
}
df5 = pd.DataFrame.from_dict(data, orient='index')
print('from_dict orient=index:\n', df5)
```

```python
# 6. Empty DataFrame with defined columns (then fill)
df_empty = pd.DataFrame(columns=['id', 'name', 'value'])
# Append rows (use pd.concat, not .append which is deprecated)
new_rows = pd.DataFrame([{'id': 1, 'name': 'test', 'value': 42}])
df_filled = pd.concat([df_empty, new_rows], ignore_index=True)
print('After concat:\n', df_filled)
```

```python
# 7. Using pd.RangeIndex and repeating values
df_template = pd.DataFrame({
    'group': np.repeat(['A', 'B', 'C'], 3),
    'value': np.random.randn(9).round(2),
    'weight': np.ones(9)
})
print('Template DataFrame:\n', df_template)
```

```python
# EXCEPTION: mismatched list lengths
try:
    df_bad = pd.DataFrame({'a': [1, 2, 3], 'b': [10, 20]})
except ValueError as e:
    print('ValueError:', e)
```
