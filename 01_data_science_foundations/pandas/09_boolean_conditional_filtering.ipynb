# Boolean / Conditional Filtering

## Definition
Boolean filtering in pandas uses **boolean masks** (Series of True/False values)
to select rows that satisfy one or more conditions.

This is how you implement data quality checks, feature selection, and class filtering in ML pipelines.

### Operators
| Operator | Meaning | Notes |
|----------|---------|-------|
| `&` | AND | Use instead of `and` |
| `|` | OR | Use instead of `or` |
| `~` | NOT | Use instead of `not` |
| `==`, `!=` | Equal/Not equal | — |
| `<`, `>`, `<=`, `>=` | Comparisons | — |
| `.isin([...])` | Membership test | — |
| `.between(lo, hi)` | Range check | — |
| `.str.contains()` | String match | — |
| `df.query()` | SQL-like string | — |

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
df = pd.DataFrame({
    'name':   ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank', 'Grace'],
    'age':    [28, 34, 25, 45, 29, 52, 31],
    'salary': [75000, 85000, 65000, 95000, 70000, 110000, 80000],
    'dept':   ['Eng', 'Mkt', 'Eng', 'HR', 'Eng', 'Mkt', 'HR'],
    'yoe':    [3, 8, 1, 18, 4, 25, 6]
})
print(df)
```

```python
# Simple condition
mask = df['age'] > 30
print('age > 30:')
print(df[mask])
```

```python
# Combined conditions with & and |
# MUST use () around each condition!
print('Engineering dept AND salary > 70000:')
print(df[(df['dept'] == 'Eng') & (df['salary'] > 70000)])

print('\nAge < 30 OR salary > 100000:')
print(df[(df['age'] < 30) | (df['salary'] > 100000)])
```

```python
# NOT operator ~
print('NOT in Engineering:')
print(df[~(df['dept'] == 'Eng')])

# Equivalent
print('\nNot Engineering (using !=):')
print(df[df['dept'] != 'Eng'])
```

```python
# .isin() — check membership in a list
print('Dept in [Eng, HR]:')
print(df[df['dept'].isin(['Eng', 'HR'])])

print('\nNOT in [Eng, HR]:')
print(df[~df['dept'].isin(['Eng', 'HR'])])
```

```python
# .between() — range inclusive check
print('Age between 28 and 35 (inclusive):')
print(df[df['age'].between(28, 35)])

# Equivalent: (df['age'] >= 28) & (df['age'] <= 35)
```

```python
# Multiple conditions — complex filter
print('Eng or HR dept, age < 35, salary > 65000:')
result = df[
    df['dept'].isin(['Eng', 'HR']) &
    (df['age'] < 35) &
    (df['salary'] > 65000)
]
print(result)
```

```python
# df.query() — SQL-like syntax (great for readability)
print('query salary > 80000 and dept Mkt:')
print(df.query('salary > 80000 and dept == "Mkt"'))

# Using Python variable in query with @
min_salary = 75000
print(f'\nquery salary > {min_salary} (using variable):')
print(df.query('salary > @min_salary'))
```

```python
# EXCEPTION: using 'and' instead of '&'
try:
    df[(df['age'] > 25) and (df['salary'] > 70000)]
except ValueError as e:
    print('ValueError (use & not and):', e)

# EXCEPTION: missing parentheses
try:
    df[df['age'] > 25 & df['salary'] > 70000]  # & binds tighter than >!
except TypeError as e:
    print('TypeError (add parentheses):', e)
```
