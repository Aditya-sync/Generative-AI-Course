# Index and Column Concepts

## Definition
The **Index** in pandas is a special immutable array-like object that serves as the **row labels** of a DataFrame or Series.
- It enables **fast label-based lookups** (like a dictionary key)
- It aligns data automatically during operations
- Indices can be integers, strings, dates, MultiIndex, etc.

### Types of Indexes
| Type | Description |
|------|-------------|
| `RangeIndex` | Default 0,1,2,... integer index |
| `Int64Index` | Custom integer index |
| `Index` | String/object index |
| `DatetimeIndex` | Timestamps |
| `MultiIndex` | Hierarchical / compound index |
| `CategoricalIndex` | Categorical labels |

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Carol', 'Dave'],
    'Age': [28, 34, 25, 31],
    'Score': [85, 72, 91, 68]
})

print('Default RangeIndex:', df.index)
print('Columns:', df.columns)
print('Index dtype:', df.index.dtype)
print('Index values:', df.index.tolist())
```

```python
# Setting a custom index
df_custom = df.set_index('Name')
print(df_custom)
print('\nIndex:', df_custom.index.tolist())
print('Access by name:', df_custom.loc['Alice'])
```

```python
# Resetting index back to default
df_reset = df_custom.reset_index()
print(df_reset)
```

```python
# Index operations
idx = pd.Index([10, 20, 30, 40, 50])
print('Is monotonic increasing:', idx.is_monotonic_increasing)
print('Is unique:', idx.is_unique)

# Renaming index
df2 = df.rename(index={0: 'r0', 1: 'r1', 2: 'r2', 3: 'r3'})
print(df2)
```

```python
# Column operations
print('Column dtype:', df.columns.dtype)

# Rename columns
df_renamed = df.rename(columns={'Name': 'FullName', 'Score': 'Marks'})
print(df_renamed.columns.tolist())

# Reorder columns
df_reordered = df[['Score', 'Name', 'Age']]
print(df_reordered.head())
```

```python
# Index alignment during operations
s1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
s2 = pd.Series([10, 20, 30], index=['b', 'c', 'd'])

# pandas auto-aligns on index
print(s1 + s2)   # 'a' -> NaN, 'b' -> 22, 'c' -> 33, 'd' -> NaN
```

```python
# Checking index membership
df = pd.DataFrame({'val': [10, 20, 30]}, index=['x', 'y', 'z'])
print("'y' in index:", 'y' in df.index)
print("'w' in index:", 'w' in df.index)

# isin
mask = df.index.isin(['x', 'z'])
print(df[mask])
```

```python
# EXCEPTION: duplicate index causes issues
df_dup = pd.DataFrame({'val': [1, 2, 3]}, index=['a', 'a', 'b'])
print('is_unique:', df_dup.index.is_unique)

# loc on duplicate index returns ALL matching rows
print(df_dup.loc['a'])  # returns DataFrame, not Series!
```
