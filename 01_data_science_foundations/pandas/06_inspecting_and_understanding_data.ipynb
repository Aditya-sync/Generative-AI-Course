# Inspecting & Understanding Data

## Definition
Before any ML work, you must **understand your data**. Pandas provides a rich set of inspection functions
to quickly grasp the structure, content, statistics, and quality of a dataset.

### Essential Inspection Methods
| Method | Description |
|--------|-------------|
| `df.head(n)` | First n rows (default 5) |
| `df.tail(n)` | Last n rows |
| `df.shape` | (rows, cols) |
| `df.info()` | Column dtypes, non-null counts, memory |
| `df.describe()` | Statistical summary |
| `df.dtypes` | Data type per column |
| `df.columns` | Column names |
| `df.index` | Row index |
| `df.nunique()` | Count of unique values per column |
| `df.value_counts()` | Frequency of each unique value |
| `df.sample(n)` | Random sample of rows |
| `df.memory_usage()` | Memory per column |

```python
import pandas as pd
import numpy as np

# Create a realistic-ish sample dataset
rng = np.random.default_rng(42)
n = 200
df = pd.DataFrame({
    'id': range(1, n+1),
    'age': rng.integers(18, 65, n),
    'income': rng.normal(60000, 20000, n).round(2),
    'education': rng.choice(['High School', 'Bachelor', 'Master', 'PhD'], n,
                             p=[0.3, 0.4, 0.2, 0.1]),
    'score': rng.uniform(0, 100, n).round(1),
    'churned': rng.choice([0, 1], n, p=[0.8, 0.2])
})
# Inject some missing values
df.loc[rng.choice(n, 15, replace=False), 'income'] = np.nan
df.loc[rng.choice(n, 10, replace=False), 'education'] = np.nan
print('Dataset created:', df.shape)
```

```python
# head and tail
print('First 5 rows:')
print(df.head())
print('\nLast 3 rows:')
print(df.tail(3))
```

```python
# info() — single most useful function for initial inspection
df.info()
```

```python
# describe() — statistical summary
print('Numeric columns:')
print(df.describe())
print()
print('All columns including object:')
print(df.describe(include='all'))
```

```python
# Column-wise inspection
print('dtypes:')
print(df.dtypes)
print()
print('Unique values per column:')
print(df.nunique())
print()
print('Missing values:')
print(df.isna().sum())
print()
print('Missing % per column:')
print((df.isna().mean() * 100).round(2))
```

```python
# value_counts — frequency distribution
print('Education value counts:')
print(df['education'].value_counts())
print()
print('With normalize=True (proportions):')
print(df['education'].value_counts(normalize=True).round(3))
```

```python
# sample — random rows (great for sanity checks)
print('5 random rows:')
print(df.sample(5, random_state=42))
```

```python
# memory usage
print('Memory per column (bytes):')
print(df.memory_usage(deep=True))
print('\nTotal memory:', df.memory_usage(deep=True).sum() / 1024, 'KB')
```

```python
# Cross-tabulation — quick group inspection
print('Churned by Education:')
print(pd.crosstab(df['education'], df['churned'], margins=True))
```
