# Pandas Series

## Definition
A **Series** is a 1-dimensional labeled array capable of holding any data type (integers, floats, strings, Python objects, etc.).
It is the fundamental building block of a DataFrame — each column in a DataFrame is a Series.

```
index    values
  0   →   10
  1   →   20
  2   →   30
```

### Key Properties
| Property | Description |
|----------|-------------|
| `.values` | NumPy array of values |
| `.index` | Index labels |
| `.dtype` | Data type |
| `.name` | Series name |
| `.shape` | Tuple `(n,)` |
| `.size` | Number of elements |

````python
import pandas as pd
import numpy as np

# Create from Python list (default integer index 0, 1, 2, ...)
s = pd.Series([10, 20, 30, 40, 50])
print(s)
print('dtype:', s.dtype)
print('index:', s.index.tolist())
````

````python
# Create with custom index
s = pd.Series([85, 92, 78, 95], index=['Math', 'Science', 'English', 'History'])
print(s)
print()
print('Science score:', s['Science'])
print('Math score:', s.iloc[0])   # positional access
````

````python
# Create from dict (keys become index)
data = {'Alice': 28, 'Bob': 34, 'Carol': 25, 'Dave': 31}
ages = pd.Series(data, name='Age')
print(ages)
print('Name:', ages.name)
````

````python
# Create with scalar (broadcast)
s = pd.Series(5.0, index=['a', 'b', 'c', 'd'])
print(s)
````

````python
# Series operations — vectorized like NumPy
s = pd.Series([10, 20, 30, 40, 50])
print('s + 5:\n', s + 5)
print('s * 2:\n', s * 2)
print('s ** 2:\n', s ** 2)
print('np.sqrt(s):\n', np.sqrt(s))
````

````python
# Alignment by index — very powerful!
s1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
s2 = pd.Series([10, 20, 30], index=['b', 'c', 'd'])
print('s1 + s2:')
print(s1 + s2)  # 'a' and 'd' are NaN because they don't overlap
````

````python
# Useful Series methods
s = pd.Series([5, 3, 8, 1, 9, 2, 7, 4, 6])
print('max:', s.max())
print('min:', s.min())
print('mean:', s.mean())
print('std:', s.std())
print('sum:', s.sum())
print('sorted:\n', s.sort_values())
print('value_counts:\n', pd.Series([1,2,2,3,3,3]).value_counts())
````

````python
# Boolean indexing on Series
scores = pd.Series({'Alice': 85, 'Bob': 72, 'Carol': 91, 'Dave': 68, 'Eve': 88})
print('Scores above 80:')
print(scores[scores > 80])
````

````python
# Handling NaN in Series
s = pd.Series([1.0, np.nan, 3.0, np.nan, 5.0])
print('isna:', s.isna())
print('dropna:', s.dropna().values)
print('fillna(0):', s.fillna(0).values)
print('mean (ignores NaN):', s.mean())
````

````python
# EXCEPTIONS
s = pd.Series([10, 20, 30])

# KeyError: label not found
try:
    print(s['x'])
except KeyError as e:
    print('KeyError:', e)

# IndexError: positional index out of range
try:
    print(s.iloc[10])
except IndexError as e:
    print('IndexError:', e)

# TypeError: unsupported op
try:
    s + 'hello'
except TypeError as e:
    print('TypeError:', e)
````
