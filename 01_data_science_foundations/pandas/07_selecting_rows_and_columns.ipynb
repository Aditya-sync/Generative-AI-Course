# Selecting Rows & Columns

## Definition
Pandas provides multiple ways to select subsets of data. Understanding each is critical:

### Selection Methods Overview
| Method | Access By | Returns |
|--------|-----------|---------|
| `df['col']` | Column name | Series |
| `df[['c1','c2']]` | List of names | DataFrame |
| `df[bool_mask]` | Boolean array | DataFrame |
| `df.loc[rows, cols]` | Labels | Series or DataFrame |
| `df.iloc[rows, cols]` | Integer positions | Series or DataFrame |
| `df.at[row, col]` | Single label-based | Scalar |
| `df.iat[row, col]` | Single positional | Scalar |

### Golden rule
- Use **`.loc`** when you know the **label**
- Use **`.iloc`** when you know the **position**

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name':  ['Alice', 'Bob', 'Carol', 'Dave', 'Eve'],
    'age':   [28, 34, 25, 31, 29],
    'score': [85.0, 72.0, 91.0, 68.0, 88.0],
    'dept':  ['Eng', 'Mkt', 'Eng', 'HR', 'Eng']
}, index=['r0', 'r1', 'r2', 'r3', 'r4'])
print(df)
```

```python
# Column selection
print('Single column (Series):')
print(df['name'])

print('\nMultiple columns (DataFrame):')
print(df[['name', 'score']])
```

```python
# Row selection with boolean mask
print('Scores above 80:')
print(df[df['score'] > 80])

print('\nEngineering dept:')
print(df[df['dept'] == 'Eng'])
```

```python
# .loc — label-based selection
print('Single row by label:')
print(df.loc['r1'])

print('\nRows r0 to r2 (INCLUSIVE):')
print(df.loc['r0':'r2'])   # INCLUSIVE of endpoint!

print('\nSpecific rows and columns:')
print(df.loc[['r0', 'r2', 'r4'], ['name', 'score']])
```

```python
# .loc with conditions
print('Loc with boolean condition:')
print(df.loc[df['score'] > 80, ['name', 'score']])

# Modify via .loc (safe assignment)
df.loc[df['dept'] == 'Eng', 'score'] += 5
print('\nAfter bonus to Eng dept:')
print(df)
```

```python
# .iloc — position-based (integer) selection
print('First row:')
print(df.iloc[0])

print('\nRows 1 to 3 (EXCLUSIVE of 3):')
print(df.iloc[1:3])   # Python slice, NOT inclusive!

print('\nSpecific rows and cols by position:')
print(df.iloc[[0, 2, 4], [0, 2]])
```

```python
# .at and .iat — fast scalar access
print('df.at[r1, name]:', df.at['r1', 'name'])
print('df.iat[1, 0]:', df.iat[1, 0])

# Setting a single value
df.at['r0', 'score'] = 99.0
print('After modifying r0 score:', df.at['r0', 'score'])
```

```python
# Selecting with query() — SQL-like string syntax
print('query() for score > 80:')
print(df.query('score > 80'))

print('\nquery() with AND condition:')
print(df.query('score > 70 and dept == "Eng"'))
```

```python
# EXCEPTIONS
# KeyError: column doesn't exist
try:
    df['nonexistent']
except KeyError as e:
    print('KeyError:', e)

# IndexError: iloc out of range
try:
    df.iloc[100]
except IndexError as e:
    print('IndexError:', e)

# SettingWithCopyWarning simulation
import warnings
sub = df[df['dept'] == 'Eng']  # this is a view or copy
with warnings.catch_warnings(record=True) as w:
    warnings.simplefilter('always')
    sub['age'] = 30  # may trigger SettingWithCopyWarning!
    if w:
        print('Warning:', str(w[0].message)[:100])
# CORRECT: use .loc on original df
df.loc[df['dept'] == 'Eng', 'age'] = 30
```
