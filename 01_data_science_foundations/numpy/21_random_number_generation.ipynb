# Random Number Generation (numpy.random)

## Definition
NumPy provides a powerful module for generating pseudo-random numbers.
Pseudo-random means the numbers are generated by a deterministic algorithm seeded with an initial value.

### Two APIs (Modern vs Legacy)
| | Modern (recommended) | Legacy |
|--|---------------------|--------|
| Create | `rng = np.random.default_rng(seed)` | `np.random.seed(42)` |
| Uniform | `rng.random(size)` | `np.random.rand(...)` |
| Normal | `rng.standard_normal(size)` | `np.random.randn(...)` |
| Integers | `rng.integers(low, high, size)` | `np.random.randint(...)` |
| Shuffle | `rng.shuffle(arr)` | `np.random.shuffle(arr)` |
| Choice | `rng.choice(arr, size)` | `np.random.choice(...)` |

**Always use the modern API** (`np.random.default_rng`) for reproducibility and better statistical properties.

### Why seeding matters in ML
- Reproducible weight initialization
- Consistent train/test splits
- Debugging: reproduce exact same random state

```python
import numpy as np

# ─── Modern API ───
rng = np.random.default_rng(seed=42)

# Uniform [0, 1)
u = rng.random(size=(3, 4))
print('Uniform [0,1):\n', np.round(u, 3))

# Integers [low, high)
i = rng.integers(low=0, high=10, size=(2, 5))
print('Integers [0,10):\n', i)

# Standard normal (mean=0, std=1)
n = rng.standard_normal(size=(3, 3))
print('Standard normal:\n', np.round(n, 3))
```

```python
# Seeding for reproducibility
rng1 = np.random.default_rng(seed=123)
rng2 = np.random.default_rng(seed=123)

a = rng1.random(5)
b = rng2.random(5)
print('Same seed -> same results:', np.allclose(a, b))

rng3 = np.random.default_rng(seed=999)
c = rng3.random(5)
print('Different seed -> different results:', not np.allclose(a, c))
```

```python
# Common distributions
rng = np.random.default_rng(42)

# Uniform in custom range [low, high)
uni = rng.uniform(low=-1, high=1, size=5)
print('Uniform [-1, 1):', np.round(uni, 3))

# Normal with custom mean and std
norm = rng.normal(loc=5.0, scale=2.0, size=5)
print('Normal(mu=5, sigma=2):', np.round(norm, 3))

# Binomial (n=10 trials, p=0.3 success probability)
binom = rng.binomial(n=10, p=0.3, size=5)
print('Binomial(n=10, p=0.3):', binom)

# Poisson (lambda=3)
pois = rng.poisson(lam=3.0, size=5)
print('Poisson(lambda=3):', pois)
```

```python
# Shuffle and permutation
arr = np.arange(10)
rng = np.random.default_rng(42)

# shuffle — in-place
arr_copy = arr.copy()
rng.shuffle(arr_copy)
print('Shuffled:', arr_copy)

# permutation — returns new array (original unchanged)
perm = rng.permutation(arr)
print('Permutation:', perm)
print('Original unchanged:', arr)
```

```python
# np.random.choice — sample from array
rng = np.random.default_rng(42)
population = np.array(['cat', 'dog', 'bird', 'fish', 'hamster'])

# Without replacement
sample = rng.choice(population, size=3, replace=False)
print('Sample without replacement:', sample)

# With replacement (for bootstrapping)
bootstrap = rng.choice(population, size=5, replace=True)
print('Bootstrap sample:', bootstrap)

# With custom probabilities
probs = np.array([0.5, 0.2, 0.1, 0.1, 0.1])
weighted = rng.choice(population, size=5, p=probs)
print('Weighted sample:', weighted)
```

```python
# ML use case: Mini-batch sampling
rng = np.random.default_rng(42)
N = 1000    # dataset size
batch_size = 32

# Method 1: random indices
idx = rng.integers(0, N, size=batch_size)

# Method 2: shuffle and take first batch_size
all_idx = np.arange(N)
rng.shuffle(all_idx)
batch_idx = all_idx[:batch_size]
print('Batch indices shape:', batch_idx.shape)
print('No duplicates in epoch shuffled batch:', len(np.unique(batch_idx)) == batch_size)
```

```python
# Xavier (Glorot) weight initialization
def xavier_uniform(fan_in, fan_out, rng):
    limit = np.sqrt(6.0 / (fan_in + fan_out))
    return rng.uniform(-limit, limit, size=(fan_in, fan_out))

def xavier_normal(fan_in, fan_out, rng):
    std = np.sqrt(2.0 / (fan_in + fan_out))
    return rng.normal(0, std, size=(fan_in, fan_out))

rng = np.random.default_rng(42)
W1 = xavier_uniform(784, 256, rng)
W2 = xavier_normal(256, 10, rng)
print('Xavier uniform W1:', W1.shape, 'std:', np.round(np.std(W1), 4))
print('Xavier normal  W2:', W2.shape, 'std:', np.round(np.std(W2), 4))
```
