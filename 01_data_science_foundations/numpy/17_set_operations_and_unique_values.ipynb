# Set Operations & Unique Values

## Definition
NumPy treats 1D arrays as **sets** and provides standard set operations.
These work on sorted, unique elements.

### Set Operation Functions
| Function | Description |
|----------|-------------|
| `np.unique(arr)` | Unique sorted values |
| `np.union1d(a, b)` | Union (elements in a OR b) |
| `np.intersect1d(a, b)` | Intersection (elements in both) |
| `np.setdiff1d(a, b)` | Difference (in a but NOT in b) |
| `np.setxor1d(a, b)` | Symmetric difference (in one but not both) |
| `np.in1d(a, b)` | Boolean: is each element of a in b? |
| `np.isin(a, b)` | Like in1d but works on N-D arrays |

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])
b = np.array([4, 5, 6, 7, 8, 9])

print('a:', a)
print('b:', b)
print()
print('union1d:     ', np.union1d(a, b))       # all unique elements
print('intersect1d: ', np.intersect1d(a, b))   # in both
print('setdiff1d:   ', np.setdiff1d(a, b))     # in a, not in b
print('setdiff1d rev:', np.setdiff1d(b, a))    # in b, not in a
print('setxor1d:    ', np.setxor1d(a, b))      # in exactly one
```

```python
# np.unique — comprehensive example
arr = np.array([3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5])
print('Original:', arr)

# Just unique values
vals = np.unique(arr)
print('unique:', vals)

# Return indices of first occurrence
vals, idx = np.unique(arr, return_index=True)
print('unique values:', vals)
print('first occurrence idx:', idx)

# Return inverse (reconstruct original from unique)
vals, inv = np.unique(arr, return_inverse=True)
print('inverse map:', inv)
print('reconstructed:', vals[inv])  # same as original

# Return counts
vals, counts = np.unique(arr, return_counts=True)
print('counts:', dict(zip(vals, counts)))
```

```python
# np.isin — test membership (works on N-D arrays!)
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
test_vals = [2, 4, 6]

mask = np.isin(arr, test_vals)
print('isin mask:\n', mask)
print('Selected elements:', arr[mask])
```

```python
# np.in1d — 1D membership test
haystack = np.array([1, 5, 9, 13, 17, 21])
needles  = np.array([5, 7, 13, 20])
print('Needles in haystack:', np.in1d(needles, haystack))
```

```python
# Practical: Remove duplicates from a dataset
# Find unique rows in 2D array
data = np.array([[1, 2, 3],
                 [4, 5, 6],
                 [1, 2, 3],
                 [7, 8, 9],
                 [4, 5, 6]])

unique_rows = np.unique(data, axis=0)
print('Original rows:', data.shape[0])
print('Unique rows:', unique_rows.shape[0])
print(unique_rows)
```

```python
# Practical: Vocabulary intersection for NLP
vocab1 = np.array(['apple', 'banana', 'cherry', 'date', 'fig'])
vocab2 = np.array(['banana', 'date', 'grape', 'kiwi'])

common = np.intersect1d(vocab1, vocab2)
only_v1 = np.setdiff1d(vocab1, vocab2)
print('Common words:', common)
print('Only in vocab1:', only_v1)
```
