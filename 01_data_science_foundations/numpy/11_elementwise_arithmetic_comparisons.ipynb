# Element-wise Arithmetic & Comparisons

## Definition
NumPy performs arithmetic operations **element-by-element** — no loops needed.
This is called **vectorization** and is the foundation of fast numerical computing in Python.

### Arithmetic Operators
| Operator | NumPy Equivalent | Description |
|----------|-----------------|-------------|
| `+` | `np.add` | Addition |
| `-` | `np.subtract` | Subtraction |
| `*` | `np.multiply` | Multiplication |
| `/` | `np.divide` | Division (float) |
| `//` | `np.floor_divide` | Integer division |
| `%` | `np.mod` | Modulo |
| `**` | `np.power` | Exponentiation |

### Comparison Operators → boolean arrays
| Operator | NumPy Equivalent |
|----------|-----------------|
| `==` | `np.equal` |
| `!=` | `np.not_equal` |
| `<` | `np.less` |
| `<=` | `np.less_equal` |
| `>` | `np.greater` |
| `>=` | `np.greater_equal` |

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print('a + b:', a + b)
print('a - b:', a - b)
print('a * b:', a * b)
print('b / a:', b / a)
print('b // a:', b // a)   # floor division
print('b % a:', b % a)     # modulo
print('a ** 2:', a ** 2)   # element-wise power
```

```python
# Scalar operations (broadcasting against scalar)
arr = np.array([1, 2, 3, 4, 5])
print('arr + 10:', arr + 10)
print('arr * 3:', arr * 3)
print('arr / 2:', arr / 2)
print('arr ** 2:', arr ** 2)
```

```python
# 2D element-wise operations
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('A + B:\n', A + B)
print('A * B (element-wise, NOT matrix mult):\n', A * B)
print('A @ B (matrix multiplication):\n', A @ B)  # different!
```

```python
# Comparison operators -> boolean arrays
a = np.array([1, 5, 3, 8, 2, 7])
b = np.array([3, 3, 3, 3, 3, 3])

print('a == 3:', a == 3)
print('a > b:', a > b)
print('a != 3:', a != 3)
print('a >= 5:', a >= 5)
```

```python
# Logical operators on boolean arrays
arr = np.array([1, 5, 3, 8, 2, 7])

mask1 = arr > 2
mask2 = arr < 6

print('AND (& ):', mask1 & mask2)
print('OR  (| ):', mask1 | mask2)
print('NOT (~  ):', ~mask1)
print('XOR (^ ):', mask1 ^ mask2)

# np.logical_* functions (equivalent)
print('logical_and:', np.logical_and(mask1, mask2))
```

```python
# In-place operations (modify array directly, no copy)
arr = np.array([1.0, 2.0, 3.0, 4.0])
arr += 10      # in-place addition
print('After +=10:', arr)
arr *= 2       # in-place multiply
print('After *=2:', arr)
arr **= 0.5    # in-place sqrt
print('After **=0.5:', arr)
```

```python
# Division edge cases
a = np.array([1.0, 0.0, -1.0, 0.0])
b = np.array([0.0, 0.0,  0.0, 1.0])

import warnings
with warnings.catch_warnings(record=True):
    warnings.simplefilter('always')
    result = a / b
print('1.0/0.0 -> inf, 0.0/0.0 -> nan:', result)

# Integer division by zero raises ZeroDivisionError differently
try:
    np.array([1]) // np.array([0])
except Exception as e:
    print(type(e).__name__, ':', e)
```

```python
# np.allclose — compare floating-point arrays with tolerance
a = np.array([0.1 + 0.2, 1.0])
b = np.array([0.3, 1.0])
print('== check:', a == b)            # may be False due to FP!
print('allclose:', np.allclose(a, b)) # True (within tolerance)
```
