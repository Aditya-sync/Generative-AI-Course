# Array Creation Methods

## Definition
NumPy provides many functions to create arrays without manually specifying each element.
Understanding these is essential because in AI/ML you rarely build arrays element-by-element —
you initialize weight matrices, create sequences, build identity matrices, etc.

| Function | Description |
|----------|-------------|
| `np.array()` | From Python list/tuple |
| `np.zeros()` | All zeros |
| `np.ones()` | All ones |
| `np.full()` | Constant fill value |
| `np.arange()` | Like Python `range()` |
| `np.linspace()` | Evenly spaced (N points) |
| `np.eye() / np.identity()` | Identity matrix |
| `np.random.*` | Random arrays |
| `np.empty()` | Uninitialized (fastest) |

```python
import numpy as np

# From list / tuple
a = np.array([1, 2, 3])
b = np.array((4.0, 5.0, 6.0))
print('from list:', a)
print('from tuple:', b)
```

```python
# zeros, ones, full
z = np.zeros((3, 4))          # 3x4 matrix of 0.0
o = np.ones((2, 3), dtype=int) # 2x3 matrix of 1 (int)
f = np.full((2, 2), 7)        # 2x2 filled with 7

print('zeros:\n', z)
print('ones:\n', o)
print('full:\n', f)
```

```python
# zeros_like, ones_like, full_like — match shape+dtype of existing array
existing = np.array([[1.0, 2.0], [3.0, 4.0]])
zl = np.zeros_like(existing)
ol = np.ones_like(existing)
print('zeros_like:\n', zl)
print('ones_like:\n', ol)
```

```python
# arange — integer / float ranges
r1 = np.arange(10)          # 0..9
r2 = np.arange(2, 20, 3)   # start=2, stop=20, step=3
r3 = np.arange(0.0, 1.0, 0.1)  # float range (prefer linspace instead)
print('arange 10:', r1)
print('arange 2-20 step 3:', r2)
print('arange float:', np.round(r3, 2))
```

```python
# linspace — N evenly spaced points between start and stop (INCLUSIVE)
ls = np.linspace(0, 1, 11)   # 11 points from 0 to 1
print('linspace:', ls)

# logspace — points evenly spaced on log scale
lgs = np.logspace(0, 3, 4)  # 10^0, 10^1, 10^2, 10^3
print('logspace:', lgs)
```

```python
# Identity / diagonal matrices
eye3 = np.eye(3)            # 3x3 identity
diag = np.diag([1, 2, 3])  # diagonal matrix from 1D array
tri  = np.tri(3, 4, k=0)   # lower triangular

print('eye(3):\n', eye3)
print('diag([1,2,3]):\n', diag)
print('tri(3,4):\n', tri)
```

```python
# Random creation
rng = np.random.default_rng(seed=42)  # modern recommended API
rand_uni  = rng.random((2, 3))        # uniform [0, 1)
rand_int  = rng.integers(0, 10, size=(2, 4))
rand_norm = rng.standard_normal((3, 3))
print('uniform:\n', rand_uni)
print('integers:\n', rand_int)
print('normal:\n', rand_norm)
```

```python
# empty — allocated but NOT initialized (fastest, use carefully)
e = np.empty((2, 3))
print('empty (garbage values):\n', e)
```

```python
# EXCEPTION: arange with float step can cause floating-point surprises
import numpy as np
r = np.arange(0, 1, 0.1)
print('float arange length:', len(r), '-> values:', r)
# Use linspace for reliable float ranges
# np.arange(0.0, 1.0, 0.3) may give 3 or 4 elements depending on FP errors
print('Safer: linspace(0, 1, 4):', np.linspace(0, 1, 4))
```
