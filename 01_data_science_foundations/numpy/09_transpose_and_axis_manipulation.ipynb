# Transpose and Axis Manipulation

## Definition
**Transpose** reorders the axes of an array. For 2D arrays, this swaps rows and columns (the classic matrix transpose).
For higher dimensions, you can specify any permutation of axes.

In deep learning:
- Switching between `(batch, C, H, W)` and `(batch, H, W, C)` for PyTorch vs TensorFlow
- Transposing weight matrices for matrix multiplication
- Rearranging sequence tensors `(seq, batch, features)` ↔ `(batch, seq, features)`

### Key Functions
| Function | Description |
|----------|-------------|
| `arr.T` | Reverse all axes (2D: row↔col) |
| `np.transpose(arr, axes)` | Custom axis permutation |
| `np.moveaxis(arr, src, dst)` | Move specific axes |
| `np.swapaxes(arr, a, b)` | Swap two axes |
| `np.rollaxis(arr, axis, start)` | Roll axis (legacy) |

```python
import numpy as np

# 2D Transpose
mat = np.array([[1, 2, 3],
                [4, 5, 6]])
print('Original shape:', mat.shape)   # (2, 3)
print('Transposed shape:', mat.T.shape)  # (3, 2)
print('Transposed:\n', mat.T)
```

```python
# 1D Transpose has no effect!
arr_1d = np.array([1, 2, 3, 4])
print('1D array shape:', arr_1d.shape)
print('1D .T shape:', arr_1d.T.shape)  # Still (4,) — no change!

# To create a column vector, use reshape or expand_dims
col = arr_1d[:, np.newaxis]
print('Column vector shape:', col.shape)  # (4, 1)
```

```python
# 3D Transpose with custom axis order
arr3d = np.ones((2, 3, 4))
print('Original shape:', arr3d.shape)  # (2, 3, 4)

# Reverse all axes (default)
print('.T shape:', arr3d.T.shape)  # (4, 3, 2)

# Custom permutation: move axis 0->2, 1->0, 2->1
t = np.transpose(arr3d, (1, 2, 0))
print('transpose(1,2,0) shape:', t.shape)  # (3, 4, 2)
```

```python
# Deep Learning use case: PyTorch (N,C,H,W) <-> TF (N,H,W,C)
pytorch_tensor = np.random.randn(8, 3, 224, 224)  # (batch, C, H, W)
print('PyTorch format:', pytorch_tensor.shape)

# Convert to TensorFlow format: (batch, H, W, C)
tf_tensor = np.transpose(pytorch_tensor, (0, 2, 3, 1))
print('TF format:', tf_tensor.shape)

# Convert back
back = np.transpose(tf_tensor, (0, 3, 1, 2))
print('Back to PyTorch:', back.shape)
```

```python
# np.moveaxis — easier to read than transpose for specific axes
arr = np.ones((2, 3, 4, 5))
print('Original:', arr.shape)

# Move axis 1 to position 3
moved = np.moveaxis(arr, 1, 3)
print('After moveaxis(1->3):', moved.shape)  # (2, 4, 5, 3)

# Move multiple axes at once
moved2 = np.moveaxis(arr, [0, 1], [-1, -2])
print('After moveaxis multiple:', moved2.shape)
```

```python
# np.swapaxes — swap exactly two axes
arr = np.ones((2, 3, 4))
print('Original:', arr.shape)
swapped = np.swapaxes(arr, 0, 2)
print('After swapaxes(0,2):', swapped.shape)  # (4, 3, 2)
```

```python
# Transpose returns a VIEW
mat = np.array([[1, 2, 3], [4, 5, 6]])
t = mat.T
t[0, 0] = 999
print('Original modified too:', mat)  # mat[0,0] = 999
```

```python
# EXCEPTION: non-square matrix multiplication after wrong transpose
a = np.ones((2, 3))
b = np.ones((2, 4))
try:
    result = a @ b      # (2,3) @ (2,4) incompatible
except ValueError as e:
    print('ValueError:', e)

# Correct: transpose b
result = a @ b.T   # wait... a=(2,3), b.T=(4,2) still incompatible
try:
    r = a @ b.T
except ValueError as e:
    print('ValueError after .T:', e)

# The right way: (2,3) @ (3,4)
c = np.ones((3, 4))
result = a @ c
print('Correct matmul shape:', result.shape)
```
