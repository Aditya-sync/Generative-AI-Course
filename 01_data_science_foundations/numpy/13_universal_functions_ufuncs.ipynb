# Universal Functions (ufuncs)

## Definition
A **ufunc** (Universal Function) is a NumPy function that operates **element-wise** on ndarrays.
They are implemented in C for maximum speed and support broadcasting, type casting, and in-place operations.

Every operator like `+`, `-`, `*` is backed by a ufunc under the hood.

### Categories of ufuncs
| Category | Examples |
|----------|---------|
| Math | `np.add`, `np.subtract`, `np.multiply`, `np.divide`, `np.power`, `np.sqrt`, `np.exp`, `np.log` |
| Trig | `np.sin`, `np.cos`, `np.tan`, `np.arcsin`, `np.arctan2` |
| Comparison | `np.maximum`, `np.minimum`, `np.fmax`, `np.fmin` |
| Bit ops | `np.bitwise_and`, `np.bitwise_or`, `np.left_shift` |
| Rounding | `np.floor`, `np.ceil`, `np.round_`, `np.trunc` |
| Logic | `np.isnan`, `np.isinf`, `np.isfinite` |

### ufunc methods
| Method | Description |
|--------|-------------|
| `.reduce(arr)` | Apply ufunc cumulatively along axis |
| `.accumulate(arr)` | Return running result |
| `.outer(a, b)` | Outer product |
| `.at(arr, idx, value)` | Apply ufunc at specific indices |

```python
import numpy as np

arr = np.array([1.0, 4.0, 9.0, 16.0, 25.0])

# Common math ufuncs
print('sqrt:', np.sqrt(arr))
print('exp:', np.exp(np.array([0, 1, 2])))
print('log:', np.log(arr))
print('log2:', np.log2(arr))
print('log10:', np.log10(arr))
print('abs:', np.abs(np.array([-3, -1, 0, 2, 4])))
```

```python
# Trigonometric ufuncs
angles = np.array([0, np.pi/6, np.pi/4, np.pi/3, np.pi/2])
print('sin:', np.round(np.sin(angles), 4))
print('cos:', np.round(np.cos(angles), 4))
print('tan:', np.round(np.tan(angles), 4))

# arctan2 — angle from x-axis (handles quadrant correctly)
print('arctan2(1,1):', np.degrees(np.arctan2(1, 1)))   # 45 degrees
```

```python
# Rounding ufuncs
arr = np.array([-2.7, -1.5, 0.3, 1.5, 2.9])
print('floor:', np.floor(arr))    # round down
print('ceil:', np.ceil(arr))      # round up
print('trunc:', np.trunc(arr))    # towards zero
print('round:', np.round(arr))    # round to nearest (banker's rounding)
```

```python
# np.maximum / np.minimum — element-wise max/min of two arrays
a = np.array([1, 5, 3, 8, 2])
b = np.array([3, 2, 6, 4, 7])

print('maximum:', np.maximum(a, b))   # [3, 5, 6, 8, 7]
print('minimum:', np.minimum(a, b))   # [1, 2, 3, 4, 2]

# vs np.max/np.min which reduce along axis
print('np.max(a):', np.max(a))  # 8 (single value)
```

```python
# ufunc.reduce — apply cumulatively
# np.add.reduce = sum; np.multiply.reduce = product
arr = np.array([1, 2, 3, 4, 5])
print('add.reduce (sum):', np.add.reduce(arr))          # 15
print('multiply.reduce (product):', np.multiply.reduce(arr))  # 120
print('maximum.reduce:', np.maximum.reduce(arr))         # 5
```

```python
# ufunc.accumulate — running result
arr = np.array([1, 2, 3, 4, 5])
print('add.accumulate (cumsum):', np.add.accumulate(arr))
print('multiply.accumulate (cumprod):', np.multiply.accumulate(arr))
```

```python
# ufunc.outer — outer product
a = np.array([1, 2, 3])
b = np.array([10, 20, 30, 40])
outer = np.multiply.outer(a, b)
print('multiply.outer shape:', outer.shape)  # (3, 4)
print(outer)
```

```python
# out parameter — store result in existing array (avoids allocation)
arr = np.array([1.0, 4.0, 9.0, 16.0])
result = np.empty_like(arr)
np.sqrt(arr, out=result)   # no new allocation!
print('Result stored in pre-allocated array:', result)
```

```python
# Custom ufunc with np.frompyfunc
def clamp_py(x, low, high):
    return min(max(x, low), high)

clamp = np.frompyfunc(clamp_py, 3, 1)  # 3 inputs, 1 output
arr = np.array([-5, 0, 3, 7, 12])
print('Clamped [0,5]:', clamp(arr, 0, 5))
```

```python
# ufuncs handle NaN specially
arr = np.array([1.0, np.nan, 3.0, np.nan, 5.0])
print('np.sqrt with nan:', np.sqrt(arr))
print('np.isnan:', np.isnan(arr))

# nan-safe ufuncs
print('np.nansum:', np.nansum(arr))
print('np.nanmax:', np.nanmax(arr))
```
