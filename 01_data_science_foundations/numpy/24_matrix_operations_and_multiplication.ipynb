# Matrix Operations & Matrix Multiplication

## Definition
Matrix multiplication is the **single most important operation** in deep learning.
Every layer of a neural network is essentially:
```
output = activation(W @ input + b)
```

### Matrix Multiplication Rules
For `A @ B`:
- A has shape `(m, k)` and B has shape `(k, n)` → result has shape `(m, n)`
- The **inner dimensions must match** (both `k`)
- Outer dimensions form the result shape

### Operations Covered
| Operation | How | Notes |
|-----------|-----|-------|
| Dot product | `a @ b`, `np.dot(a, b)` | 1D: scalar, 2D: matmul |
| Matrix multiply | `A @ B`, `np.matmul(A, B)` | No reduction for 3D+ |
| Batched matmul | `np.matmul(batch_A, batch_B)` | Broadcast batch dims |
| Hadamard (elem-wise) | `A * B` | Same shape required |
| Outer product | `np.outer(a, b)` | (m,) × (n,) = (m, n) |
| Kronecker product | `np.kron(A, B)` | Block matrix |

````python
import numpy as np

# Vector dot product
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print('1D @ 1D (scalar):', a @ b)      # 32

# Matrix-vector product
A = np.array([[1, 2, 3],
              [4, 5, 6]])   # (2, 3)
x = np.array([1, 1, 1])    # (3,)
print('2D @ 1D (vector):', A @ x)     # (2,)

# Matrix-matrix product
B = np.array([[1, 0],
              [0, 1],
              [1, 1]])     # (3, 2)
print('2D @ 2D (matrix):\n', A @ B)  # (2, 2)
````

````python
# np.dot vs np.matmul vs @
A = np.ones((2, 3))
B = np.ones((3, 4))

# All three are equivalent for 2D:
r1 = np.dot(A, B)
r2 = np.matmul(A, B)
r3 = A @ B
print('All equal:', np.all(r1 == r2) and np.all(r2 == r3))

# np.dot vs matmul differ for ND arrays!
# np.dot(A, B) for ND: sum over last axis of A, second-to-last of B
# np.matmul(A, B) for ND: treats as stack of 2D matrices
````

````python
# Batched matrix multiplication (critical for deep learning)
batch_size = 4
seq_len = 10
d_model = 64

# Q, K matrices in attention
Q = np.random.randn(batch_size, seq_len, d_model)  # (4, 10, 64)
K = np.random.randn(batch_size, seq_len, d_model)  # (4, 10, 64)

# Attention scores: Q @ K^T for each item in batch
# K.transpose(0, 2, 1) -> (4, 64, 10)
scores = Q @ K.transpose(0, 2, 1)
print('Attention scores shape:', scores.shape)  # (4, 10, 10)
````

````python
# Outer product — creates a matrix from two vectors
a = np.array([1, 2, 3])     # (3,)
b = np.array([10, 20, 30, 40])  # (4,)

outer = np.outer(a, b)
print('Outer product shape:', outer.shape)   # (3, 4)
print(outer)

# Equivalent to a[:, None] * b[None, :] (broadcasting)
print('Broadcasting equiv:\n', a[:, None] * b[None, :])
````

````python
# Hadamard product (element-wise) vs matrix multiply
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('Hadamard (*):\n', A * B)      # element-wise
print('Matrix mult (@):\n', A @ B)  # true matrix multiply
````

````python
# Efficient neural network forward pass using vectorized matmul
rng = np.random.default_rng(42)

def relu(x): return np.maximum(x, 0)
def sigmoid(x): return 1.0 / (1.0 + np.exp(-x))

# Random initialization
batch  = 32
W1 = rng.normal(0, 0.01, (784, 256))
b1 = np.zeros(256)
W2 = rng.normal(0, 0.01, (256, 10))
b2 = np.zeros(10)

# Mini-batch forward pass
X = rng.standard_normal((batch, 784))
H = relu(X @ W1 + b1)      # (32, 256)
Y = sigmoid(H @ W2 + b2)   # (32, 10)

print('Input:', X.shape)
print('Hidden:', H.shape)
print('Output:', Y.shape)
````

````python
# EXCEPTION: incompatible dimensions
A = np.ones((3, 4))
B = np.ones((3, 5))   # 4 != 3 — inner dims don't match
try:
    A @ B
except ValueError as e:
    print('ValueError:', e)
````
