# Reshaping and Flattening

## Definition
**Reshaping** changes the **shape** (arrangement of dimensions) of an array without changing its elements or total size.
**Flattening** collapses a multi-dimensional array into 1D.

These operations are fundamental in deep learning:
- Flattening image features before a fully-connected layer
- Reshaping batch tensors between conv and linear layers
- Adding/removing batch or channel dimensions

### Key Functions
| Function | Description |
|----------|-------------|
| `arr.reshape(shape)` | Return new shape view |
| `arr.ravel()` | 1D view (C-order) |
| `arr.flatten()` | 1D copy (always) |
| `np.expand_dims()` | Add a dimension |
| `np.squeeze()` | Remove size-1 dimensions |
| `arr.T` | Transpose (axes reversed) |

```python
import numpy as np

arr = np.arange(1, 13)  # [1, 2, ..., 12]
print('Original (1D):', arr)

# Reshape to 2D (3 rows x 4 cols)
mat = arr.reshape(3, 4)
print('\nReshape to (3,4):\n', mat)

# Reshape to 3D
cube = arr.reshape(2, 2, 3)
print('\nReshape to (2,2,3):\n', cube)
```

```python
# -1 as a wildcard dimension — NumPy infers it
arr = np.arange(24)

a = arr.reshape(4, -1)   # 4 rows, NumPy figures out cols (6)
b = arr.reshape(-1, 6)   # NumPy figures out rows (4)
c = arr.reshape(2, 3, -1) # (2, 3, 4)
print('reshape(4,-1) shape:', a.shape)
print('reshape(-1,6) shape:', b.shape)
print('reshape(2,3,-1) shape:', c.shape)
```

```python
# reshape returns a VIEW when possible
arr = np.arange(6)
view = arr.reshape(2, 3)
view[0, 0] = 999
print('After modifying view, original:', arr)  # arr[0] changed!

# flatten() always returns a copy
mat = np.array([[1, 2, 3], [4, 5, 6]])
flat_copy = mat.flatten()
flat_copy[0] = 999
print('After modifying flatten copy:', mat)  # mat unchanged

# ravel() returns a view when possible
flat_view = mat.ravel()
flat_view[0] = 999
print('After modifying ravel view:', mat)   # mat may change!
```

```python
# np.expand_dims — add a new axis (common in deep learning)
arr = np.array([1, 2, 3])  # shape (3,)
print('Original shape:', arr.shape)

# Add batch dimension at front -> (1, 3)
batch = np.expand_dims(arr, axis=0)
print('After expand_dims(axis=0):', batch.shape)

# Add channel dimension at end -> (3, 1)
channel = np.expand_dims(arr, axis=-1)
print('After expand_dims(axis=-1):', channel.shape)

# Equivalent with None / newaxis
print('arr[np.newaxis,:] shape:', arr[np.newaxis, :].shape)
print('arr[:,np.newaxis] shape:', arr[:, np.newaxis].shape)
```

```python
# np.squeeze — remove size-1 dimensions
arr = np.zeros((1, 3, 1, 4))
print('Before squeeze:', arr.shape)  # (1, 3, 1, 4)
print('After squeeze():', np.squeeze(arr).shape)        # (3, 4)
print('Squeeze axis=0:', np.squeeze(arr, axis=0).shape) # (3, 1, 4)
print('Squeeze axis=2:', np.squeeze(arr, axis=2).shape) # (1, 3, 4)
```

```python
# Practical: flatten image batch before FC layer
batch_images = np.random.randn(32, 3, 28, 28)  # (batch, C, H, W)
print('Image batch shape:', batch_images.shape)

# Flatten all but batch dimension
flat = batch_images.reshape(32, -1)
print('Flattened for FC layer:', flat.shape)  # (32, 2352)
```

```python
# EXCEPTION 1: Total size must remain same
arr = np.arange(6)
try:
    arr.reshape(4, 4)
except ValueError as e:
    print('ValueError:', e)

# EXCEPTION 2: squeeze on non-size-1 axis
arr = np.zeros((2, 3))
try:
    np.squeeze(arr, axis=1)  # axis=1 has size 3, not 1
except ValueError as e:
    print('ValueError:', e)
```
