# Concatenation, Stacking and Splitting

## Definition
These operations combine or divide arrays along specific axes.

### Combining Arrays
| Function | Description |
|----------|-------------|
| `np.concatenate(arrays, axis)` | Join along **existing** axis |
| `np.stack(arrays, axis)` | Join along **new** axis (same shapes required) |
| `np.vstack()` | Vertical stack (axis=0) |
| `np.hstack()` | Horizontal stack (axis=1) |
| `np.dstack()` | Depth stack (axis=2) |
| `np.r_[]` | Row-wise concatenation shorthand |
| `np.c_[]` | Column-wise concatenation shorthand |

### Splitting Arrays
| Function | Description |
|----------|-------------|
| `np.split(arr, indices, axis)` | Split at given indices |
| `np.vsplit()` | Split vertically |
| `np.hsplit()` | Split horizontally |
| `np.array_split()` | Like split but allows unequal chunks |

```python
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

# concatenate along existing axes
print('concat axis=0 (rows):\n', np.concatenate([a, b], axis=0))
print('concat axis=1 (cols):\n', np.concatenate([a, b], axis=1))
```

```python
# np.stack — creates a NEW axis
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

s0 = np.stack([a, b], axis=0)   # shape (2, 3)
s1 = np.stack([a, b], axis=1)   # shape (3, 2)

print('stack axis=0 shape:', s0.shape, '\n', s0)
print('stack axis=1 shape:', s1.shape, '\n', s1)
```

```python
# vstack, hstack, dstack
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print('vstack (vertical / row-wise):\n', np.vstack([a, b]))
print('hstack (horizontal / col-wise):\n', np.hstack([a, b]))

# dstack works on 2D by adding depth (3rd axis)
print('dstack shape:', np.dstack([a, b]).shape)  # (2, 2, 2)
```

```python
# Deep Learning: batching samples
sample1 = np.random.randn(3, 224, 224)  # single image C,H,W
sample2 = np.random.randn(3, 224, 224)
sample3 = np.random.randn(3, 224, 224)

# Stack to create batch
batch = np.stack([sample1, sample2, sample3], axis=0)
print('Batch shape:', batch.shape)  # (3, 3, 224, 224) = (batch, C, H, W)
```

```python
# np.split — divide array into equal parts
arr = np.arange(12)
parts = np.split(arr, 4)   # split into 4 equal parts
for i, p in enumerate(parts):
    print(f'Part {i}:', p)

print()
# Split at specific indices
parts2 = np.split(arr, [3, 7])   # split at index 3 and 7
for i, p in enumerate(parts2):
    print(f'Part {i}:', p)
```

```python
# vsplit and hsplit
mat = np.arange(1, 25).reshape(4, 6)
print('Matrix:\n', mat)

rows = np.vsplit(mat, 2)   # split into 2 equal row groups
print('\nvsplit[0]:\n', rows[0])
print('vsplit[1]:\n', rows[1])

cols = np.hsplit(mat, 3)   # split into 3 equal column groups
print('\nhsplit shapes:', [c.shape for c in cols])
```

```python
# array_split allows unequal chunks
arr = np.arange(10)
parts = np.array_split(arr, 3)  # can't be split equally
for i, p in enumerate(parts):
    print(f'Part {i}: {p}')
```

```python
# EXCEPTION 1: concatenate requires same shape on all but concat axis
a = np.ones((2, 3))
b = np.ones((2, 4))  # different number of cols
try:
    np.concatenate([a, b], axis=0)  # axis=0 -> rows must have same cols
except ValueError as e:
    print('ValueError:', e)

# EXCEPTION 2: stack requires exact same shape
a = np.ones((2, 3))
b = np.ones((2, 4))
try:
    np.stack([a, b], axis=0)
except ValueError as e:
    print('ValueError:', e)

# EXCEPTION 3: split with unequal division
arr = np.arange(10)
try:
    np.split(arr, 3)  # 10 is not divisible by 3
except ValueError as e:
    print('ValueError (use array_split instead):', e)
```
