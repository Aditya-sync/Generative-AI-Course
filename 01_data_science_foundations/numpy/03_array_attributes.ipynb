# Array Attributes — shape, ndim, size, dtype, itemsize

## Definition
Every NumPy array carries **metadata** (attributes) that describe its structure and memory layout.
These attributes are accessed as **properties** (no parentheses).

| Attribute | Type | Description |
|-----------|------|-------------|
| `.ndim` | int | Number of axes/dimensions |
| `.shape` | tuple | Size along each axis |
| `.size` | int | Total element count |
| `.dtype` | dtype | Data type of elements |
| `.itemsize` | int | Bytes per element |
| `.nbytes` | int | Total bytes (`size × itemsize`) |
| `.strides` | tuple | Bytes to step in each dimension |
| `.flags` | dict | Memory layout info |
| `.T` | ndarray | Transposed view |

```python
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]], dtype=np.float32)

print('ndim:    ', arr.ndim)      # 2
print('shape:   ', arr.shape)     # (2, 3)
print('size:    ', arr.size)      # 6
print('dtype:   ', arr.dtype)     # float32
print('itemsize:', arr.itemsize)  # 4 bytes (float32 = 32 bits / 8)
print('nbytes:  ', arr.nbytes)    # 24 bytes total
```

```python
# Strides — bytes to step per dimension
arr_64 = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int64)
print('strides (int64):', arr_64.strides)
# (24, 8) => moving to next row needs 24 bytes, next col needs 8 bytes

arr_32 = arr_64.astype(np.int32)
print('strides (int32):', arr_32.strides)
# (12, 4)
```

```python
# Shape of 1D, 2D, 3D, 4D arrays
a1 = np.ones(5)
a2 = np.ones((3, 4))
a3 = np.ones((2, 3, 4))
a4 = np.ones((10, 3, 224, 224))  # batch of images

for arr in [a1, a2, a3, a4]:
    print(f'ndim={arr.ndim}  shape={arr.shape}  size={arr.size}')
```

```python
# .flags — C vs F contiguous, writeable, etc.
c_arr = np.array([[1,2],[3,4]], order='C')
f_arr = np.asfortranarray(c_arr)

print('C array flags:\n', c_arr.flags)
print('F array flags:\n', f_arr.flags)
```

```python
# Scalar attributes
scalar_arr = np.float64(3.14)
print('scalar ndim:', scalar_arr.ndim)    # 0
print('scalar shape:', scalar_arr.shape)  # ()
print('scalar size:', scalar_arr.size)    # 1
```

```python
# EXCEPTION: shape is read-only if array is a view with incompatible strides
# You cannot directly set .shape that would require a copy
arr = np.array([[1,2,3],[4,5,6]])
try:
    arr.shape = (3, 3)  # wrong total size
except ValueError as e:
    print('ValueError:', e)

# Correct: total elements must remain same
arr.shape = (6,)  # OK! reshape to 1D
print('After reshape:', arr)
```
