# axis — Deep Understanding

## Definition
The `axis` parameter in NumPy specifies **which dimension** to operate along.
Getting this right is one of the most common sources of confusion in NumPy for beginners.

### Mental Model
> **axis=N** means "collapse (reduce) along dimension N"

For a 2D array of shape `(rows, cols)`:
- `axis=0` → operate **down the rows** → result has shape `(cols,)`
- `axis=1` → operate **across the columns** → result has shape `(rows,)`

For a 3D array of shape `(batch, rows, cols)`:
- `axis=0` → collapse batch → shape `(rows, cols)`
- `axis=1` → collapse rows → shape `(batch, cols)`
- `axis=2` → collapse cols → shape `(batch, rows)`

**Negative axes**: `axis=-1` = last axis, `axis=-2` = second to last

```python
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print('Array shape:', arr.shape)   # (2, 3)
print()
print('sum(axis=None):', np.sum(arr))          # 21 — all elements
print('sum(axis=0):', np.sum(arr, axis=0))     # [5 7 9] — down rows
print('sum(axis=1):', np.sum(arr, axis=1))     # [6 15] — across cols
print()
print('max(axis=0):', np.max(arr, axis=0))     # [4 5 6]
print('max(axis=1):', np.max(arr, axis=1))     # [3 6]
```

```python
# 3D axis understanding
arr3d = np.arange(24).reshape(2, 3, 4)  # (batch=2, rows=3, cols=4)
print('Shape:', arr3d.shape)
print()
print('sum(axis=0) — collapse batch:')
r0 = np.sum(arr3d, axis=0)
print('  shape:', r0.shape, '\n ', r0)  # (3, 4)

print('sum(axis=1) — collapse rows:')
r1 = np.sum(arr3d, axis=1)
print('  shape:', r1.shape, '\n ', r1)  # (2, 4)

print('sum(axis=2) — collapse cols:')
r2 = np.sum(arr3d, axis=2)
print('  shape:', r2.shape, '\n ', r2)  # (2, 3)
```

```python
# Negative axis indexing
arr = np.ones((2, 3, 4))
print('axis=-1 == axis=2?', 
      np.all(np.sum(arr, axis=-1) == np.sum(arr, axis=2)))

print('axis=-2 == axis=1?', 
      np.all(np.sum(arr, axis=-2) == np.sum(arr, axis=1)))
```

```python
# axis as tuple — reduce along multiple axes at once
arr = np.ones((2, 3, 4))
r = np.sum(arr, axis=(0, 2))  # collapse batch AND cols
print('sum(axis=(0,2)) shape:', r.shape)  # (3,)
```

```python
# keepdims — preserve reduced dimensions as size-1
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
s = np.sum(arr, axis=1, keepdims=True)
print('sum keepdims shape:', s.shape)  # (2, 1) not (2,)

# Now can broadcast back against original (2, 3)
normalized = arr / s
print('Row-normalized:\n', normalized)
```

```python
# Real example: batch normalization insight
# Layer norm: normalize across features (last axis)
batch = np.random.randn(4, 8)  # (batch_size=4, features=8)

mean = np.mean(batch, axis=-1, keepdims=True)   # (4, 1)
std  = np.std(batch, axis=-1, keepdims=True)    # (4, 1)
normalized = (batch - mean) / (std + 1e-8)

print('Batch shape:', batch.shape)
print('Mean per sample (should be ~0):', np.round(np.mean(normalized, axis=1), 6))
print('Std per sample  (should be ~1):', np.round(np.std(normalized,  axis=1), 6))
```

```python
# argmax / argmin along axis
scores = np.array([[0.1, 0.6, 0.3],
                   [0.7, 0.2, 0.1],
                   [0.2, 0.3, 0.5]])  # (batch=3, classes=3)

# Which class has highest score for each sample?
predictions = np.argmax(scores, axis=1)  # reduce along axis=1 (classes)
print('Predictions:', predictions)  # [1, 0, 2]

# Which sample has highest score for each class?
best_sample = np.argmax(scores, axis=0)  # reduce along axis=0 (samples)
print('Best sample per class:', best_sample)
```
