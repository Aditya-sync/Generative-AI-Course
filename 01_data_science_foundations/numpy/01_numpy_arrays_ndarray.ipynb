# NumPy Arrays (ndarray)

## What is ndarray?
The `ndarray` (N-dimensional array) is the **core data structure** of NumPy. It is a fixed-size, homogeneous multidimensional array — all elements must share the **same data type**.

Unlike Python lists:
- All elements share the same dtype → predictable, efficient memory
- Memory is stored **contiguously** → blazing fast access
- Supports **vectorized math** without Python loops
- Supports **broadcasting**, slicing, and linear algebra natively

In AI/ML, virtually all data (images, text embeddings, weight matrices) lives inside ndarrays.

| Property | Description |
|----------|-------------|
| `ndim` | Number of dimensions (axes) |
| `shape` | Tuple of size along each axis |
| `size` | Total number of elements |
| `dtype` | Data type of elements |
| `itemsize` | Bytes per single element |

```python
import numpy as np

# 1D array from a Python list
a = np.array([1, 2, 3, 4, 5])
print('1D array:', a)
print('Type:    ', type(a))
print('dtype:   ', a.dtype)
print('ndim:    ', a.ndim)
print('shape:   ', a.shape)
print('size:    ', a.size)
print('itemsize:', a.itemsize, 'bytes')
```

```python
# 2D array (matrix)
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print('2D array:\n', b)
print('ndim:', b.ndim)    # 2
print('shape:', b.shape)  # (2, 3) -> 2 rows, 3 cols
print('size:', b.size)    # 6
```

```python
# 3D array (e.g. batch of RGB images)
c = np.array([[[1, 2], [3, 4]],
              [[5, 6], [7, 8]]])
print('3D array:\n', c)
print('ndim:', c.ndim)    # 3
print('shape:', c.shape)  # (2, 2, 2)
```

```python
# Performance: NumPy vs Python list
import time
N = 1_000_000
py_list = list(range(N))
np_arr  = np.arange(N)

t0 = time.time()
_ = sum([x**2 for x in py_list])
print(f'Python list: {time.time()-t0:.4f}s')

t0 = time.time()
_ = np.sum(np_arr**2)
print(f'NumPy array: {time.time()-t0:.4f}s')
```

```python
# Homogeneous typing - mixed input gets UPCAST
mixed = np.array([1, 2.5, 3])      # all become float64
print('Mixed upcast:', mixed, mixed.dtype)

# Force a specific dtype
forced = np.array([1, 2, 3], dtype=np.float32)
print('Forced float32:', forced, forced.dtype)
```

```python
# Memory layout flags
arr_c = np.array([[1,2,3],[4,5,6]], order='C')  # Row-major (default)
arr_f = np.array([[1,2,3],[4,5,6]], order='F')  # Column-major
print('C-contiguous:', arr_c.flags['C_CONTIGUOUS'])
print('F-contiguous:', arr_f.flags['F_CONTIGUOUS'])
```

```python
# EXCEPTION 1: Inhomogeneous shape -> ValueError
try:
    bad = np.array([[1, 2, 3], [4, 5]])
except ValueError as e:
    print('ValueError:', e)
```

```python
# EXCEPTION 2: dtype overflow (silent wrap-around)
arr_int8 = np.array([100, 200], dtype=np.int8)  # 200 > 127 -> wraps!
print('int8 overflow:', arr_int8)  # 200 wraps to -56

# EXCEPTION 3: IndexError
arr = np.array([10, 20, 30])
try:
    print(arr[5])
except IndexError as e:
    print('IndexError:', e)
```
